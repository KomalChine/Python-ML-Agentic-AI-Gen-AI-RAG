# 13. Conditional Statements

Conditional statements allow a Python program to make decisions based on conditions.

Python commonly uses:

- `if`
- `if-else`
- `if-elif-else`
- nested `if`
- `match-case`

## 1. Indentation and Code Blocks

Python uses indentation to define which statements belong to a conditional block.

```python
if age >= 18:
    print("Adult")
    print("Eligible")
```

The indented statements are part of the `if` block.

Incorrect indentation causes an error:

```python
if age >= 18:
print("Adult")
```

This produces an `IndentationError`.

````python
age = 20

if age >= 18:
    print("You are an adult.")
````

## 2. `if` Statement

The `if` statement executes a block only when its condition evaluates to `True`.

### Syntax

```python
if condition:
    statement
```

````python
age = 20

if age >= 18:
    print("You are eligible to vote.")
````

````python
age = 15

if age >= 18:
    print("You are eligible to vote.")

print("Program continues...")
````

## 3. `if-else` Statement

Use `if-else` when there are two possible paths.

- If the condition is `True`, the `if` block runs.
- Otherwise, the `else` block runs.

````python
age = 15

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are too young to vote.")
````

## 4. `if-elif-else` Ladder

Use an `if-elif-else` ladder when there are multiple possible conditions.

Python checks the conditions **from top to bottom** and executes the first matching branch.

````python
marks = 75

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
else:
    print("Grade: D")
````

### Order of Conditions Matters

The first condition that evaluates to `True` is executed.

For example, this does **not** print `A`:

```python
marks = 95

if marks >= 60:
    print("C")
elif marks >= 90:
    print("A")
```

The first condition is already `True`, so Python does not check the later `elif`.

````python
marks = 95

if marks >= 60:
    print("C")
elif marks >= 90:
    print("A")
````

## 5. Nested `if` Statement

A nested `if` is an `if` statement inside another `if` statement.

It is useful when one decision depends on another condition.

````python
has_ticket = True
id_verified = False

if has_ticket:
    print("Ticket verified.")

    if id_verified:
        print("ID verified.")
````

## 6. `match-case` Statement

`match-case` was introduced in Python 3.10.

It can be used when you want to compare a value against multiple cases or patterns.

For basic use, think of it as another way to organize multiple possible choices.

````python
command = "start"

match command:
    case "start":
        print("System starting...")
    case "stop":
        print("System stopping...")
    case "restart":
        print("System restarting...")
    case _:
        print("Unknown command.")
````

## 7. Summary

| Structure | Purpose |
|---|---|
| `if` | One condition |
| `if-else` | Two possible paths |
| `if-elif-else` | Multiple possible paths |
| Nested `if` | A condition inside another condition |
| `match-case` | Compare a value against different cases/patterns |

### Key points

- Python uses indentation to define conditional blocks.
- `if` runs a block only when its condition is `True`.
- `if-else` provides two possible paths.
- `if-elif-else` handles multiple conditions.
- Python checks an `if-elif-else` ladder from top to bottom.
- A nested `if` places one conditional inside another.
- `match-case` provides another way to handle multiple cases.
