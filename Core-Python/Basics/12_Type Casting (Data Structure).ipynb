Type Casting (Data Structure)

```python
# List / Tuple / Set to List

tup = (1, 2, 3)
s = {4, 5, 6}

print(list(tup)) # Output: [1, 2, 3]
print(list(s))   # Output: [4, 5, 6] (order not guaranteed for sets)
```

```python
# List / Tuple / Set to Tuple

lst = [1, 2, 3]
s = {4, 5, 6}

print(tuple(lst)) # Output: (1, 2, 3)
print(tuple(s))   # Output: (4, 5, 6)
```

```python
# List / Tuple / Set to Set (Removes Duplicates)

lst = [1, 2, 2, 3, 3, 3]
tup = ("a", "b", "a")

print(set(lst)) # Output: {1, 2, 3}
print(set(tup)) # Output: {'a', 'b'}
```

```python
# Dictionary to List / Tuple / Set

d = {"a": 1, "b": 2}

# Default: Extracts keys
print(list(d))  # Output: ['a', 'b']
print(tuple(d)) # Output: ('a', 'b')
print(set(d))   # Output: {'a', 'b'}

# Extracting values explicitly
print(list(d.values())) # Output: [1, 2]

# Extracting key-value pairs as tuples
print(list(d.items()))  # Output: [('a', 1), ('b', 2)]
```

```python
# List / Tuple / Set of Pairs to Dictionary

# List of 2-item tuples
pairs_list = [("name", "Alice"), ("age", 25)]
print(dict(pairs_list)) 
# Output: {'name': 'Alice', 'age': 25}

# Set of 2-item tuples
pairs_set = {("x", 10), ("y", 20)}
print(dict(pairs_set))  
# Output: {'x': 10, 'y': 20}
```
