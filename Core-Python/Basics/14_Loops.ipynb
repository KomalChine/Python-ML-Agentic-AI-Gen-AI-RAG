Loops

range() function

```python
# Single Parameter: range(stop)
# Generates numbers starting from 0 up to stop - 1.

for i in range(4):
    print(i)

# Output:
# 0
# 1
# 2
# 3
```

```python
# Two Parameters: range(start, stop)
# Generates numbers starting from start up to stop - 1.

for i in range(2, 6):
    print(i)

# Output:
# 2
# 3
# 4
# 5
```

```python
# Three Parameters: range(start, stop, step)
# Generates numbers with a specific step value (increment).

for i in range(1, 10, 2):
    print(i)

# Output:
# 1
# 3
# 5
# 7
# 9
```

```python
# Negative Step (Counting Down)
# Pass a negative step to count backwards.

for i in range(5, 0, -1):
    print(i)

# Output:
# 5
# 4
# 3
# 2
# 1
```

```python
nums = list(range(1, 6))
print(nums)  
# Output: [1, 2, 3, 4, 5]
```

for Loop

```python
# Iterating Over a Range (range(start, stop, step)):

for i in range(1, 6):  # Stops before 6
    print(i)
# Output: 1 2 3 4 5
```

```python
# Iterating Over Collections:

fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# Using enumerate() for Index & Value:

names = ["Alice", "Bob"]
for index, name in enumerate(names):
    print(f"{index}: {name}")
# Output:
# 0: Alice
# 1: Bob
```

while Loop

```python
count = 1

while count <= 3:
    print(f"Count is: {count}")
    count += 1  # Increment to prevent an infinite loop!

# Output:
# Count is: 1
# Count is: 2
# Count is: 3
```

Loop Control Statements

```python
# 'break' example
for num in range(1, 10):
    if num == 5:
        break  # Exit loop when num reaches 5
    print(num)  # Prints 1, 2, 3, 4

# 'continue' example
for num in range(1, 6):
    if num == 3:
        continue  # Skip printing 3
    print(num)  # Prints 1, 2, 4, 5
```

Loop else Clause

```python
numbers = [1, 3, 5, 7]

for num in numbers:
    if num % 2 == 0:
        print("Found an even number!")
        break
else:
    print("No even numbers found.")

# Output: No even numbers found.
```

Comprehension

List Comprehension

```python
# Traditional loop:
# squares = []
# for x in range(5): squares.append(x**2)

squares = [x**2 for x in range(5)]
print(squares)  # Output: [0, 1, 4, 9, 16]
```

```python
# With Filtering (if):

# Only include even numbers
evens = [x for x in range(10) if x % 2 == 0]
print(evens)  # Output: [0, 2, 4, 6, 8]
```

```python
# With Conditional Assignment (if-else):

# Apply label based on value
labels = ["Even" if x % 2 == 0 else "Odd" for x in range(4)]
print(labels)  # Output: ['Even', 'Odd', 'Even', 'Odd']
```

Dictionary Comprehension

```python
# Map numbers to their squares
square_map = {x: x**2 for x in range(1, 4)}
print(square_map)  # Output: {1: 1, 2: 4, 3: 9}

# Filter a dictionary
prices = {"apple": 1.5, "banana": 0.8, "cherry": 2.5}
expensive = {k: v for k, v in prices.items() if v > 1.0}
print(expensive)   # Output: {'apple': 1.5, 'cherry': 2.5}
```

Set Comprehension

```python
word = "abracadabra"
unique_vowels = {char for char in word if char in "aeiou"}
print(unique_vowels)  # Output: {'a'}
```
