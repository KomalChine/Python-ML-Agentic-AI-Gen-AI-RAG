Constructor

```python
class Car:
    def __init__(self, brand, model, color):
        # Assigning incoming parameter values to instance attributes
        self.brand = brand
        self.model = model
        self.color = color

# Creating an object automatically triggers __init__
my_car = Car("Ford", "Mustang", "Black")

print(my_car.brand)  # Output: Ford
print(my_car.color)  # Output: Black
```

```python
# Just like standard functions, you can give parameters default values inside __init__.
class Smartphone:
    def __init__(self, brand, storage=128):
        self.brand = brand
        self.storage = storage  # Defaults to 128 GB if omitted

phone1 = Smartphone("Apple", 256)
phone2 = Smartphone("Samsung")  # Uses default storage

print(f"{phone1.brand}: {phone1.storage}GB")  # Output: Apple: 256GB
print(f"{phone2.brand}: {phone2.storage}GB")  # Output: Samsung: 128GB
```

self keyword

```python
class Car:
    def __init__(self, brand):
        self.brand = brand  # Binds 'brand' to THIS specific car instance

    def show_brand(self):
        print(f"This car is a {self.brand}")

car1 = Car("Tesla")
car2 = Car("BMW")

car1.show_brand()  # Output: This car is a Tesla
car2.show_brand()  # Output: This car is a BMW
```

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def add_and_double(self, a, b):
        # Calls the 'add' method of the current object
        result = self.add(a, b)
        return result * 2

calc = Calculator()
print(calc.add_and_double(3, 4))  # Output: 14
```
