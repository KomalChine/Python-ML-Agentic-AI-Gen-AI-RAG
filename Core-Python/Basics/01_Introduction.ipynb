## 1. What is Python?

Python is a high-level, interpreted, general-purpose programming language. It is known for its simple, readable syntax and is widely used in web development, data science, automation, artificial intelligence, machine learning, scientific computing, and software development.

Python programs are usually easy to read and write because the language syntax is designed to be close to natural language.

## 2. History / Background

- Python was created by **Guido van Rossum**.
- Development began in the late 1980s, and Python was first released publicly in **1991**.
- The name "Python" was inspired by the British comedy group **Monty Python**.
- Python 2 and Python 3 were two major generations of the language. **Python 3** is the modern version used for current development.
- Python is developed as an open-source project with a large global community.

## 3. Features of Python

Important features of Python include:

1. **Simple and readable** – Python code is generally easy to understand.
2. **High-level language** – It handles many low-level programming details for the programmer.
3. **Interpreted** – Python code can be executed by a Python interpreter without a separate traditional compilation step.
4. **Dynamically typed** – Variable types are determined at runtime.
5. **Object-oriented** – Python supports classes and objects.
6. **Cross-platform** – Python programs can run on Windows, macOS, Linux, and other platforms.
7. **Large standard library** – Python provides many built-in modules and tools.
8. **Extensive ecosystem** – Thousands of third-party packages are available.
9. **Open source** – Python can be freely used, modified, and distributed under its license.

## 4. Applications of Python

Python is used in many areas:

- **Web development** – Django, Flask, FastAPI
- **Data analysis** – pandas, NumPy
- **Data visualization** – Matplotlib, Seaborn
- **Artificial intelligence and machine learning** – scikit-learn, TensorFlow, PyTorch
- **Automation and scripting** – file processing, repetitive tasks, system utilities
- **Scientific computing** – numerical and scientific applications
- **Desktop applications** – GUI applications using libraries such as Tkinter
- **Testing** – automated software testing
- **Education** – teaching programming and computer science concepts
- **Cybersecurity and networking** – automation, analysis, and network programming

## 5. Why Learn Python?

Python is a popular first programming language because it provides a relatively gentle learning curve while remaining useful for advanced applications.

Reasons to learn Python:

- Easy-to-read syntax
- Useful for beginners and professionals
- Large community and extensive documentation
- Many libraries and frameworks
- Widely used in industry

- Strong presence in data science, AI, and machine learning

## 6. Python Interpreter

A **Python interpreter** is the program that reads and executes Python code. It processes Python instructions and produces the corresponding results.

For example, when you execute:

```python
print("Hello, Python!")
```

the Python interpreter executes the `print()` instruction and displays the text.

## 7. Python Script

A **Python script** is a file containing Python code, normally saved with the `.py` extension.

Example file: `hello.py`

```python
print("Hello, World!")
```

A script can contain multiple statements and can be executed whenever the program is needed.

## 8. Interactive Mode

**Interactive mode** allows you to enter Python statements one at a time and immediately see their results.

Example:

```text
>>> 10 + 5
15
>>> print("Hello")
Hello
```

Interactive mode is useful for learning, experimenting, testing small pieces of code, and checking how Python expressions behave.

> **Note:** A Jupyter Notebook also provides an interactive environment, where code can be executed cell by cell.

## 9. Comments

Comments are notes written in source code for humans. Python ignores comments when executing the program.

### Single-line comment

Use `#` to create a comment:

```python
# This is a comment
print("Hello")
```

Comments can also appear after code:

```python
x = 10  # Store 10 in x
```

Good comments explain **why** something is done when the reason is not obvious from the code itself.

## 10. Basic Syntax

Python has a relatively simple syntax. Some important rules are:

- Python uses **indentation** to define blocks of code.
- Statements normally do not require a semicolon at the end.
- Python is **case-sensitive**: `name` and `Name` are different identifiers.
- Strings can be written using single or double quotes.
- Variables can be created by assigning values with `=`.

Example:

```python
name = "Komal"
age = 25

if age >= 18:
    print(name, "is an adult")
```

Notice that the `print()` statement is indented inside the `if` block.

## 11. Hello World

The traditional first program in a programming language is the **Hello World** program.

In Python, it is very simple:

````python
print("Hello, World!")
````

### Expected Output

```text
Hello, World!
```
