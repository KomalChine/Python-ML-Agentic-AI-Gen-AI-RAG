# 15. Comprehensions

Comprehensions provide a concise way to create a new collection from an iterable.

Python provides three basic collection comprehensions:

- List comprehension
- Dictionary comprehension
- Set comprehension

Comprehensions are closely related to loops, but they are best understood as a concise way to express common collection-building patterns.

## 1. What is a Comprehension?

Consider the normal loop:

```python
squares = []

for x in range(1, 6):
    squares.append(x ** 2)

print(squares)
```

The same operation can be written as a list comprehension:

````python
squares = [x ** 2 for x in range(1, 6)]

print(squares)
````

### Why use comprehensions?

A comprehension can make a simple transformation concise and readable.

Use a normal loop when the logic becomes complex or when multiple statements are needed.

## 2. List Comprehension

A list comprehension creates a new list from an iterable.

### Basic syntax

```python
[expression for item in iterable]
```

The expression produces each value in the new list.

````python
numbers = [1, 2, 3, 4]

doubled = [x * 2 for x in numbers]

print(doubled)
````

### Using `range()`

````python
numbers = [x for x in range(1, 6)]
squares = [x ** 2 for x in range(1, 6)]
cubes = [x ** 3 for x in range(1, 6)]

print(numbers)
print(squares)
print(cubes)
````

### Using an Existing Iterable

Comprehensions are not limited to `range()`. They can work with lists, strings, tuples, and other iterables.

````python
numbers = [1, 2, 3, 4]

squares = [x ** 2 for x in numbers]

print(squares)
````

````python
word = "Python"

letters = [char for char in word]

print(letters)
````

### List Comprehension with `if`

Use an `if` condition to filter which items are included.

Syntax:

```python
[expression for item in iterable if condition]
```

````python
numbers = range(1, 11)

evens = [x for x in numbers if x % 2 == 0]

print(evens)
````

````python
words = ["apple", "banana", "cat", "dog"]

long_words = [word for word in words if len(word) > 4]

print(long_words)
````

### List Comprehension with `if-else`

A conditional expression can transform each item differently.

Syntax:

```python
[expression_if_true if condition else expression_if_false for item in iterable]
```

Compare this with filtering:

```python
[x for x in numbers if x % 2 == 0]
```

Here, the `if` comes **after** the `for` because it filters items.

With `if-else`:

```python
["Even" if x % 2 == 0 else "Odd" for x in numbers]
```

the `if-else` comes **before** the `for` because it chooses the resulting value.

````python
numbers = range(5)

labels = ["Even" if x % 2 == 0 else "Odd" for x in numbers]

print(labels)
````

### Nested List Comprehension

A simple nested comprehension can replace a nested loop.

Example:

````python
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

flattened = [num for row in matrix for num in row]

print(flattened)
````

The equivalent normal loop is:

```python
flattened = []

for row in matrix:
    for num in row:
        flattened.append(num)
```

Keep nested comprehensions simple. If the logic becomes difficult to read, use normal loops.

## 3. Dictionary Comprehension

A dictionary comprehension creates a new dictionary.

### Basic syntax

```python
{key_expression: value_expression for item in iterable}
```

````python
numbers = range(1, 6)

squares = {x: x ** 2 for x in numbers}

print(squares)
````

### Dictionary Comprehension from an Existing Dictionary

Dictionary comprehensions work especially well with `.items()`.

````python
prices = {
    "apple": 100,
    "banana": 40,
    "mango": 120
}

discounted = {
    fruit: price * 0.9
    for fruit, price in prices.items()
}

print(discounted)
````

### Dictionary Comprehension with `if`

Use an `if` condition to select which key-value pairs are included.

````python
scores = {
    "Alice": 85,
    "Bob": 45,
    "Charlie": 72
}

passed = {
    name: score
    for name, score in scores.items()
    if score >= 50
}

print(passed)
````

## 4. Set Comprehension

A set comprehension creates a new set.

### Basic syntax

```python
{expression for item in iterable}
```

Because the result is a set, duplicate values are automatically removed.

````python
numbers = [1, 2, 2, 3, 3, 4]

squares = {x ** 2 for x in numbers}

print(squares)
````

### Set Comprehension with `if`

````python
word = "programming"

vowels = {
    char
    for char in word
    if char in "aeiou"
}

print(vowels)
````

## 5. Comprehension vs Normal Loop

### Normal loop

````python
squares = []

for x in range(1, 6):
    squares.append(x ** 2)

print(squares)
````

### Comprehension

````python
squares = [x ** 2 for x in range(1, 6)]

print(squares)
````

### Choosing between them

**Comprehension**
- concise
- readable for simple transformations and filtering
- useful for creating collections

**Normal loop**
- easier for complex logic
- useful when multiple statements are needed
- often easier to debug when the logic is complicated

If a comprehension becomes difficult to read, use a normal loop instead.

## 6. When to Use Comprehensions

Use a comprehension when it makes a simple operation clearer and more concise.

For example:

````python
numbers = range(1, 11)

even_squares = [x ** 2 for x in numbers if x % 2 == 0]

print(even_squares)
````

Avoid forcing complicated logic into a single line.

Instead of a difficult one-liner, a normal loop may be clearer:

```python
result = []

for item in data:
    if condition(item):
        # additional logic
        ...
```

The goal is **clear Python**, not the shortest possible code.

## 7. Summary

| Type | Syntax pattern | Common use |
|---|---|---|
| List comprehension | `[expression for item in iterable]` | Create a list |
| List with filter | `[expression for item in iterable if condition]` | Filter and transform |
| List with `if-else` | `[a if condition else b for item in iterable]` | Choose a value |
| Dictionary comprehension | `{key: value for item in iterable}` | Create a dictionary |
| Set comprehension | `{expression for item in iterable}` | Create a set and remove duplicates |

### Basic progression

```text
Normal loop
    ↓
List comprehension
    ↓
List comprehension with if
    ↓
List comprehension with if-else
    ↓
Dictionary comprehension
    ↓
Set comprehension
```
