```python
a = 20
b = 20
#id function used to know the memory address
#print function is used to print statement
print(id(a)) # both address of a and b will be same cause they are refering ton same object that is 20
print(id(b))
```

Using Variable in Expression

```python
# = assign the value to the left side
x = 10
y = 3.5
result = x + y
print(result) # output: 13.5
```

Changing the Value of a Variable

```python
# Initial value
temp = 30
print(temp) # output: 30
# Changing the value of 'temp'
temp = 36
print(temp) # output: 36
```
