Operators

Arithmetic Operators

```python
a = 24
b = 12
x = 4.5
y = "Hi"
z = "Good Morning"
```

```python
print("a + b = ",a + b) # int output
print("a + x = ",a + x) # float output
print("y + z = ",y + z)
print("y + ' ' + z = ",y+" "+z)
print("y + a = ",y + a) # Error cause int or float cannot be join with strings 
```

```python
print("a - b = ",a - b) # int output
print("a - x = ",a - x) # float output
```

```python
print("a * b = ",a * b) # int output
print("a * x = ",a * x) # float output
```

```python
print("a / b = ",a / b) # always output will be in float
print("a / x = ",a / x) 
```

```python
print("a // b = ",a // b) # output will always will be the low and integer means floor
print("a // x = ",a // x) 
print("10 // -3 = ",10 // -3) # notice here output is -4 not -3 cause floor is -4
```

```python
print("a % b = ",a % b) # int output
print("a % x = ",a % x) # float output
```

```python
print("a ** b = ",a ** b) # int output
print("a ** x = ",a ** x) # float output
```

Bitwise Operator

```python
"""  101   (5)
   & 011   (3)
   -----
     001   (1)"""
5 & 3
```

```python
'''  101
   | 011
   -----
     111  '''
5 | 3
```

```python
'''  101
   ^ 011
   -----
     110  '''
5 ^ 3
```

```python
~5 # ~n = -(n + 1)
```

```python
~10
# 10 = 1100
# first convert it in 8 bits
# 10 = 00001100
# ~10 = 11110011
# the first bit is 1 means  it is negative number, so to find decimal number use two's complement
# Flip the bits again: 00001100 and then add 1
# 00001100 + 1 = 00001101 = 11 (Remeber it is binary addition not decimal addition, rules are different)
# Since it was negative so -11
```

```python
15<<2
```

```python
15>>2
```

Operator Precedence

```python
result = 10 + 2 * 3 ** 2 > 25 and not (8 // 3 == 2 or 5 % 2 != 1)
result
# (8 // 3 == 2 or 5 % 2 != 1)
# (2 == 2 or 5 % 2 != 1)
# (2 == 2 or 1 != 1)
# (True or 1 != 1)
# (True or False)
# True
# 10 + 2 * 3 ** 2 > 25 and not (True)
# 10 + 2 * 9 > 25 and not (True)
# 10 + 18 > 25 and not (True)
# 28 > 25 and not (True)
# True and not (True)
# True and False
# False
```
