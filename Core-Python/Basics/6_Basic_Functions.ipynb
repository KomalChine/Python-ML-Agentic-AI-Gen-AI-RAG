Basic Functions

print function

```python
print("Hello")
```

```python
print("Hello", "Python") # More than one object
```

```python
print("Hello", "Python", sep="-") 
# Here separator separate each object using specified string or object
```

```python
print("Hello", end=" ") 
# Here end=" " so at the end of string there will be blank space and next print statemnt wiil be displayed on same line
print("Python")
```

```python
print("Hello",end="\t")
# Here end="\t" means tab so after 4 spaces next print statement will be displayed
print("Python")
```

```python
print("Hello",end="\n")
# Here end="\n" means next line so next print statement will be displayed on next line
print("Python")
```

input function

```python
age = input("Enter age: ")
print(type(age))
```

```python
# Can convert string into other datatype by:
age = int(input("Enter age: "))
print(type(age))
```

```python
flt = float(input("Enter float: "))
print(type(flt))
```

```python
# Also we can perform indexing and sclicing
text = input("Enter a text: ")[2]
# If we give "Hello" as input "l" will be output
print(text)
```

```python
aadhar_no = input("Enter a aadhar number: ")[8:]
# last 4 digit will be printed
print("xxxxxxxx" + aadhar_no)
```
