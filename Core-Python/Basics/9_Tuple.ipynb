Tuple

```python
numbers = (10, 20, 30, 40)
print(numbers)
print(type(numbers))
```

```python
mixed_tuple = (1, "Python", 3.14, True)
mixed_tuple
```

```python
empty_tuple = ()
type(empty_tuple)
```

```python
# If you want to create a tuple with only one item, you must include a trailing comma. Otherwise, Python will treat it as a regular data type (like a string or integer)
single_item = ("apple",)  # Tuple
not_a_tuple = ("apple")   # String!
print(type(single_item))
print(type(not_a_tuple))
```

```python
# Indexing and slicing work identically to lists
colors = ("red", "green", "blue", "yellow")

print(colors[0])     # Output: 'red'
print(colors[1:3])   # Output: ('green', 'blue')

# Negative Indexing and Backward Slicing
print()
print("Negative Indexing and Backward Slicing")
print(colors[-2])    # Output: "blue"
print(colors[::-2])  # Output: ('yellow', 'green')
```

Tuple Unpacking

```python
coordinates = (12.97, 77.59)
latitude, longitude = coordinates

print(latitude)   # Output: 12.97
print(longitude)  # Output: 77.59
```

```python
# Can't modify a tuple, Python will raise a TypeError
numbers = (1, 2, 3)
numbers[0] = 99  # ❌ TypeError: 'tuple' object does not support item assignment
```

Built-in Methods

```python
letters = ('a', 'b', 'c', 'a', 'a')

# 1. count() - Returns how many times a value appears
print(letters.count('a'))  # Output: 3

# 2. index() - Returns the first index of a given value
print(letters.index('b'))  # Output: 1
```
