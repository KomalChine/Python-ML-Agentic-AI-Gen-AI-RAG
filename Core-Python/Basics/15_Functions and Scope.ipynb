Functions

```python
def greet(name):
    """Docstring: Returns a greeting message."""
    return f"Hello, {name}!"

# Function call
message = greet("Alice")
print(message)  # Output: Hello, Alice!
```

```python
def create_profile(name, age):
    print(f"Name: {name}, Age: {age}")

create_profile("Bob", 30)          # Positional
create_profile(age=30, name="Bob") # Keyword
```

```python
def power(base, exponent=2):
    return base ** exponent

print(power(5))     # Output: 25 (uses default exponent=2)
print(power(5, 3))  # Output: 125
```

```python
def summarize(*args, **kwargs):
    print("Positional args:", args)
    print("Keyword args:", kwargs)

summarize(10, 20, 30, status="Active", role="Admin")
# Output:
# Positional args: (10, 20, 30)
# Keyword args: {'status': 'Active', 'role': 'Admin'}
```

Variable Scope (LEGB Rule)

```python
x = "Global"

def outer():
    x = "Enclosing"
    
    def inner():
        x = "Local"
        print(x)  # Looked up in L -> Prints 'Local'
        
    inner()

outer()
```

```python
count = 0

def increment():
    global count  # Allows modification of the global 'count' variable
    count += 1

increment()
print(count)  # Output: 1
```

```python
def outer():
    num = 10
    def inner():
        nonlocal num  # Modifies 'num' in the enclosing outer() function
        num = 20
    inner()
    print(num)  # Output: 20

outer()
```

```python
# Regular function vs Lambda
square = lambda x: x ** 2
print(square(4))  # Output: 16

# Sorting a list of tuples by second element using lambda
pairs = [(1, "one"), (3, "three"), (2, "two")]
pairs.sort(key=lambda pair: pair[1])
print(pairs)  # Output: [(1, 'one'), (3, 'three'), (2, 'two')]
```

```python
words = ["python", "cat", "lambda", "dog", "function"]

# Filter out words with length <= 4
long_words = list(filter(lambda word: len(word) > 4, words)) # Here lamda acts as a condition

print(long_words)  
# Output: ['python', 'lambda', 'function']
```
