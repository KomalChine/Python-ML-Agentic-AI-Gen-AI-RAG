This notebook introduces Python's **List** data structure and the most important operations and methods used with lists.

## 1. Introduction to Data Structures

A **data structure** is a way of organizing and storing data so that it can be accessed and manipulated efficiently.

Common Python data structures include:

- List
- Tuple
- Set
- Dictionary

This notebook focuses specifically on the **List** data structure.


### Data Type vs Data Structure

A **data type** describes the kind of value an object represents.

Examples:

```text
int
float
str
bool
complex
```

A **data structure** organizes multiple values or objects in a particular way.

Examples:

```text
list
tuple
set
dictionary
```

The other data structures will be covered in their own lessons.


## 2. Introduction to Lists

A **list** is an ordered, mutable collection of objects.

### Main characteristics of a List

| Characteristic | Meaning |
|---|---|
| Ordered | Items retain their sequence |
| Mutable | Items can be changed after creation |
| Indexed | Items can be accessed using indexes |
| Heterogeneous | Different data types can coexist |
| Duplicates allowed | The same value can appear multiple times |


### Ordered

Lists preserve the order in which elements are stored.


````python
numbers = [30, 10, 20, 40]

print(numbers)
````

### Mutable

Lists are mutable, meaning their elements can be changed after the list is created.

This is demonstrated in detail in the **Modifying List Elements** section.


### Indexed

Each list element has a position called an **index**.

Indexes start at `0`.


### Heterogeneous

A list can contain values of different data types.


````python
items = [10, "Hello", 3.14, True, 2 + 8j]

print(items)
````

### Duplicate Values

Lists allow duplicate values.


````python
numbers = [10, 20, 10, 30, 10]

print(numbers)
````

## 3. Creating Lists

### Empty List

Use square brackets to create an empty list.


````python
numbers = []

print(numbers)
print(type(numbers))
````

> Avoid using `list` as a variable name because `list` is already the name of Python's built-in list type.


### List with Values


````python
numbers = [10, 20, 30, 40, 50]

print(numbers)
````

### Mixed Data Types


````python
items = [10, "Python", 3.14, True, None]

print(items)
````

### Nested Lists

A list can contain another list as an element.


````python
nested_list = [1, 2, [3, 4, 5]]

print(nested_list)
````

### Quick Inspection with `type()` and `len()`

`type()` and `len()` were introduced earlier. Here we use them briefly with lists.


````python
numbers = [10, 20, 30]

print(type(numbers))
print(len(numbers))
````

## 4. List Indexing

List elements can be accessed using their index.

```text
Positive indexing:
0    1    2    3    4
10  20   30   40   50

Negative indexing:
-5  -4  -3  -2  -1
10  20  30  40  50
```


### Positive Indexing


````python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])
print(numbers[2])
print(numbers[4])
````

### Negative Indexing


````python
numbers = [10, 20, 30, 40, 50]

print(numbers[-1])
print(numbers[-2])
print(numbers[-5])
````

### Nested List Indexing


````python
nested_list = [1, 2, [3, 4, 5]]

print(nested_list[2])
print(nested_list[2][0])
print(nested_list[2][2])
````

## 5. List Slicing

List slicing extracts a portion of a list using:

```python
list[start:stop:step]
```

- `start` → starting index, included
- `stop` → ending index, excluded
- `step` → distance between selected indexes


### Basic Slicing


````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
````

### Omitting `start`


````python
numbers = [10, 20, 30, 40, 50]

print(numbers[:3])
````

### Omitting `stop`


````python
numbers = [10, 20, 30, 40, 50]

print(numbers[2:])
````

### Using `start`, `stop`, and `step`


````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:6:2])
````

### Negative Slicing


````python
numbers = [10, 20, 30, 40, 50]

print(numbers[-4:-1])
print(numbers[-1:-5:-1])
````

### Reversing with Slicing

A step of `-1` selects elements from right to left.


````python
numbers = [10, 20, 30, 40, 50]

print(numbers[::-1])
````

### Nested List Slicing


````python
nested_list = [1, 2, [3, 4, 5], 6, 7]

print(nested_list[1:4])
print(nested_list[2][1:])
````

### Slicing Creates a New List

List slicing returns a new list and does not modify the original list.


````python
numbers = [1, 2, 3, 4, 5]

part = numbers[1:4]

print("Sliced list:", part)
print("Original list:", numbers)
````

## 6. Modifying List Elements

Lists are mutable, so individual elements can be changed after creation.


````python
numbers = [10, 20, 30, 40, 50]

print("Before:", numbers)

numbers[3] = 7.1

print("After:", numbers)
````

You can also modify an element inside a nested list.


````python
nested_list = [1, 2, [3, 4, 5]]

nested_list[2][1] = 40

print(nested_list)
````

## 7. List Membership

The membership operators `in` and `not in` can be used to test whether a value occurs in a list.


````python
numbers = [10, 20, 30, 40]

print(20 in numbers)
print(50 in numbers)

print(20 not in numbers)
print(50 not in numbers)
````

## 8. List Concatenation & Repetition

### Repetition with `*`

### 9.1 Adding Elements

#### `append()`

`append()` adds **one item** to the end of a list.

It modifies the original list and returns `None`.


````python
numbers = [10, 20, 30]

result = numbers.append(40)

print(numbers)
print("Return value:", result)
````

`append()` accepts exactly one argument.


````python
numbers = [1, 2]

numbers.append(3)
print(numbers)

numbers.append(3)
print(numbers)
````

### Appending a List

If the object being appended is itself a list, the entire list becomes one element.


````python
items = [1, 2]

items.append([3, 4])

print(items)
````

This is different from `extend()`, which adds the elements individually.


#### `extend()`

`extend()` adds each element from an iterable to the end of the list.

It modifies the original list and returns `None`.


````python
numbers = [1, 2]

result = numbers.extend([3, 4, 5])

print(numbers)
print("Return value:", result)
````

### `append()` vs `extend()`


````python
a = [1, 2]
a.append([3, 4])

print("append():", a)

b = [1, 2]
b.extend([3, 4])

print("extend():", b)
````

Because `extend()` accepts an iterable, it can also extend a list with the characters of a string.


````python
words = ["hi"]

words.extend("bye")

print(words)
````

For lists, `+=` performs an in-place extension using the values from the right-hand iterable.


````python
numbers = [1, 2]

numbers += [3, 4]

print(numbers)
````

#### `insert()`

`insert(index, value)` inserts a value at a specified position.


````python
numbers = [10, 20, 30]

numbers.insert(1, 15)

print(numbers)
````

Elements at and after the insertion position are shifted to the right.


````python
numbers = [10, 20, 30, 40]

numbers.insert(2, 25)

print(numbers)
````

`insert()` returns `None`.


````python
numbers = [1, 2, 3]

result = numbers.insert(1, 10)

print(numbers)
print("Return value:", result)
````

A very large positive index inserts at the end.


````python
numbers = [1, 2, 3]

numbers.insert(100, 4)

print(numbers)
````

A negative index can also be used.


````python
numbers = [1, 2, 3, 4]

numbers.insert(-1, 99)

print(numbers)
````

> **Note:** Inserting into the middle of a very large list can require shifting existing elements. This can make `insert()` slower than appending to the end.


### 9.2 Removing Elements

#### `clear()`

`clear()` removes all elements from the list.

It modifies the original list and returns `None`.


````python
numbers = [10, 20, 30]

result = numbers.clear()

print(numbers)
print("Return value:", result)
````

`clear()` is different from assigning a new empty list.

With:

```python
numbers.clear()
```

the existing list object is emptied.

With:

```python
numbers = []
```

the variable is rebound to a new list object.


````python
numbers = [1, 2, 3]
same_reference = numbers

numbers.clear()

print("numbers:", numbers)
print("same_reference:", same_reference)
````

#### `pop()`

`pop()` removes and **returns** an item.

Without an argument, it removes the last item.


````python
numbers = [10, 20, 30, 40]

removed = numbers.pop()

print("List:", numbers)
print("Removed:", removed)
````

An index can be supplied to remove a specific item.


````python
numbers = [10, 20, 30, 40]

removed = numbers.pop(1)

print("List:", numbers)
print("Removed:", removed)
````

#### `remove()`

`remove(value)` removes the **first occurrence** of a specified value.

It modifies the list and returns `None`.


````python
numbers = [10, 20, 10, 30]

result = numbers.remove(10)

print(numbers)
print("Return value:", result)
````

If the value occurs multiple times, only the first matching occurrence is removed.


````python
numbers = [10, 20, 10, 30, 10]

numbers.remove(10)

print(numbers)
````

If the value is not found, `remove()` raises a `ValueError`.

```python
numbers.remove(100)
```


````python
numbers = [10, 20, 30]

try:
    numbers.remove(100)
except ValueError as error:
    print("ValueError:", error)
````

### `pop()` vs `remove()`

| Method | Removes by | Returns |
|---|---|---|
| `pop()` | Index | Removed item |
| `remove()` | Value | `None` |



### 9.3 Searching & Counting

#### `count()`

`count(value)` returns the number of times a value occurs in the list.


````python
numbers = [10, 20, 10, 30, 10]

print(numbers.count(10))
print(numbers.count(50))
````

#### `index()`

`index(value)` returns the index of the first occurrence of a value.


````python
fruits = ["apple", "banana", "orange", "banana"]

print(fruits.index("banana"))
````

`index()` can also accept optional `start` and `end` positions:

```python
list.index(value, start, end)
```


````python
numbers = [10, 20, 30, 20, 40, 20]

print(numbers.index(20))
print(numbers.index(20, 2))
print(numbers.index(20, 3, 6))
````

If the value is not found, `index()` raises a `ValueError`.


````python
numbers = [10, 20, 30]

try:
    print(numbers.index(100))
except ValueError as error:
    print("ValueError:", error)
````

### 9.4 Copying

#### `copy()`

`copy()` creates a **shallow copy** of a list.

The new outer list is a different object from the original.


````python
original = [1, 2, 3]
copied = original.copy()

print(original)
print(copied)

print("Same values:", original == copied)
print("Same object:", original is copied)
````

### Shallow Copy and Nested Lists

A shallow copy creates a new outer list, but nested mutable objects are still shared.


````python
original = [[1, 2], [3, 4]]
copied = original.copy()

copied[0][0] = 99

print("Original:", original)
print("Copied:", copied)
````

This notebook only introduces the idea of a shallow copy. Deep copying is a separate topic and is not required for basic list operations.


### 9.5 Reordering

#### `reverse()`

`reverse()` reverses the order of elements **in place**.

It modifies the original list and returns `None`.


````python
numbers = [10, 20, 30, 40]

result = numbers.reverse()

print(numbers)
print("Return value:", result)
````

#### `sort()`

`sort()` sorts a list **in place**.

By default, values are sorted in ascending order.


````python
numbers = [40, 10, 30, 20]

numbers.sort()

print(numbers)
````

### Descending Order

Use `reverse=True` to sort in descending order.


````python
numbers = [40, 10, 30, 20]

numbers.sort(reverse=True)

print(numbers)
````

### Sorting with `key`

The `key` parameter lets you specify the value used for sorting.

For example, strings can be sorted according to their length.


````python
words = ["Python", "is", "interesting"]

words.sort(key=len)

print(words)
````

### Mixed Incomparable Types

Sorting values that cannot be meaningfully compared can raise a `TypeError`.

For example, integers and strings cannot be directly ordered against one another.


````python
items = [10, "Python", 3.14]

try:
    items.sort()
except TypeError as error:
    print("TypeError:", error)
````

## 10. List Method Quick Reference

| Method | Purpose | Returns |
|---|---|---|
| `append(x)` | Add one item to the end | `None` |
| `extend(iterable)` | Add items from an iterable | `None` |
| `insert(i, x)` | Insert an item at an index | `None` |
| `clear()` | Remove all items | `None` |
| `pop()` / `pop(i)` | Remove and return an item | Removed item |
| `remove(x)` | Remove first matching value | `None` |
| `count(x)` | Count occurrences | Integer |
| `index(x)` | Find first matching index | Integer |
| `copy()` | Create a shallow copy | New list |
| `reverse()` | Reverse in place | `None` |
| `sort()` | Sort in place | `None` |


## List Operations: Important Comparisons

### `append()` vs `extend()`

```python
a = [1, 2]
a.append([3, 4])
# [1, 2, [3, 4]]

b = [1, 2]
b.extend([3, 4])
# [1, 2, 3, 4]
```

### `pop()` vs `remove()`

```text
pop()    → removes by index and returns the removed item
remove() → removes by value and returns None
```

### `reverse()` vs slicing

```python
numbers.reverse()
```

modifies the original list, while:

```python
numbers[::-1]
```

creates a new reversed list.
