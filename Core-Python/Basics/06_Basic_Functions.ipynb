## 2. `print()`

The `print()` function displays information on the screen.


````python
print("Hello")
print("Welcome to Python")
print(100)
````

### Printing Multiple Objects

`print()` can display multiple objects separated by commas.


````python
name = "Komal"
age = 25

print("Name:", name)
print("Age:", age)
````

### The `sep` Parameter

`sep` controls the separator placed between multiple objects. The default is a space.


````python
print("Python", "Java", "C++")
print("Python", "Java", "C++", sep="-")
print("2026", "08", "28", sep="/")
````

### The `end` Parameter

`end` controls what is printed after the final object. The default is a newline (`\n`).


````python
print("Hello", end=" ")
print("Python")

print("A", end="-")
print("B", end="-")
print("C")
````

You can also use escape characters such as `\t` and `\n` with `end`.


````python
print("Hello", end="\t")
print("Python")

print("First line", end="\n")
print("Second line")
````

For this beginner course, focus on these `print()` parameters:

| Parameter | Purpose |
|---|---|
| `*objects` | Values to display |
| `sep` | Separator between objects |
| `end` | Text printed after the final object |

The full `print()` function also supports `file` and `flush`, but those are outside the scope of this introductory notebook.


## 3. `type()`

The `type()` function returns the type (class) of an object.


````python
x = 10
name = "Python"
price = 99.5
is_active = True

print(type(x))
print(type(name))
print(type(price))
print(type(is_active))
````

`type()` is useful for inspecting values while learning Python or debugging a program.

Data types are covered in more detail in **03_Datatypes_and_Type_Casting**.


## 4. `input()`

The `input()` function allows a program to receive information from the user.


````python
name = input("Enter your name: ")
print("Hello,", name)
````

### Important: `input()` Always Returns a String

Even when the user enters a number, `input()` returns the entered value as a `str`.


````python
age = input("Enter your age: ")

print("Value:", age)
print("Type:", type(age))
````

### Converting Input to `int`

If you need an integer, convert the input using `int()`.


````python
age = int(input("Enter your age: "))

print("Age:", age)
print("Type:", type(age))
````

### Converting Input to `float`

If you need a decimal number, convert the input using `float()`.


````python
height = float(input("Enter your height in meters: "))

print("Height:", height)
print("Type:", type(height))
````

### Input with Indexing and Slicing

Because the result of `input()` is a string, string indexing and slicing can be applied to it.


````python
text = input("Enter a word: ")

if len(text) >= 3:
    print("Character at index 2:", text[2])
else:
    print("Please enter at least 3 characters.")
````

````python
number = input("Enter a 10-digit number: ")

if len(number) >= 4:
    print("Last four characters:", number[-4:])
else:
    print("Please enter at least 4 characters.")
````

### A Note About `eval()`

You may encounter examples such as:

```python
expression = eval(input("Enter an expression: "))
```

This is **not recommended for beginner input handling**. `eval()` evaluates a string as Python code and can create serious security problems when used with untrusted input.

For normal user input, prefer explicit conversions such as `int(input(...))` and `float(input(...))`.

`eval()` is intentionally not included in this Basic Functions notebook.


## 5. `len()`

The `len()` function returns the number of items in an object. For a string, it returns the number of characters.


````python
text = "Python"

print("Text:", text)
print("Length:", len(text))
````

````python
numbers = [10, 20, 30]

print("List:", numbers)
print("Number of items:", len(numbers))
````

`len()` only works with objects that provide a length. For example, calling it on an integer raises `TypeError`.


````python
try:
    print(len(100))
except TypeError as error:
    print("TypeError:", error)
````

## 6. `abs()`

The `abs()` function returns the **absolute value** of a number.


````python
print(abs(10))
print(abs(-10))
print(abs(5.5))
print(abs(-5.5))
````

## 7. `round()`

The `round()` function rounds a number.

- `round(3.14159)` → nearest integer
- `round(3.14159, 2)` → two decimal places
- `round(12345, -2)` → nearest hundred


````python
print(round(3.14159))
print(round(3.14159, 2))
print(round(12345, -2))
````

> **Note:** Python uses "round half to even" for exact halfway cases, so some results can differ from the traditional "always round .5 upward" rule.


## 8. `min()`

The `min()` function returns the smallest value from an iterable or from multiple arguments.


````python
numbers = [10, 20, 5, 30]

print(min(numbers))
print(min(10, 20, 5, 30))
````

## 9. `max()`

The `max()` function returns the largest value from an iterable or from multiple arguments.


````python
numbers = [10, 20, 5, 30]

print(max(numbers))
print(max(10, 20, 5, 30))
````

## 10. `sum()`

The `sum()` function adds the items in an iterable containing numbers.


````python
numbers = [10, 20, 5, 30]

print(sum(numbers))
````

### Combining `min()`, `max()`, and `sum()`


````python
numbers = [10, 20, 5, 30]

total = sum(numbers)
minimum = min(numbers)
maximum = max(numbers)

print("Numbers:", numbers)
print("Total:", total)
print("Minimum:", minimum)
print("Maximum:", maximum)
````

## 11. Python Exploration Functions

Two useful built-in functions for exploring Python are:

```python
help()
dir()
```


### `help()`

`help()` displays documentation about an object, function, class, module, or other Python feature.


````python
help(print)
````

You can also ask for help about a type or method:

```python
help(str)
help(str.upper)
```

In Jupyter, `?` is another convenient interactive inspection feature, for example `print?`.


### `dir()`

`dir()` returns a list of names and attributes available on an object.


````python
print(dir("Python"))
````

````python
print(dir(min))
````

`dir()` is useful when you want to explore what an object provides. After finding a method, `help()` can provide its documentation.


````python
help(str.upper)
````
