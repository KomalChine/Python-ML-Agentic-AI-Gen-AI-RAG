## 1. Introduction to Tuples

A **tuple** is an ordered, immutable collection of objects.

### Tuple Characteristics

- **Ordered** — elements maintain their order.
- **Immutable** — the tuple itself cannot be changed after creation.
- **Allows duplicates** — the same value can appear more than once.
- **Heterogeneous** — different data types can be stored together.
- **Nestable** — a tuple can contain other tuples or collections.
- **Indexed** — elements can be accessed using indexes.



## 2. Why Use Tuples?

### Immutability and Data Integrity

Use a tuple when values should not be changed after they are created.

For example, coordinates can naturally be represented as a tuple:

```python
coordinates = (12.97, 77.59)
```

### Performance

Tuples can be slightly more memory-efficient and sometimes faster than lists because they are immutable.

> Do not interpret this as "tuples are always faster." The main reason to choose a tuple is usually that the collection should not be modified.

### Dictionary Keys / Hashability — Brief Note

A tuple can be used as a dictionary key **only when all of its elements are hashable**.

For example, this is valid:

```python
key = (1, "Python", 3)
```

A tuple containing a list cannot be used as a dictionary key:

```python
# ([1, 2], 3) is not hashable
```

Deeper hashability concepts are covered later.



## 3. Creating Tuples

### Normal Tuple

Parentheses are commonly used to display and create tuples.



````python
numbers = (10, 20, 30, 40)

print(numbers)
print(type(numbers))

````

### Heterogeneous Tuple

A tuple can contain different data types.



````python
mixed_tuple = (1, "Python", 3.14, True)

print(mixed_tuple)

````

### Empty Tuple

An empty tuple is created with `()`.



````python
empty_tuple = ()

print(empty_tuple)
print(type(empty_tuple))

````

### Single-Element Tuple

A single-element tuple requires a **trailing comma**.

> **The comma creates the tuple, not the parentheses.**



````python
single_item = ("apple",)
not_a_tuple = ("apple")

print(single_item)
print(type(single_item))

print(not_a_tuple)
print(type(not_a_tuple))

````

The comma can also create a tuple without parentheses:

```python
x = 10,
```



````python
x = 10,

print(x)
print(type(x))

````

## 4. Tuple Indexing

Tuple indexing works like list indexing.

Indexes start at `0`.

### Positive Indexing

```text
Index:   0       1       2       3
Value:  red    green    blue   yellow
```



````python
colors = ("red", "green", "blue", "yellow")

print(colors[0])
print(colors[1])
print(colors[3])

````

### Negative Indexing

Negative indexes count from the end.



````python
colors = ("red", "green", "blue", "yellow")

print(colors[-1])
print(colors[-2])

````

## 5. Tuple Slicing

Tuple slicing uses the same syntax as strings and lists:

```python
tuple[start:stop:step]
```

The `stop` index is excluded.



````python
colors = ("red", "green", "blue", "yellow")

print(colors[1:3])
print(colors[:3])
print(colors[::2])
print(colors[::-1])

````

## 6. Tuple Immutability

A tuple itself cannot be modified after creation.

Attempting item assignment raises `TypeError`.



````python
numbers = (1, 2, 3)

try:
    numbers[0] = 99
except TypeError as error:
    print("TypeError:", error)

````

### A Tuple Can Contain Mutable Objects

Immutable means the **tuple structure itself** cannot be changed. A tuple can still contain a mutable object such as a list.

The list inside the tuple can be modified.



````python
data = ([1, 2], 10)

data[0].append(3)

print(data)

````

> This is only a basic demonstration. Deeper reference, shallow-copy, and deep-copy concepts are covered later.



## 7. Tuple Unpacking

Tuple unpacking assigns tuple elements to multiple variables.



````python
coordinates = (12.97, 77.59)

latitude, longitude = coordinates

print(latitude)
print(longitude)

````

### Another Example



````python
person = ("Komal", 25)

name, age = person

print(name)
print(age)

````

### Extended Unpacking

The `*` operator can collect multiple remaining elements into a list.



````python
numbers = (1, 2, 3, 4, 5)

first, *middle, last = numbers

print(first)
print(middle)
print(last)

````

## 8. Tuple Operators

### Concatenation (`+`)

Two tuples can be combined with `+`.



````python
a = (1, 2)
b = (3, 4)

print(a + b)

````

### Repetition (`*`)

A tuple can be repeated with `*`.



````python
a = (1, 2)

print(a * 3)

````

### Membership (`in` and `not in`)



````python
numbers = (10, 20, 30)

print(20 in numbers)
print(50 not in numbers)

````

## 9. Tuple Methods

Tuples provide two commonly used methods:

```text
count()
index()
```



### `count()`

`count(value)` returns the number of times a value appears in the tuple.



````python
letters = ("a", "b", "c", "a", "a")

print(letters.count("a"))
print(letters.count("z"))

````

### `index()`

`index(value)` returns the index of the first occurrence of a value.



````python
letters = ("a", "b", "c", "a", "a")

print(letters.index("a"))
print(letters.index("c"))

````

If the value is not present, `index()` raises `ValueError`.



````python
letters = ("a", "b", "c")

try:
    print(letters.index("z"))
except ValueError as error:
    print("ValueError:", error)

````

## 10. Tuple Conversion

A tuple can be created from another iterable using `tuple()`.

This is only a brief introduction here; detailed data-structure conversions are covered in `12_Type_Casting_Data_Structures`.



````python
numbers = tuple([1, 2, 3])

print(numbers)
print(type(numbers))

````
