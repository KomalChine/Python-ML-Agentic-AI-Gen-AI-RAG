# 14. Loops

Loops allow us to repeat a block of code without writing the same code multiple times.

Python mainly provides two types of loops:

- `for`
- `while`

## 1. `for` Loop

A `for` loop is used to iterate over an iterable.

### Basic syntax

```python
for variable in iterable:
    statements
```

The loop takes each item from the iterable and executes the indented block.

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

### Iterating Over a `range()`

`range()` is commonly used when we want a sequence of numbers for a `for` loop.

````python
for i in range(1, 6):
    print(i)
````

### Iterating Over Other Iterables

A `for` loop can iterate over many Python objects, including:

- lists
- tuples
- strings
- ranges
- dictionaries

````python
for char in "Python":
    print(char)
````

````python
numbers = (10, 20, 30)

for number in numbers:
    print(number)
````

````python
student = {"name": "Alice", "age": 25}

for key in student:
    print(key)
````

## 2. `range()` Function

`range()` generates a sequence of numbers. It is commonly used with `for` loops.

### `range(stop)`

Starts at `0` and stops before `stop`.

````python
for i in range(4):
    print(i)
````

### `range(start, stop)`

Starts at `start` and stops before `stop`.

````python
for i in range(2, 6):
    print(i)
````

### `range(start, stop, step)`

The `step` controls how much the value changes each time.

````python
for i in range(1, 10, 2):
    print(i)
````

### Negative Step

A negative `step` can be used to count backwards.

````python
for i in range(5, 0, -1):
    print(i)
````

### Converting `range()` to a List

A `range` can be converted to a list when you need to see or store all of its values.

````python
numbers = list(range(1, 6))
print(numbers)
````

## 3. `enumerate()`

`enumerate()` is useful when you need both the index and the value while iterating.

````python
names = ["Alice", "Bob", "Charlie"]

for index, name in enumerate(names):
    print(f"{index}: {name}")
````

## 4. `while` Loop

A `while` loop repeatedly executes a block of code as long as its condition remains `True`.

### Basic syntax

```python
while condition:
    statements
```

Make sure the condition eventually becomes `False`; otherwise, the loop can become infinite.

````python
count = 1

while count <= 3:
    print(f"Count is: {count}")
    count += 1
````

### Avoiding Infinite Loops

The variable controlling the condition should normally be updated so that the condition eventually becomes `False`.

For example:

```python
count = 1

while count <= 3:
    print(count)
    count += 1
```

Here, `count += 1` moves the loop toward its stopping condition.

## 5. `for` vs `while`

| `for` | `while` |
|---|---|
| Commonly used to iterate over an iterable | Commonly used when repetition depends on a condition |
| Useful when iterating over a sequence | Useful when the number of iterations depends on a changing condition |
| Often used with `range()` | Requires a condition that eventually becomes `False` |

````python
# for: repeat a known number of times
for i in range(5):
    print(i)
````

````python
# while: repeat while a condition is true
count = 1

while count <= 5:
    print(count)
    count += 1
````

## 6. Loop Control Statements

Loop-control statements change the normal flow of a loop.

- `break` → immediately terminates the nearest enclosing loop.
- `continue` → skips the current iteration and moves to the next iteration.

````python
# break: stop the loop

for num in range(1, 10):
    if num == 5:
        break
    print(num)
````

````python
# continue: skip the current iteration

for num in range(1, 6):
    if num == 3:
        continue
    print(num)
````

### `break` vs `continue`

```text
break
→ stop the loop completely

continue
→ skip the current iteration
→ continue with the next iteration
```

## 7. Loop `else` Clause

Python loops can have an optional `else` block.

The `else` block runs when the loop finishes normally, without executing `break`.

```text
Normal completion → else runs
break             → else does not run
```

````python
numbers = [1, 3, 5, 7]

for num in numbers:
    if num % 2 == 0:
        print("Found an even number!")
        break
else:
    print("No even numbers found.")
````
