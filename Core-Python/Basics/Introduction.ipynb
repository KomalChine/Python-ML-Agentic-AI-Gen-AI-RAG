Module

```python
import math # math is a module
math.floor(5.7) # floor is a function to calculate floor of object
```

```python
# We can use "as" keyword to refer the module with different name
import math as m
m.ceil(4.7)
```

```python
# We can use "from" keyword to directly call functions from modules or modules from packages
from math import pow, floor
pow(2,3) # calculate power
```
