```python
name = 'Python'
name1 = "Python"
message = """Hello
How are you?
Welcome to Python."""
```

```python
# Indexing
# Forward
print("Index 0 ",name[0]) 
print("Index 2 ",name[3])
# Backward

print("Index -1 ",name[-1])
print("Index -2 ",name[-2])
```

```python
# Slicing
print(name[0:3])
print(name[-6:-2])
print(name[-2:-6]) # empty string
# Since step is +1, so -2->-1->0->..., but stop is -6 which is behind therefore it return empty string
print(name[:])
print(name[:5])
print(name[2:])
print(name[2:15]) # thon
# if the stop index is beyond the length of the string, Python simply stops at the end of the string instead of giving an error.
# Slicing with step
# When step is positive we go left to right and when step is negative we go right to left
print(name[0:6:2])
#Reverse string
print(name[::-1])
print(name[-5:-1:-2])
```

```python
# Concatenation
first = "Hello"
second = "World"
result = first + " " + second
print(result)
```

```python
# Repetition
text = "Hi "
print(text * 3)
```

```python
# Methods
text = "hello"
print("Upper- ",text.upper())
print("Lower- ",text.lower())

text1 = "hello world"
print("Capitalize- ",text1.capitalize())
print("Title- ",text1.title())

text2 = "  hello  "
print("Strip- ",text2.strip())
print("LStrip- ",text2.lstrip())
print("RStrip- ",text2.rstrip())

# Remember strings are immutable, so replace() returns new string. It does change original strings.
text3 = "I like Java"
text3 = text3.replace("Java", "Python")
print("Replaced- ",text3)

text4 = "apple,banana,mango"
print("Split- ",text4.split(",")) # Specify separator in argument, by default sparator = whitespace

words = ["Python", "is", "easy"]
result = " ".join(words)
print("Joined- ",result)

print("Find- ",text1.find("world")) # if not present give -1

print("Index- ",text1.index("world")) # if not present throw error

text5 = "banana"
print("Count- ",text5.count("a"))

text6 = "Python programming"
print("StartWith- ",text6.startswith("Python"))
print("EndsWith- ",text6.endswith("programming"))
```
