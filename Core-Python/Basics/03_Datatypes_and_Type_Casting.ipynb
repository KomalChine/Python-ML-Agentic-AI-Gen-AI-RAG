# 03_Datatypes_and_Type_Casting

This notebook introduces Python's fundamental built-in data types and explains how to convert values from one type to another using type casting.

## 1. Introduction to Data Types

A **data type** describes the kind of value an object represents and determines what operations can be performed on that object.

Python provides many built-in data types. In this notebook, we focus on:

- `int` — integers
- `float` — decimal numbers
- `str` — text
- `bool` — Boolean values
- `complex` — complex numbers
- `NoneType` — the type of `None`

Python is dynamically typed, so a variable does not need a type declaration. The type belongs to the object referenced by the variable.

````python
age = 25
price = 99.99
name = "Python"
is_active = True
z = 2 + 3j
result = None

print(age, price, name, is_active, z, result)
````

## 2. Integer (`int`)

The `int` type represents **whole numbers** without a fractional part. Integers can be positive, negative, or zero.

Examples:

````python
positive_number = 100
negative_number = -25
zero = 0

print(positive_number)
print(negative_number)
print(zero)
print(type(positive_number))
````

Python integers have arbitrary precision, so they can represent integers much larger than the fixed-size integer types commonly found in lower-level languages, subject to available memory.

## 3. Float (`float`)

The `float` type represents numbers with a fractional part. Floating-point values are commonly written using a decimal point or scientific notation.

````python
temperature = 36.5
price = 149.99
scientific_value = 1.5e3

print(temperature)
print(price)
print(scientific_value)
print(type(temperature))
````

> **Note:** Floating-point numbers are represented approximately in binary. Therefore, some decimal calculations may produce small rounding differences.

````python
value = 0.1 + 0.2
print(value)
````

## 4. String (`str`)

A `str` object represents **text**. Strings can be written using single quotes, double quotes, or triple quotes.

````python
single_quoted = 'Hello'
double_quoted = "Python"
multi_line = """This is
a multi-line string."""

print(single_quoted)
print(double_quoted)
print(multi_line)
print(type(double_quoted))
````

Strings are sequences of Unicode characters and are **immutable**, meaning their contents cannot be changed in place.

````python
first_name = "Komal"
greeting = "Hello, " + first_name + "!"
print(greeting)
print(first_name[0])
````

## 5. Boolean (`bool`)

The `bool` type represents logical truth values. It has exactly two values:

- `True`
- `False`

Booleans are commonly produced by comparisons and logical expressions.

````python
is_student = True
has_permission = False
age = 20

print(is_student)
print(has_permission)
print(age >= 18)
print(type(is_student))
````

Python's `bool` type is a subclass of `int`. In many numeric contexts, `True` behaves like `1` and `False` behaves like `0`.

````python
print(True + True)
print(False + 10)
````

## 6. Complex (`complex`)

The `complex` type represents **complex numbers**, which contain a real part and an imaginary part.

Python uses `j` to represent the imaginary component.

````python
z1 = 2 + 3j
z2 = 4 - 1j

print(z1)
print(z2)
print(type(z1))
print("Real part:", z1.real)
print("Imaginary part:", z1.imag)
print("Sum:", z1 + z2)
````

Complex numbers are useful in areas such as engineering, signal processing, mathematics, and scientific computing.

## 7. `NoneType`

`None` is a special singleton object that represents the **absence of a value** or "no value". Its type is `NoneType`.

````python
result = None

print(result)
print(type(result))
print(result is None)
````

`None` is different from `0`, `False`, and an empty string. It is commonly used to indicate that a value is missing, unavailable, or intentionally absent.

````python
value = None

if value is None:
    print("No value is available.")
````

## 8. `type()` Function

The built-in `type()` function returns the type (class) of an object.

````python
values = [42, 3.14, "Python", True, 2 + 3j, None]

for value in values:
    print(repr(value), "->", type(value))
````

Typical results include:

```text
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
<class 'complex'>
<class 'NoneType'>
```

`type()` is useful when learning Python, debugging programs, and inspecting data.

# 9. Type Casting

**Type casting** (or type conversion) means converting a value from one data type to another. Python provides built-in functions such as `int()`, `float()`, `str()`, `bool()`, and `complex()` for common conversions.

> **Important:** Not every conversion is possible. Some conversions raise a `ValueError` or `TypeError`.

## 9.1 `int()`

`int()` converts compatible values to an integer.

- A float is truncated toward zero.
- A numeric string can be converted if it represents a valid integer.
- `bool` values convert to `1` or `0`.

````python
print(int(10.9))
print(int(-10.9))
print(int("250"))
print(int(True))
print(int(False))
````

`int()` does not perform normal rounding when converting a float. For example, `int(10.9)` gives `10`.

## 9.2 `float()`

`float()` converts compatible values to floating-point numbers.

````python
print(float(10))
print(float("25.5"))
print(float(True))
print(float(False))
````

A string must contain a valid floating-point representation. For example, `float("3.14")` works, while `float("hello")` raises `ValueError`.

## 9.3 `str()`

`str()` creates a string representation of a value. It is commonly used when displaying or combining non-string values with text.

````python
age = 25
price = 99.5
is_active = True

print(str(age))
print(str(price))
print(str(is_active))
print("Age: " + str(age))
````

For formatted output, f-strings are often more convenient than manually converting values with `str()`:

````python
name = "Komal"
age = 25
print(f"My name is {name} and I am {age} years old.")
````

## 9.4 `bool()`

`bool()` converts a value to a Boolean using Python's truth-value rules.

Common **falsy** values include:

- `False`
- `None`
- `0` and numeric zero values
- `""` (empty string)
- Empty collections such as `[]`, `()`, `{}`, and `set()`

Most other objects are truthy.

````python
print(bool(1))
print(bool(0))
print(bool("Python"))
print(bool(""))
print(bool([1, 2, 3]))
print(bool([]))
print(bool(None))
````

## 9.5 `complex()`

`complex()` creates a complex number. It can be called with a real value, or with real and imaginary components.

````python
print(complex(5))
print(complex(2, 3))
print(complex("4+5j"))
````

The two-argument form `complex(real, imag)` produces `real + imag*j`.

## Type Casting Summary

| Function | Purpose | Example | Result |
|---|---|---|---|
| `int()` | Convert to integer | `int(3.8)` | `3` |
| `float()` | Convert to float | `float(5)` | `5.0` |
| `str()` | Convert to string | `str(25)` | `'25'` |
| `bool()` | Convert to Boolean | `bool(1)` | `True` |
| `complex()` | Convert/create complex number | `complex(2, 3)` | `(2+3j)` |
