Clear Method

```python
# Empty the list. Clear all elements
print(l2)
l2.clear() 
print(l2)
```

```python
# Using clear() (Affects all references):
original = [1,2,3]
alias = original  # Both point to the same list

original.clear()
print(alias)      
# Output: [] (The alias is also emptied)
```

```python
# Reassigning original = [] (Breaks the reference):
original = [1, 2, 3]
alias = original  # Both point to the same list

original = []     # Creates a BRAND NEW empty list
print(alias)      
# Output: [1, 2, 3] (The alias stays unchanged)
```

Copy Method

```python
# copy() returns a shallow copy of the list.
print(l2)
l2 = l.copy()
print(l2)
```

```python
# Modifying the copied list does not alter the original list because they occupy different locations in memory.
original = [1, 2, 3]
clone = original.copy()

clone.append(4)

print(original)  # Output: [1, 2, 3]
print(clone)     # Output: [1, 2, 3, 4]
```

```python
# The copy() method only duplicates the outer list structure. If your list contains nested mutable objects (like sublists or dictionaries),
# both lists will still share references to those inner objects.
original = [[1, 2], [3, 4]]
clone = original.copy()

# Modifying a nested item affects BOTH lists!
clone[0].append(99)

print(original)  # Output: [[1, 2, 99], [3, 4]]
print(clone)     # Output: [[1, 2, 99], [3, 4]]
```

Count Method

```python
# Count the number of occurence of the object
print(l)
print("Total number of 'Hello' = ",l.count("Hello"))
print("Total number of '10' = ",l.count(10))
print("Total number of '7.1' = ",l.count(7.1))
# It does not look inside nested sublists. It only looks at the top-level items.
print("Total number of 'mango' = ",l.count("mango")) # output: 1
# If the item is not in the list, it returns 0 instead of raising an error.
print("Total number of 'cherry' = ",l.count("cherry"))
```

Pop Method

```python
letters = ["a", "b", "c"]
last_item = letters.pop()

print(last_item)  # Output: 'c' (Returns the removed item)
print(letters)    # Output: ['a', 'b'] (Modifies the list)
```

```python
fruits = ["apple", "banana", "cherry"]
second_fruit = fruits.pop(1)

print(second_fruit)  # Output: 'banana'
print(fruits)        # Output: ['apple', 'cherry']
```

Sort Method

```python
# Ascending Order (Default)
numbers = [5, 2, 9, 1]
numbers.sort()
print(numbers)  # Output: [1, 2, 5, 9]

strings = ["cherry", "apple", "banana"]
strings.sort()
print(strings)  # Output: ['apple', 'banana', 'cherry']
```

```python
# Descending Order
numbers = [5, 2, 9, 1]
numbers.sort(reverse=True)
print(numbers)  # Output: [9, 5, 2, 1]
```

```python
# By default, uppercase letters come before lowercase letters in Python
words = ["banana", "apple", "Cherry"]
words.sort()
print(words)  # Output: ['Cherry', 'Apple', 'banana']
```

```python
# Sorting by Length
words = ["elephant", "cat", "banana"]
words.sort(key=len)
print(words)  # Output: ['cat', 'banana', 'elephant']
```
