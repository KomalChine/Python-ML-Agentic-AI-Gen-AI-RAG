Conditional Statemnet

if Statement

```python
age = 20

if age >= 18:
    print("You are eligible to vote.")

# Output: You are eligible to vote.
```

```python
age = 15

if age >= 18:
    print("You are eligible to vote.")

# Output: no output
```

if-else Statement

```python
age = 15

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are too young to vote.")

# Output: You are too young to vote.
```

```python
age = 20

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are too young to vote.")

# Output: You are eligible to vote.
```

if-elif-else Ladder

```python
marks = 75

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
else:
    print("Grade: Fail")

# Output: Grade: B
```

Nested if Statement

```python
has_ticket = True
id_verified = False

if has_ticket:
    print("Ticket verified.")

    if id_verified:
        print("Access granted to the venue.")
    else:
        print("Access denied: Photo ID missing.")
else:
    print("Access denied: No ticket provided.")

# Output:
# Ticket verified.
# Access denied: Photo ID missing.
```

match-case Statement

```python
command = "start"

match command:
    case "start":
        print("System starting...")
    case "stop":
        print("System stopping...")
    case "pause":
        print("System paused.")
    case _:
        print("Unknown command!")

# Output: System starting...
```
