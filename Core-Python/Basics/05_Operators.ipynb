## 1. Introduction to Operators

An **operator** is a symbol or keyword that tells Python to perform an operation on one or more values.

An **operand** is a value on which an operator acts.

For example:

```python
a + b
```

- `a` → operand
- `+` → operator
- `b` → operand

Python has several categories of operators, each designed for a different purpose.


````python
a = 10
b = 5

result = a + b

print("a =", a)
print("Operator = +")
print("b =", b)
print("Result =", result)
````

## 2. Arithmetic Operators

Arithmetic operators perform mathematical calculations.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `10 + 3` |
| `-` | Subtraction | `10 - 3` |
| `*` | Multiplication | `10 * 3` |
| `/` | Division | `10 / 3` |
| `//` | Floor division | `10 // 3` |
| `%` | Modulus / remainder | `10 % 3` |
| `**` | Exponentiation | `10 ** 3` |


````python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)  # Always return float value
print("Floor division:", a // b) # Always return integer value
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
````

### Floor Division: `//`

`//` performs **floor division**. It returns the floor of the mathematical division result, meaning it rounds toward negative infinity.

For positive numbers:

```python
7 // 2
```

gives `3`.

But:

```python
-7 // 2
```

gives `-4`, not `-3`, because the mathematical result is `-3.5` and the floor is `-4`.


````python
print(7 // 2)
print(-7 // 2)
print(7 / 2)
print(-7 / 2)
````

### Arithmetic Operators with Strings

Some operators also have useful behavior with strings:

- `+` → concatenation
- `*` → repetition

These are not numeric arithmetic operations when used with strings.


````python
print("Hello" + " " + "Python")
print("Hi! " * 3)
````

## 3. Assignment Operators

Assignment operators assign values to variables or update existing values.

| Operator | Meaning |
|---|---|
| `=` | Assignment |
| `+=` | Add and assign |
| `-=` | Subtract and assign |
| `*=` | Multiply and assign |
| `/=` | Divide and assign |
| `//=` | Floor-divide and assign |
| `%=` | Modulus and assign |
| `**=` | Exponentiate and assign |


````python
x = 10
print("x =", x)

x += 5
print("After x += 5:", x)

x -= 3
print("After x -= 3:", x)

x *= 2
print("After x *= 2:", x)

x /= 4
print("After x /= 4:", x)

x //= 2
print("After x //= 2:", x)

x %= 3
print("After x %= 3:", x)

x **= 2
print("After x **= 2:", x)
````

The assignment operator can also be used with strings when the operation is supported:

```python
message = "Hello"
message += " Python"
```


````python
message = "Hello"
message += " Python"
print(message)
````

## 4. Comparison Operators

Comparison operators compare values and return a Boolean result: `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |


````python
a = 10
b = 5

print("a == b:", a == b)
print("a != b:", a != b)
print("a > b:", a > b)
print("a < b:", a < b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)
````

Comparison operators are commonly used in conditions:

```python
age >= 18
```

The expression evaluates to `True` or `False`.


````python
age = 20

print(age >= 18)
print(age < 18)
````

## 5. Logical Operators

Logical operators combine or negate Boolean expressions.

### `and`

`and` is `True` only when both operands are true.

| A | B | `A and B` |
|---|---|---|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |


````python
print(True and True)
print(True and False)
print(False and True)
print(False and False)
````

### `or`

`or` is `True` when at least one operand is true.

| A | B | `A or B` |
|---|---|---|
| True | True | True |
| True | False | True |
| False | True | True |
| False | False | False |


````python
print(True or True)
print(True or False)
print(False or True)
print(False or False)
````

### `not`

`not` reverses a Boolean value.

| A | `not A` |
|---|---|
| True | False |
| False | True |


````python
print(not True)
print(not False)
````

### Logical Operators with Conditions

````python
age = 22
has_id = True

can_enter = age >= 18 and has_id
print("Can enter:", can_enter)

is_weekend = False
is_holiday = True

can_relax = is_weekend or is_holiday
print("Can relax:", can_relax)
````

## 6. Identity Operators

Identity operators compare **object identity**, not just value.

- `is` → `True` when two references point to the same object.
- `is not` → `True` when two references do not point to the same object.

### `==` versus `is`

Use:

```python
==
```

to compare **values**.

Use:

```python
is
```

when you specifically need to check **object identity**.


````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b:", a == b)
print("a is b:", a is b)
print("a is c:", a is c)
print("a is not b:", a is not b)
````

The two lists `a` and `b` contain equal values, but they are separate list objects. `c` refers to the same object as `a`.

A common and important use of `is` is checking for `None`:

```python
value is None
```

Do not use `is` as a general replacement for `==` when comparing numbers, strings, or other values.


````python
value = None

if value is None:
    print("No value was provided.")
````

## 7. Membership Operators

Membership operators test whether a value exists inside a container or sequence.

- `in` → membership exists
- `not in` → membership does not exist


````python
text = "Python programming"

print("Python" in text)
print("Java" in text)
print("Java" not in text)

numbers = [10, 20, 30, 40]

print(20 in numbers)
print(50 not in numbers)
````

Membership tests with strings are case-sensitive:

```python
"Python" in "Python programming"   # True
"python" in "Python programming"   # False
```


````python
text = "Python programming"

print("Python" in text)
print("python" in text)
````

## 8. Bitwise Operators

Bitwise operators work with the binary representation of integers.

| Operator | Name |
|---|---|
| `&` | Bitwise AND |
| `|` | Bitwise OR |
| `^` | Bitwise XOR |
| `~` | Bitwise NOT |
| `<<` | Left shift |
| `>>` | Right shift |


### Bitwise AND: `&`

A bit is `1` only when both corresponding bits are `1`.

```text
6  = 110
3  = 011
---------
&  = 010
```

So `6 & 3` is `2`.


````python
a = 6
b = 3

print(a & b)
````

### Bitwise OR: `|`

A bit is `1` when at least one corresponding bit is `1`.


````python
print(6 | 3)
````

### Bitwise XOR: `^`

A bit is `1` when the corresponding bits are different.


````python
print(6 ^ 3)
````

### Bitwise NOT: `~`

Python integers use signed, arbitrary-precision integer semantics. A useful identity is:

```python
~n == -(n + 1)
```

For example:

```python
~5
```

produces `-6`.


````python
print(~5)
print(~10)
print(~(-1))
````

### Left Shift: `<<`

A left shift moves the binary bits to the left. For non-negative integers, shifting left by one position is equivalent to multiplying by `2`.

```python
5 << 1
```

produces `10`.


````python
print(5 << 1)
print(5 << 2)
````

### Right Shift: `>>`

A right shift moves bits to the right. For non-negative integers, shifting right by one position is equivalent to floor-dividing by `2`.


````python
print(20 >> 1)
print(20 >> 2)
````

## 9. Unary Operators

A **unary operator** works with one operand.

Common unary operations include:

- `+x` → unary plus
- `-x` → unary minus
- `~x` → bitwise NOT
- `not x` → logical NOT

Some of these operators have already appeared in earlier sections; this section groups them by the fact that they operate on a single value.


````python
x = 5

print("+x =", +x)
print("-x =", -x)
print("~x =", ~x)
print("not x =", not x)
````

## 10. Operator Precedence

When an expression contains multiple operators, Python follows a precedence order to determine which operations are evaluated first.

A simplified order from higher to lower precedence is:

1. Parentheses: `(...)`
2. Exponentiation: `**`
3. Unary `+`, unary `-`, `~`
4. Multiplication, division, floor division, modulus: `*`, `/`, `//`, `%`
5. Addition and subtraction: `+`, `-`
6. Comparisons: `==`, `!=`, `<`, `>`, `<=`, `>=`
7. `not`
8. `and`
9. `or`

When in doubt, use parentheses to make the intended order explicit.


### Progressive Examples

Start with a simple arithmetic expression:


````python
result = 2 + 3 * 4
print(result)
````

Multiplication has higher precedence than addition, so this is equivalent to:

```python
2 + (3 * 4)
```


````python
result = (2 + 3) * 4
print(result)
````

Parentheses override the normal precedence and make addition happen first.

Exponentiation has higher precedence than multiplication:


````python
result = 2 ** 3 * 4
print(result)
````

This is evaluated as:

```python
(2 ** 3) * 4
```

Finally, consider a more complex expression:


````python
result = 10 + 2 * 3 ** 2 > 25 and not (8 // 3 == 2 or 5 % 2 != 1)
print(result)
````

### Best Practice

Do not try to memorize every precedence rule before writing programs. Use parentheses when an expression becomes difficult to read.

For example:

```python
result = (10 + (2 * (3 ** 2))) > 25
```

Clear code is usually better than relying on a reader to remember a complicated precedence chain.
