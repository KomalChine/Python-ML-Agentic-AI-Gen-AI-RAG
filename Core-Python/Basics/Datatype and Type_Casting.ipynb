Datatype

```python
it = 3 #integer
f = 6.7 # float
s = "komal" # string
b = True # boolean
c = 3 + 5j # complex
# type function tells datatype of object
print("Datatye of it ",type(it))
print("Datatye of f ",type(f))
print("Datatye of s ",type(s))
print("Datatye of b ",type(b))
print("Datatye of c ",type(c))
```

Type Casting

```python
# Changing from integer to other datatype
i = float(9) # To float
print(i) # output:9.0
print(type(i))

i = str(4) # To string
# print(i) will give only 4 as output not '4' since it doesn't print "" for string
print(i) # output:4
print(type(i))

i = bool(9) # To boolean. Every integer or float is True other than 0, even if it is negative
print(i) # output:True
print(type(i))

i = complex(9) # To complex
print(i) # output:9+0j
print(type(i))
```

```python
# Changing from float to other datatype
flt = int(7.8)
print(flt) # output:7
print(type(flt))

flt = complex(7.8)
print(flt) # output:7.8+0j
print(type(flt))

flt = str(7.8)
print(flt) # output:7.8
print(type(flt))

flt = bool(7.8)
print(flt) # output:True
print(type(flt))
```

```python
# Changing from string to other datatype
text = int("7")
print(text) # output:7
print(type(text))

text = bool("komal")
print(text) # output:True
print(type(text))

text = bool("")
print(text) # output:False
print(type(text))

text = float("7")
print(text) # output:7.0
print(type(text))

text = complex("7")
print(text) # output:7+0j
print(type(text))
```

```python
# Changing from boolean to other datatype
boolean = int(True)
print(boolean) # output:1, if False:0
print(type(boolean))

boolean = float(True)
print(boolean) # output:1.0, if False:0.0
print(type(boolean))

boolean = str(True)
print(boolean) # output:True, if False:False
print(type(boolean))

boolean = complex(True)
print(boolean) # output:1+0l,, if False:0j
print(type(boolean))
```

```python
# Changing from complex to other datatype
# int and float does not take complex as a argument, it throws TypeError
com = str(2+9j)
print(com) # output:2+9j
print(type(com))

com = bool(2+9j)
print(com) # output:True, if 0j then False
print(type(com))
```
