Datatype

```python
it = 3 #integer
f = 6.7 # float
s = "komal" # string
b = True # boolean
c = 3 + 5j # complex
# type function tells datatype of object
print("Datatye of it ",type(it))
print("Datatye of f ",type(f))
print("Datatye of s ",type(s))
print("Datatye of b ",type(b))
print("Datatye of c ",type(c))
```

Type Casting

```python
flt = int(7.8)
print(flt) # output:7
print(type(flt))

text = int("7")
print(text) # output:7
print(type(text))

text = int("A1",16)
print(text) # output:161
print(type(text))

boolean = int(True)
print(boolean) # output:1
print(type(boolean))

boolean = int(False)
print(boolean) # output:0
print(type(boolean))
```

```python
i = float(9) 
print(i) # output:9.0
print(type(i))

text = float("7")
print(text) # output:7.0
print(type(text))

boolean = float(True)
print(boolean) # output:1.0, if False:0.0
print(type(boolean))
```

```python
i = str(4) 
# print(i) will give only 4 as output not '4' since it doesn't print "" for string
print(i) # output:4
print(type(i))

flt = str(7.8)
print(flt) # output:7.8
print(type(flt))

boolean = str(True)
print(boolean) # output:True, if False:False
print(type(boolean))

com = str(2+9j)
print(com) # output:2+9j
print(type(com))
```

```python
i = bool(9) # Every integer or float is True other than 0, even if it is negative
print(i) # output:True
print(type(i))

flt = bool(7.8) # Same as integer
print(flt) # output:True
print(type(flt))

text = bool("komal") 
print(text) # output:True
print(type(text))

text = bool("")
print(text) # output:False
print(type(text))

com = bool(2+9j)
print(com) # output:True, if 0j then False
print(type(com))
```

```python
i = complex(2,4)
print(i) # output: 2+4j
print(type(i))

flt = complex(3.6)
print(flt) # output: 3.6+0j
print(type(flt))

text = complex("7")
print(text) # output:7+0j
print(type(text))

text = complex("7",6)
print(text) # output:7+6j
print(type(text))

boolean = complex(True)
print(boolean) # output:1+0j, if False:0j
print(type(boolean))

text = complex("7","6")
print(text) # output: TypeError
print(type(text))
```
