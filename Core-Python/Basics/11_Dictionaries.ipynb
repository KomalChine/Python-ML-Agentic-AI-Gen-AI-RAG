# 11. Dictionaries

A dictionary is a **mutable collection of key-value pairs**.

Each item has:

```text
key → value
```

**Keys**
- must be unique
- must be hashable

**Values**
- can be any Python object
- can be duplicated

## 1. Dictionary Characteristics

- **Mutable** → dictionaries can be changed after creation.
- **Ordered** → dictionaries preserve insertion order.
- **Unique keys** → a key can appear only once.
- **Hashable keys** → common examples include strings, numbers, and tuples containing hashable elements.
- **Values can be any object** → values may be strings, numbers, lists, tuples, dictionaries, and more.

## 2. Creating Dictionaries

A dictionary can be created using `{}` or `dict()`.

The empty dictionary is created with `{}`. Note that `{}` is **not** an empty set.

````python
# Dictionary literal
person = {
    "name": "Alice",
    "age": 25,
    "skills": ["Python", "SQL"]
}

# Empty dictionaries
empty_dict = {}
empty_dict_2 = dict()

print(person)
print(empty_dict)
print(empty_dict_2)
````

## 3. Accessing Values

Use a key to access its corresponding value.

```python
dictionary[key]
```

If the key does not exist, `dict[key]` raises a `KeyError`.

````python
student = {"name": "Bob", "grade": "A"}

print(student["name"])
print(student["grade"])
````

````python
# Missing key → KeyError
print(student["age"])
````

## 4. Adding and Updating Values

The same syntax is used for both adding and updating:

```python
dictionary[key] = value
```

- If the key does not exist → a new key-value pair is added.
- If the key already exists → its value is updated.

````python
student = {"name": "Bob", "grade": "A"}

# Add a new key
student["age"] = 20

# Update an existing key
student["grade"] = "A+"

print(student)
````

## 5. Dictionary Membership

Use `in` and `not in` to check whether a **key** exists in a dictionary.

> Membership testing on a dictionary checks keys, not values.

````python
user = {"name": "Alice", "age": 25}

print("name" in user)
print("email" not in user)

# This checks for a key, not a value
print(25 in user)
````

## 6. Removing Items

### `pop()`

`pop(key)` removes a specific key and returns its value.

If the key is missing, a `KeyError` is raised unless a default value is provided.

### `popitem()`

`popitem()` removes and returns the **last inserted key-value pair**.

### `clear()`

`clear()` removes all key-value pairs from the dictionary.

````python
user = {"name": "Alice", "age": 25, "role": "Admin"}

role = user.pop("role")
print(role)
print(user)

email = user.pop("email", "Not Found")
print(email)
````

````python
user = {"name": "Alice", "age": 25, "role": "Admin"}

item = user.popitem()

print(item)
print(user)
````

````python
user = {"name": "Alice", "role": "Admin", "active": True}

user.clear()

print(user)
````

## 7. Dictionary Methods

Python provides several useful methods for working with dictionaries:

```text
get()
keys()
values()
items()
update()
setdefault()
```

### `get()`

`get()` returns the value for a key.

If the key does not exist, it returns `None` by default instead of raising `KeyError`.

You can also provide a custom default value.

````python
user = {"name": "Alice", "role": "Admin"}

print(user.get("name"))
print(user.get("age"))
print(user.get("role", "Guest"))
print(user.get("country", "India"))
````

### `keys()`

`keys()` returns a view containing the dictionary's keys.

````python
user = {"name": "Alice", "age": 25, "role": "Admin"}

print(user.keys())
````

### `values()`

`values()` returns a view containing all values in the dictionary.

Dictionary values may contain duplicates.

````python
scores = {"Math": 90, "Science": 85, "English": 90}

print(scores.values())
````

### `items()`

`items()` returns the dictionary's key-value pairs.

Each pair is represented as a two-element tuple:

```text
(key, value)
```

`items()` is especially useful when iterating through both keys and values.

````python
user = {"name": "Alice", "age": 25, "role": "Admin"}

print(user.items())
````

````python
scores = {"Math": 90, "Science": 85, "English": 92}

for subject, score in scores.items():
    print(f"{subject}: {score}")
````

### `update()`

`update()` adds new key-value pairs and updates existing keys.

It modifies the original dictionary and returns `None.

````python
user = {"name": "Alice", "age": 25}
new_data = {"age": 26, "city": "New York"}

user.update(new_data)

print(user)
````

````python
profile = {"username": "coder123"}

profile.update(status="Active", score=100)

print(profile)
````

### `setdefault()`

`setdefault()` returns the value of a key if it exists.

If the key does not exist, it adds the key with the specified default value and returns that value.

```text
Key exists  → return existing value
Key missing → add key with default → return default
```

````python
user = {"name": "Alice", "role": "Admin"}

role = user.setdefault("role", "Guest")
print(role)

age = user.setdefault("age", 25)
print(age)

print(user)
````

## 8. Copying Dictionaries

`copy()` creates a separate dictionary containing the same key-value pairs.

For this Basic course, the important idea is that changing the copied dictionary does not change the original dictionary in this simple example.

````python
original = {"a": 1, "b": 2}
duplicate = original.copy()

duplicate["a"] = 99

print(original)
print(duplicate)
````

## 9. Nested Dictionaries

A dictionary value can itself be another dictionary.

Nested dictionaries are useful for representing structured data.

````python
student = {
    "name": "Alice",
    "marks": {
        "math": 90,
        "science": 85
    }
}

print(student["name"])
print(student["marks"]["math"])
````

## 10. Iterating Through Dictionaries

You can iterate through a dictionary in several common ways.

### Iterate through keys

````python
user = {"name": "Alice", "age": 25, "role": "Admin"}

for key in user:
    print(key)
````

### Iterate through key-value pairs

Use `items()` when you need both the key and the value.

````python
user = {"name": "Alice", "age": 25, "role": "Admin"}

for key, value in user.items():
    print(key, value)
````

## 11. `fromkeys()`

`dict.fromkeys()` creates a new dictionary using a sequence of keys.

If no value is supplied, each key receives `None`.

````python
keys = ["name", "age", "role"]

user = dict.fromkeys(keys)

print(user)
````

````python
subjects = ["math", "science", "english"]

gradebook = dict.fromkeys(subjects, 0)

print(gradebook)
````
