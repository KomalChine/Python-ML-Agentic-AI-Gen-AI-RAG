Dictionary

```python
# 1. Using curly braces {}
person = {
    "name": "Alice",
    "age": 25,
    "skills": ["Python", "SQL"]
}

# 2. Creating an empty dictionary
empty_dict = {}       # Preferred
empty_dict_2 = dict() # Alternative

print(person)
# Output: {'name': 'Alice', 'age': 25, 'skills': ['Python', 'SQL']}
```

```python
# If the key is not there in dictionary and you try to access it, then Python throw KeyError
print(person["book"])
```

```python
student = {"name": "Bob", "grade": "A"}

# Accessing a value
# For accessing values we use keys not index in dictionary
print(student["name"])  # Output: Bob
```

```python
# Adding a new key-value pair
student["age"] = 20

print(student)  # Output: {'name': 'Bob', 'grade': 'A', 'age': 20}
```

```python
# Updating an existing value
student["grade"] = "A+"

print(student)
# Output: {'name': 'Bob', 'grade': 'A+', 'age': 20}
```

Clear Method

```python
user = {"name": "Alice", "role": "Admin", "active": True}

user.clear()

print(user)  
# Output: {}
```

Copy Method

```python
original = {"a": 1, "b": 2}
duplicate = original.copy()

# Updating the copy
duplicate["a"] = 99

print(original)   # Output: {'a': 1, 'b': 2}  (Unchanged)
print(duplicate)  # Output: {'a': 99, 'b': 2}
```

Fromkeys Method

```python
keys = ["name", "age", "role"]

user = dict.fromkeys(keys)

print(user)  
# Output: {'name': None, 'age': None, 'role': None}
```

```python
scores = ["math", "science", "english"]

# Setting initial score to 0 for all subjects
gradebook = dict.fromkeys(scores, 0)

print(gradebook)  
# Output: {'math': 0, 'science': 0, 'english': 0}
```

```python
# Dangerous: All keys share the same list instance!
data = dict.fromkeys(["a", "b"], [])
data["a"].append(100)

print(data)  
# Output: {'a': [100], 'b': [100]}  <-- Both 'a' and 'b' changed!
```

Get Method

```python
user = {"name": "Alice", "role": "Admin"}

# Existing key
print(user.get("name"))  
# Output: Alice

# Missing key (returns None by default)
print(user.get("age"))   
# Output: None
```

```python
# Providing a Custom Default Value
user = {"name": "Alice"}

# Key doesn't exist, so custom default is returned
role = user.get("role", "Guest")

print(role)  
# Output: Guest
```

Items Method

```python
user = {"name": "Alice", "age": 25, "role": "Admin"}

print(user.items())  
# Output: dict_items([('name', 'Alice'), ('age', 25), ('role', 'Admin')])
```

```python
# Looping with Tuple Unpacking
scores = {"Math": 90, "Science": 85, "English": 92}

for subject, score in scores.items():
    print(f"{subject}: {score}")

# Output:
# Math: 90
# Science: 85
# English: 92
```

```python
# Dynamic View Property
data = {"a": 1, "b": 2}
items_view = data.items()

data["c"] = 3  # Add a new item to original dict

print(items_view)  
# Output: dict_items([('a', 1), ('b', 2), ('c', 3)])
```

Keys Method

```python
user = {"name": "Alice", "age": 25, "role": "Admin"}

print(user.keys())  
# Output: dict_keys(['name', 'age', 'role'])
```

```python
scores = {"Math": 90, "Science": 85, "English": 92}

user_keys = user.keys()
scores_keys = scores.keys()

# Cannot use method like union, so use operators instead
Union = user_keys | scores_keys
print(Union)
```

Pop and Popitem Method

```python
user = {"name": "Alice", "age": 25, "role": "Admin"}

# Remove specific key
role = user.pop("role")
print(role)  # Output: Admin
print(user)  # Output: {'name': 'Alice', 'age': 25}

# Safe removal with default (key doesn't exist)
email = user.pop("email", "Not Found")
print(email) # Output: Not Found
```

```python
user = {"name": "Alice", "age": 25, "role": "Admin"}

# Removes the last item ("role": "Admin")
item = user.popitem()

print(item) # Output: ('role', 'Admin')
print(user) # Output: {'name': 'Alice', 'age': 25}
```

Setdefault Method

```python
user = {"name": "Alice", "role": "Admin"}

# 1. Key EXISTS: returns existing value "Admin" (does NOT overwrite)
role = user.setdefault("role", "Guest")
print(role)  # Output: Admin

# 2. Key MISSING: adds 'age': 25 to user and returns 25
age = user.setdefault("age", 25)
print(age)   # Output: 25

print(user)  
# Output: {'name': 'Alice', 'role': 'Admin', 'age': 25}
```

Update Method

```python
# Updating with Another Dictionary
user = {"name": "Alice", "age": 25}
new_data = {"age": 26, "city": "New York"}  # 'age' exists, 'city' is new

user.update(new_data)

print(user)  
# Output: {'name': 'Alice', 'age': 26, 'city': 'New York'}
```

```python
# Updating with Keyword Arguments (kwargs)
profile = {"username": "coder123"}

profile.update(status="Active", score=100)

print(profile)  
# Output: {'username': 'coder123', 'status': 'Active', 'score': 100}
```

```python
# Updating with a List of Tuples
scores = {"Math": 90}

scores.update([("Science", 85), ("English", 92)])

print(scores)  
# Output: {'Math': 90, 'Science': 85, 'English': 92}
```

Values Method

```python
scores = {"Math": 90, "Science": 85, "English": 90}

print(scores.values())  
# Output: dict_values([90, 85, 90])
```
