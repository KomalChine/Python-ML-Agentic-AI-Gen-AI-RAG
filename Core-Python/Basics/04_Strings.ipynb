# 04_Strings

This notebook provides a practical introduction to Python strings, including indexing, slicing, concatenation, repetition, membership, string methods, f-strings, and string immutability.

## 1. Introduction to Strings

### What is a String?

A **string** (`str`) is a sequence of Unicode characters used to represent text. Strings can contain letters, numbers, spaces, punctuation, and special characters.

Examples:

````python
name = "Komal"
message = "Welcome to Python!"
number_as_text = "12345"

print(name)
print(message)
print(number_as_text)
print(type(name))
````

### Single Quotes

Strings can be enclosed in single quotes (`'...'`).

````python
city = 'Hyderabad'
print(city)
````

### Double Quotes

Strings can also be enclosed in double quotes (`"..."`). Both single and double quotes create the same `str` type.

````python
language = "Python"
print(language)
print(type(language))
````

Using double quotes is convenient when the string contains an apostrophe:

````python
sentence = "Python's syntax is readable."
print(sentence)
````

### Triple Quotes

Triple single quotes (`'''...'''`) or triple double quotes (`\"\"\"...\"\"\"`) can be used for multi-line strings.

````python
multi_line = """Python is easy to learn.
It is widely used in many fields.
This is a multi-line string."""

print(multi_line)
````

Triple-quoted strings are also commonly used for **docstrings**, which document functions, classes, and modules.

## 2. String Indexing

Indexing accesses one character from a string. Python uses **zero-based indexing**, so the first character has index `0`.

### Positive Indexing

For the string `Python`, the indexes are:

```text
Character:  P  y  t  h  o  n
Index:      0  1  2  3  4  5
```

````python
text = "Python"

print(text[0])
print(text[1])
print(text[5])
````

### Negative Indexing

Negative indexes count from the end of the string. The last character has index `-1`.

````python
text = "Python"

print(text[-1])
print(text[-2])
print(text[-6])
````

Negative indexing can be visualized as:

```text
Character:  P   y   t   h   o   n
Positive:   0   1   2   3   4   5
Negative:  -6  -5  -4  -3  -2  -1
```

### IndexError

Trying to access an index outside the valid range raises `IndexError`.

````python
text = "Python"

try:
    print(text[10])
except IndexError as error:
    print("IndexError:", error)
````

The example above catches the error so that the notebook continues running. Without the `try`/`except`, `text[10]` would stop that cell with an `IndexError`.

## 3. String Slicing

Slicing extracts a portion of a string using the syntax:

```python
string[start:stop:step]
```

- `start` — where slicing begins (included)
- `stop` — where slicing ends (excluded)
- `step` — how many positions to move at a time

If an argument is omitted, Python uses an appropriate default.

### `start`

````python
text = "Python Programming"

print(text[7:])
````

### `stop`

````python
text = "Python Programming"

print(text[:6])
print(text[0:6])
````

### `step`

````python
text = "Python Programming"

print(text[::2])
print(text[1::2])
````

### Negative Slicing

Negative values can be used for `start`, `stop`, or `step`.

````python
text = "Python Programming"

print(text[-11:])
print(text[-11:-1])
print(text[::-2])
````

### Reversing a String

A step of `-1` reverses a string.

````python
text = "Python"
reversed_text = text[::-1]

print(reversed_text)
````

## 4. String Concatenation

**Concatenation** means joining strings together. The `+` operator can be used to concatenate strings.

````python
first_name = "Komal"
last_name = "Chine"
full_name = first_name + " " + last_name

print(full_name)
````

Python does not automatically concatenate strings with numbers. Convert the number to a string first, or use an f-string.

````python
age = 25
message = "Age: " + str(age)
print(message)
````

## 5. String Repetition

The `*` operator repeats a string a specified number of times.

````python
pattern = "Python! "
print(pattern * 3)

print("-" * 30)
````

## 6. String Membership

The `in` and `not in` operators check whether a substring or character is present in a string.

### `in`

````python
text = "Python programming"

print("Python" in text)
print("program" in text)
print("Java" in text)
````

### `not in`

````python
text = "Python programming"

print("Java" not in text)
print("Python" not in text)
````

Membership checks are **case-sensitive**:

````python
text = "Python"
print("python" in text)
print("Python" in text)
````

## 7. String Methods

String methods are functions associated with string objects. They provide convenient ways to transform, search, validate, split, and otherwise work with text.

> **Important:** Most string methods return a **new string** rather than changing the original string, because strings are immutable.

### Case Conversion

#### `upper()`

````python
text = "Python Programming"
print(text.upper())
````

#### `lower()`

````python
text = "Python Programming"
print(text.lower())
````

#### `capitalize()`

Converts the first character to uppercase and the remaining characters to lowercase.

````python
text = "python PROGRAMMING"
print(text.capitalize())
````

#### `title()`

Converts the first character of each word to uppercase and the remaining characters to lowercase.

````python
text = "python programming language"
print(text.title())
````

#### `swapcase()`

Swaps uppercase characters to lowercase and lowercase characters to uppercase.

````python
text = "Python PROGRAMMING"
print(text.swapcase())
````

### Searching

#### `find()`

`find()` returns the lowest index where a substring is found. If it is not found, it returns `-1`.

````python
text = "Python programming"
print(text.find("program"))
print(text.find("Java"))
````

#### `index()`

`index()` is similar to `find()`, but raises `ValueError` if the substring is not found.

````python
text = "Python programming"
print(text.index("program"))

try:
    print(text.index("Java"))
except ValueError as error:
    print("ValueError:", error)
````

#### `count()`

`count()` returns the number of non-overlapping occurrences of a substring.

````python
text = "banana"
print(text.count("a"))
print(text.count("an"))
````

### Checking

#### `startswith()`

````python
text = "Python Programming"
print(text.startswith("Python"))
print(text.startswith("Java"))
````

#### `endswith()`

````python
filename = "report.pdf"
print(filename.endswith(".pdf"))
print(filename.endswith(".txt"))
````

#### `isalpha()`

Returns `True` if all characters are alphabetic and the string is not empty.

````python
print("Python".isalpha())
print("Python3".isalpha())
print("Hello World".isalpha())
````

#### `isdigit()`

Returns `True` when all characters are digits and the string is not empty.

````python
print("12345".isdigit())
print("123a".isdigit())
print("12.5".isdigit())
````

#### `isalnum()`

Returns `True` if all characters are alphanumeric (letters or digits) and the string is not empty.

````python
print("Python3".isalnum())
print("Python 3".isalnum())
print("Python_3".isalnum())
````

### Modification

#### `replace()`

`replace(old, new)` returns a new string with occurrences of `old` replaced by `new`.

````python
text = "I like Java. Java is popular."
new_text = text.replace("Java", "Python")

print(new_text)
print(text)
````

#### `strip()`

Removes leading and trailing whitespace by default.

````python
text = "   Python Programming   "
print(repr(text.strip()))
````

`strip()` can also remove specified characters from both ends:

````python
text = "---Python---"
print(text.strip("-"))
````

#### `lstrip()`

Removes leading whitespace (or specified characters).

````python
text = "   Python   "
print(repr(text.lstrip()))
````

#### `rstrip()`

Removes trailing whitespace (or specified characters).

````python
text = "   Python   "
print(repr(text.rstrip()))
````

### Splitting & Joining

#### `split()`

`split()` breaks a string into a list of substrings. By default, it splits on whitespace.

````python
text = "Python is easy to learn"
words = text.split()

print(words)
print(type(words))
````

You can provide a separator:

````python
data = "apple,banana,orange"
fruits = data.split(",")
print(fruits)
````

#### `join()`

`join()` combines an iterable of strings using the string on which the method is called.

````python
words = ["Python", "is", "powerful"]
sentence = " ".join(words)
print(sentence)

csv_line = ",".join(["apple", "banana", "orange"])
print(csv_line)
````

## 8. Formatted Strings (f-strings)

**F-strings** provide a concise way to insert expressions into strings. Put `f` or `F` before the opening quote and place expressions inside `{}`.

````python
name = "Komal"
age = 25
city = "Hyderabad"

message = f"My name is {name}, I am {age} years old, and I live in {city}."
print(message)
````

Expressions can also be used inside f-strings:

````python
price = 1250
quantity = 3
total = price * quantity

print(f"Price: ₹{price}")
print(f"Quantity: {quantity}")
print(f"Total: ₹{total}")
````

F-strings also support formatting specifications. For example, `.2f` displays a floating-point value with two decimal places.

````python
amount = 1234.5678
print(f"Amount: ₹{amount:.2f}")
````

## 9. String Immutability

Python strings are **immutable**. This means that once a string object is created, its characters cannot be changed in place.

The following would raise a `TypeError` if executed directly:

```python
text = "Python"
text[0] = "J"
```

Instead, create a new string:

````python
text = "Python"
new_text = "J" + text[1:]

print("Original:", text)
print("New:", new_text)
````

String methods also return new strings rather than modifying the original object:

````python
text = "python"
upper_text = text.upper()

print("Original:", text)
print("Modified copy:", upper_text)
````
