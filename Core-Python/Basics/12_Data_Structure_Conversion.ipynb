# 12. Data Structure Conversion

Python provides built-in constructor functions that can convert one data structure into another.

Common conversion functions:

```python
list()
tuple()
set()
dict()
```

This notebook focuses specifically on converting between Python's basic data structures:

```text
List ↔ Tuple ↔ Set ↔ Dictionary
```

> This notebook focuses on **data structure conversion**. Basic type casting such as `int()`, `float()`, `str()`, and `bool()` is covered separately.

## 1. Conversion Overview

| From → To | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| List | — | `tuple(lst)` | `set(lst)` | `dict(lst)` if elements are pairs |
| Tuple | `list(tup)` | — | `set(tup)` | `dict(tup)` if elements are pairs |
| Set | `list(s)` | `tuple(s)` | — | `dict(s)` if elements are pairs |
| Dictionary | `list(d)` → keys | `tuple(d)` → keys | `set(d)` → keys | — |

### Important

- Converting a **set** to a list or tuple does not guarantee a particular order.
- Converting a **list or tuple to a set** removes duplicate values.
- Converting a **dictionary** directly to a list, tuple, or set uses its **keys** by default.
- Converting pairs to a dictionary requires each element to contain exactly two values: a key and a value.

## 2. Convert to List

### Tuple → List

Use `list()` to convert a tuple into a list.

````python
tup = (1, 2, 3)

result = list(tup)

print(result)
print(type(result))
````

### Set → List

Use `list()` to convert a set into a list.

> The order of elements from a set is not guaranteed.

````python
s = {4, 5, 6}

result = list(s)

print(result)
print(type(result))
````

### List → List

Converting a list to a list does not remove duplicates.

````python
numbers = [1, 2, 2, 3, 4, 4]

result = list(numbers)

print(result)
````

## 3. Convert to Tuple

### List → Tuple

Use `tuple()` to convert a list into a tuple.

````python
lst = [1, 2, 3]

result = tuple(lst)

print(result)
print(type(result))
````

### Set → Tuple

Use `tuple()` to convert a set into a tuple.

> Because sets are unordered, the resulting tuple does not have a guaranteed order.

````python
s = {4, 5, 6}

result = tuple(s)

print(result)
print(type(result))
````

## 4. Convert to Set

### List → Set

Use `set()` to convert a list into a set.

One common use is **removing duplicate values**.

````python
lst = [1, 2, 2, 3, 3, 3]

result = set(lst)

print(result)
````

### Tuple → Set

The same conversion works with tuples.

Duplicate values are removed.

````python
tup = ("a", "b", "a", "c")

result = set(tup)

print(result)
````

## 5. Convert Dictionary to Other Data Structures

When a dictionary is converted directly to a list, tuple, or set, Python uses the dictionary's **keys** by default.

````python
d = {"a": 1, "b": 2}

print(list(d))
print(tuple(d))
print(set(d))
````

### Dictionary → Values

To convert dictionary values, explicitly use `values()`.

````python
d = {"a": 1, "b": 2}

print(list(d.values()))
print(tuple(d.values()))
print(set(d.values()))
````

### Dictionary → Key-Value Pairs

To convert dictionary key-value pairs, use `items()`.

Each item is represented as a two-element tuple:

```text
(key, value)
```

````python
d = {"a": 1, "b": 2}

print(list(d.items()))
print(tuple(d.items()))
print(set(d.items()))
````

## 6. Convert Pairs to Dictionary

A sequence of two-item pairs can be converted into a dictionary using `dict()`.

Each pair must contain:

```text
(key, value)
```

### List of Pairs → Dictionary

````python
pairs_list = [("name", "Alice"), ("age", 25)]

result = dict(pairs_list)

print(result)
````

### Tuple of Pairs → Dictionary

````python
pairs_tuple = (
    ("name", "Alice"),
    ("age", 25)
)

result = dict(pairs_tuple)

print(result)
````

### Set of Pairs → Dictionary

A set containing two-item tuples can also be converted into a dictionary.

> Because sets are unordered, the insertion order of the resulting dictionary should not be relied upon.

````python
pairs_set = {("x", 10), ("y", 20)}

result = dict(pairs_set)

print(result)
````

## 7. Conversion Requirements

When converting a sequence of pairs to a dictionary, each element must contain exactly two values.

For example:

```python
[("name", "Alice"), ("age", 25)]
```

is valid because every element contains two values.

A sequence such as:

```python
[(1, 2, 3)]
```

cannot be directly converted into a dictionary because the element contains three values.

````python
# This raises ValueError because the pair contains three values.
dict([(1, 2, 3)])
````

## 8. Conversion Summary

| Conversion | Example | Main Point |
|---|---|---|
| Tuple → List | `list((1, 2, 3))` | Creates a list |
| Set → List | `list({1, 2, 3})` | Order is not guaranteed |
| List → Tuple | `tuple([1, 2, 3])` | Creates a tuple |
| Set → Tuple | `tuple({1, 2, 3})` | Order is not guaranteed |
| List → Set | `set([1, 2, 2, 3])` | Removes duplicates |
| Tuple → Set | `set((1, 2, 2, 3))` | Removes duplicates |
| Dictionary → List | `list(d)` | Uses keys |
| Dictionary → Tuple | `tuple(d)` | Uses keys |
| Dictionary → Set | `set(d)` | Uses keys |
| Dictionary → Values | `list(d.values())` | Explicitly selects values |
| Dictionary → Items | `list(d.items())` | Produces key-value pairs |
| Pairs → Dictionary | `dict(pairs)` | Each element must contain two values |

## Key Takeaways

1. Use `list()` to create a list from another iterable.
2. Use `tuple()` to create a tuple from another iterable.
3. Use `set()` to create a set and remove duplicates.
4. Direct conversion of a dictionary to a list, tuple, or set uses its keys.
5. Use `values()` when you want dictionary values.
6. Use `items()` when you want dictionary key-value pairs.
7. Use `dict()` with two-item pairs to create a dictionary.
8. Sets do not guarantee element order.
