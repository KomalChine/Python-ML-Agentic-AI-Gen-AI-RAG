Object Oriented Programming (OOP)

Class

```python
# Defining a simple Class
class Car:
    # A class attribute (shared by all instances)
    wheels = 4

    # A method (behavior)
    def start_engine(self):
        print(f"The engine of {self.brand} {self.model} is now running!")

    def drive(self, speed):
        print(f"The {self.color} {self.model} is driving at {speed} mph.")
```

Object

```python
#Creating Concrete Instances (Objects)
car1 = Car()
car2 = Car()
car3 = Car()
```

```python
# Setting specific states (Attributes) for each Object
car1.brand = "Tesla"
car1.model = "Model 3"
car1.color = "Red"

car2.brand = "Toyota"
car2.model = "Corolla"
car2.color = "Blue"

car3.brand = "Ford"
car3.model = "Mustang"
car3.color = "Black"
```

```python
# Interacting with the Objects
print(f"Car 1: {car1.brand} {car1.model} ({car1.color})")
# Output: Car 1: Tesla Model 3 (Red)

print(f"Car 2: {car2.brand} {car2.model} ({car2.color})")
# Output: Car 2: Toyota Corolla (Blue)

# Executing methods on specific objects
car1.start_engine()
# Output: The engine of Tesla Model 3 is now running!

car3.drive(75)
# Output: The Black Mustang is driving at 75 mph.
```

```python
print(id(car1))  # Unique memory address integer 
print(id(car2))  # Different memory address integer

print(car1 is car2)  # Output: False (They are two separate objects)
```

```python
# "Everything is an Object": In Python, strings, numbers, lists, functions, and even classes themselves are objects under the hood!

x = 5
print(type(x))  # Output: <class 'int'> -> Integers are objects of the 'int' class!
```
