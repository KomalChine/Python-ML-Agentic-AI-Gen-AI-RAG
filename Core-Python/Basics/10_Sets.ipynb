# 10_Sets

A **set** is an unordered collection of unique, hashable objects.

Sets are useful when you need to:
- store unique values
- remove duplicates
- test membership
- perform mathematical set operations
- compare relationships between collections


## 1. Introduction to Sets

### Characteristics of Sets

| Characteristic | Meaning |
|---|---|
| **Mutable** | The set can be changed after creation |
| **Unordered** | Elements do not have a fixed positional order |
| **Unique** | Duplicate values are removed |
| **Unindexed** | You cannot access an element using `set[0]` |
| **Heterogeneous** | Different hashable data types can be stored together |
| **Hashable elements** | Each element must be hashable |


## 2. Creating Sets

### Set Literal

Sets are commonly created using curly braces `{}` with comma-separated values.


````python
fruits = {"apple", "banana", "cherry"}

print(fruits)
print(type(fruits))
````

### Empty Set

An empty set must be created using `set()`.

```python
empty_set = set()
```

Important:

```python
{}
```

creates an **empty dictionary**, not an empty set.


````python
empty_set = set()
empty_dict = {}

print(empty_set)
print(type(empty_set))

print(empty_dict)
print(type(empty_dict))
````

### Creating a Set from an Iterable

`set()` can create a set from another iterable.


````python
numbers = set([1, 2, 2, 3, 4])

print(numbers)
````

## 3. Unique Elements

One of the most common uses of a set is removing duplicate values from a collection.


````python
my_list = [1, 2, 2, 3, 4, 4, 4]

unique_set = set(my_list)

print("Original list:", my_list)
print("Unique values:", unique_set)
````

### `True`, `1`, `False`, and `0`

In Python:

```python
True == 1
False == 0
```

Therefore, a set treats these equivalent values as the same set element.


````python
print({True, 1})
print({False, 0})
````

## 4. Set Elements & Hashability

Set elements must be **hashable**.

Common immutable/hashable values such as numbers, strings, and tuples can be set elements.

Lists, dictionaries, and sets are mutable and cannot be direct elements of a set.


````python
my_set = {1, "Python", (2, 3)}

print(my_set)
````

````python
my_set = {1, 2}

try:
    my_set.add([3, 4])
except TypeError as error:
    print("TypeError:", error)
````

For Basic Python, remember:

> **Set elements must be hashable.**

You do not need to learn hashing algorithms or hash-table implementation details yet.


## 5. Adding & Updating

The main methods for adding elements are:

```text
add()
update()
```


### `add()`

`add(value)` adds **one element** to the set.

- It modifies the original set.
- If the value already exists, the set remains unchanged.
- It returns `None`.


````python
fruits = {"apple", "banana"}

result = fruits.add("orange")

print(fruits)
print("Return value:", result)
````

````python
fruits = {"apple", "banana"}

fruits.add("apple")

print(fruits)
````

### `update()`

`update()` adds elements from one or more iterables to the set.

It modifies the original set and returns `None`.


````python
fruits = {"apple", "banana"}

result = fruits.update(["orange", "mango"])

print(fruits)
print("Return value:", result)
````

````python
numbers = {1, 2}

numbers.update([3, 4], (5, 6))

print(numbers)
````

Because `update()` accepts an iterable, a string contributes its individual characters.


````python
letters = {"a"}

letters.update("cat")

print(letters)
````

## 6. Removing Elements

Common ways to remove elements are:

```text
remove()
discard()
pop()
clear()
```


### `remove()`

`remove(value)` removes the specified element.

If the element is not present, `remove()` raises `KeyError`.


````python
fruits = {"apple", "banana", "orange"}

fruits.remove("banana")

print(fruits)
````

````python
fruits = {"apple", "banana"}

try:
    fruits.remove("mango")
except KeyError as error:
    print("KeyError:", error)
````

### `discard()`

`discard(value)` removes the specified element if it is present.

If the element is not present, **no error is raised**.


````python
fruits = {"apple", "banana"}

fruits.discard("banana")
fruits.discard("mango")

print(fruits)
````

### `remove()` vs `discard()`

| Method | If element exists | If element is missing |
|---|---|---|
| `remove(x)` | Removes it | Raises `KeyError` |
| `discard(x)` | Removes it | Does nothing |


### `pop()`

`pop()` removes and returns an **arbitrary element** from a set.

Because sets are unordered, you cannot choose the element by position.


````python
fruits = {"apple", "banana", "orange"}

removed = fruits.pop()

print("Removed:", removed)
print("Remaining:", fruits)
````

### `clear()`

`clear()` removes all elements from the set.


````python
fruits = {"apple", "banana", "orange"}

result = fruits.clear()

print(fruits)
print("Return value:", result)
````

## 7. Copying Sets

`copy()` creates a separate set containing the same elements.

For this Basic course, remember:

```python
new_set = old_set.copy()
```


````python
set_a = {1, 2, 3}
set_b = set_a.copy()

set_b.add(100)

print("set_a:", set_a)
print("set_b:", set_b)
````

## 8. Set Membership

Sets are especially useful for fast membership testing.

Use:

```text
in
not in
```


````python
fruits = {"apple", "banana", "cherry"}

print("apple" in fruits)
print("mango" in fruits)
print("mango" not in fruits)
````

## 9. Set Operations

The four core set operations are:

1. **Union** — elements in either set
2. **Intersection** — elements common to both sets
3. **Difference** — elements in the first set but not the second
4. **Symmetric Difference** — elements in either set, but not both

For the examples below:

```python
A = {1, 2, 3}
B = {3, 4, 5}
```


### Union

Union combines all unique elements from both sets.

Operator:

```python
A | B
```

Method:

```python
A.union(B)
```


````python
A = {1, 2, 3}
B = {3, 4, 5}

print(A | B)
print(A.union(B))
````

### Intersection

Intersection returns elements that are common to both sets.

Operator:

```python
A & B
```

Method:

```python
A.intersection(B)
```


````python
A = {1, 2, 3}
B = {3, 4, 5}

print(A & B)
print(A.intersection(B))
````

### Difference

Difference returns elements that are in the first set but not in the second.

```python
A - B
```

The order matters:

```text
A - B != B - A
```


````python
A = {1, 2, 3}
B = {3, 4, 5}

print("A - B:", A - B)
print("B - A:", B - A)

print("A.difference(B):", A.difference(B))
````

### Symmetric Difference

Symmetric difference returns elements that are in either set, but **not in both**.

Operator:

```python
A ^ B
```

Method:

```python
A.symmetric_difference(B)
```


````python
A = {1, 2, 3}
B = {3, 4, 5}

print(A ^ B)
print(A.symmetric_difference(B))
````

### Set Operations — Quick Reference

| Operation | Operator | Method | Meaning |
|---|---|---|---|
| Union | `\|` | `union()` | In either set |
| Intersection | `&` | `intersection()` | In both sets |
| Difference | `-` | `difference()` | In first, not second |
| Symmetric difference | `^` | `symmetric_difference()` | In either, but not both |


## 10. Set Relationships

Sets can be compared to determine whether one contains another or whether two sets have no common elements.


### Subset

A set `A` is a subset of `B` when every element of `A` is also in `B`.

Methods and operators:

```python
A.issubset(B)
A <= B
```


````python
A = {2, 3}
B = {1, 2, 3, 4}

print(A.issubset(B))
print(A <= B)
````

### Superset

A set `A` is a superset of `B` when `A` contains every element of `B`.

Methods and operators:

```python
A.issuperset(B)
A >= B
```


````python
A = {1, 2, 3, 4}
B = {2, 3}

print(A.issuperset(B))
print(A >= B)
````

### Disjoint Sets

Two sets are disjoint when they have **no elements in common**.

Use:

```python
A.isdisjoint(B)
```


````python
A = {1, 2, 3}
B = {4, 5, 6}
C = {3, 4, 5}

print(A.isdisjoint(B))
print(A.isdisjoint(C))
````

## Set Relationship Quick Reference

| Relationship | Method | Operator |
|---|---|---|
| Subset | `A.issubset(B)` | `A <= B` |
| Superset | `A.issuperset(B)` | `A >= B` |
| Disjoint | `A.isdisjoint(B)` | No dedicated operator |
