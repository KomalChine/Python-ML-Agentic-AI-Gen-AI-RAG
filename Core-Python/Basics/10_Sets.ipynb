Set Methods

Add Method

```python
fruits = {"apple", "banana"}
fruits.add("cherry")

print(fruits)  
# Output: {'apple', 'banana', 'cherry'}
```

```python
# Adding Duplicates (Ignored)
numbers = {1, 2, 3}
numbers.add(2)  # 2 is already present

print(numbers)  
# Output: {1, 2, 3}
```

```python
# Adding Heterogeneous Types (Must be Immutable)
mixed_set = {10, "Python"}

# Adding a float
mixed_set.add(3.14)

# Adding a tuple
mixed_set.add((1, 2))

print(mixed_set)  
# Output: {10, 'Python', 3.14, (1, 2)}
```

```python
# Trying to Add a Mutable Object (Fails)
my_set = {1, 2}
my_set.add([3, 4])  

# ❌ TypeError: unhashable type: 'list'
```

Clear Method

```python
fruits = {"apple", "banana", "cherry"}
fruits.clear()

print(fruits)      # Output: set()  (Note: displays as set(), not {})
print(len(fruits)) # Output: 0
```

Copy Method

```python
set_a = {1, 2, 3}
set_b = set_a.copy()

# Add an element to set_b only
set_b.add(100)

print("Original set_a:", set_a)  # Output: {1, 2, 3}
print("Copied set_b:  ", set_b)  # Output: {1, 2, 3, 100}
```

Discard Method

```python
# Removing an Existing Element
fruits = {"apple", "banana", "cherry"}
fruits.discard("banana")

print(fruits)  
# Output: {'apple', 'cherry'}
```

```python
# Trying to Remove a Non-Existent Element
numbers = {1, 2, 3}

# 99 is not in the set, but discard handles it safely without crashing!
numbers.discard(99)  

print(numbers)  
# Output: {1, 2, 3}
```

Pop Method

```python
fruits = {"apple", "banana", "cherry"}

# Removes and returns one item
popped_fruit = fruits.pop()

print("Popped item:", popped_fruit) 
print("Remaining set:", fruits)

# Output (exact popped item may vary):
# Popped item: apple
# Remaining set: {'banana', 'cherry'}
```

```python
# Popping from an Empty Set (Fails)
empty_set = set()

empty_set.pop()  
# ❌ KeyError: 'pop from an empty set'
```

Remove Method

```python
languages = {"Python", "Java", "C++"}
languages.remove("Java")

print(languages)  
# Output: {'Python', 'C++'}
```

```python
# Attempting to Remove a Missing Element
numbers = {1, 2, 3}

# Calling .remove() on an element not in the set causes a crash
numbers.remove(10)  

# ❌ KeyError: 10
```

Update Method

```python
fruits = {"apple", "banana"}
more_fruits = {"cherry", "mango"}

fruits.update(more_fruits)

print(fruits)  
# Output: {'apple', 'banana', 'cherry', 'mango'}
```

```python
# Updating with Different Iterables (Lists, Tuples)
numbers = {1, 2}

# Passing a list and a tuple at the same time
numbers.update([3, 4], (5, 6))

print(numbers)  
# Output: {1, 2, 3, 4, 5, 6}
```

```python
# Updating with a String
letters = {"a", "b"}
letters.update("cat")

print(letters)  
# Output: {'a', 'b', 'c', 't'}
```

Set Operations

Union

```python
A = {1, 2, 3}
B = {3, 4, 5}

# Using the | operator
result1 = A | B
print(result1)  # Output: {1, 2, 3, 4, 5}

# Using the .union() method
result2 = A.union(B)
print(result2)  # Output: {1, 2, 3, 4, 5}
```

```python
# Union with Multiple Collections
set_a = {1, 2}
set_b = {2, 3}
set_c = {3, 4}

# Combines all three sets into one
result = set_a.union(set_b, set_c)

print(result)  # Output: {1, 2, 3, 4}
```

```python
# Method vs. Operator Argument Types
numbers = {1, 2, 3}
my_list = [3, 4, 5]

# Method works with a list directly
print(numbers.union(my_list))  # Output: {1, 2, 3, 4, 5}

# Operator requires both to be sets
# print(numbers | my_list)  # ❌ TypeError: unsupported operand type(s) for |: 'set' and 'list'
```

Intersection

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Using the & operator
result1 = A & B
print(result1)  # Output: {3, 4}

# Using the .intersection() method
result2 = A.intersection(B)
print(result2)  # Output: {3, 4}
```

```python
# Intersection across multiple collection
set_a = {1, 2, 3, 4}
set_b = {2, 3, 5}
set_c = {3, 6, 7}

# Only '3' is present in ALL three sets
result = set_a.intersection(set_b, set_c)

print(result)  # Output: {3}
```

```python
# Method vs. Operator Argument Types
numbers = {10, 20, 30}
my_list = [20, 30, 40]

# Method works with a list directly
print(numbers.intersection(my_list))  # Output: {20, 30}

# Operator requires both operands to be sets
# print(numbers & my_list)  # ❌ TypeError: unsupported operand type(s) for &: 'set' and 'list'
```

Difference

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

# Elements in A that are NOT in B
result1 = A.difference(B)
print(result1)  # Output: {1, 2, 3}

# Elements in B that are NOT in A (Order matters!)
result2 = B - A
print(result2)  # Output: {6, 7}
```

```python
# Difference across multiple collection
set_a = {1, 2, 3, 4, 5}
set_b = {2, 3}
set_c = {4}

# Subtracts elements from set_b AND set_c from set_a
result = set_a.difference(set_b, set_c)
result1 =  set_a - set_b - set_c

print(result)  # Output: {1, 5}
print(result1) # Output: {1, 5}
```

```python
# Method vs. Operator Argument Types
numbers = {10, 20, 30, 40}
my_list = [20, 30]

# Using method with a list (Works!)
print(numbers.difference(my_list))  # Output: {10, 40}

# Using operator with a list (Fails!)
# print(numbers - my_list)  # ❌ TypeError: unsupported operand type(s) for -: 'set' and 'list'
```

Symmetric_Difference

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Common elements are 3 and 4 (these will be excluded)

# Using the ^ operator
result1 = A ^ B
print(result1)  # Output: {1, 2, 5, 6}

# Using the .symmetric_difference() method
result2 = A.symmetric_difference(B)
print(result2)  # Output: {1, 2, 5, 6}
```

```python
# Method vs. Operator Argument Types
numbers = {1, 2, 3, 4}
my_list = [3, 4, 5, 6]

# Method works directly with a list
print(numbers.symmetric_difference(my_list))  # Output: {1, 2, 5, 6}

# Operator requires both to be sets
# print(numbers ^ my_list)  # ❌ TypeError: unsupported operand type(s) for ^: 'set' and 'list'
```

```python
# Symmetric_Difference across multiple collection
# The ^ operator allows you to chain as many sets together as you like
A = {1, 2, 3}
B = {2, 3, 4}
C = {3, 4, 5}

# Chaining multiple sets
result = A ^ B ^ C

print(result)  # Output: {1, 3, 5}
```

```python
# ❌ THIS FAILS:
# A.symmetric_difference(B, C)  
# TypeError: symmetric_difference() takes exactly one argument (2 given)

# ✅ SOLUTION: Chain the method calls
result = A.symmetric_difference(B).symmetric_difference(C)

print(result)  # Output: {1, 3, 5}
```

```python
# Mathematical Rule: An element will appear in the final result if and only if it appears in an odd number of sets (1 set, 3 sets, 5 sets, etc.).
# Notice how 3 appears in all 3 sets above, so it turns up in the final result {1, 3, 5}
```

```python
# 1. difference_update
A = {1, 2, 3, 4}
A.difference_update({3, 4, 5})
print(A)  # Output: {1, 2}

# 2. intersection_update
B = {1, 2, 3, 4}
B.intersection_update({3, 4, 5})
print(B)  # Output: {3, 4}

# 3. symmetric_difference_update
C = {1, 2, 3, 4}
C.symmetric_difference_update({3, 4, 5})
print(C)  # Output: {1, 2, 5}
```

Issuperset

```python
A = {1, 2, 3, 4, 5}
B = {2, 3}

print(A.issuperset(B))  # Output: True (A contains all of B)
print(B.issuperset(A))  # Output: False (B is missing 1, 4, and 5)
```

```python
# Equal sets
A = {1, 2, 3}
B = {1, 2, 3}

print(A.issuperset(B))  # Output: True
print(A > B)            # Output: False (Not a STRICT superset because they are equal)
```

```python
# Method vs. Operator
A = {1, 2, 3, 4}
my_list = [2, 3]

# Method works directly with a list
print(A.issuperset(my_list))  # Output: True

# Operator requires both to be sets
# print(A >= my_list)  # ❌ TypeError
```

Issubset

```python
A = {2, 3}
B = {1, 2, 3, 4}

print(A.issubset(B))  # Output: True  (All elements of A are inside B)
print(B.issubset(A))  # Output: False (B has extra items: 1 and 4)
```

Isdisjoint

```python
A = {1, 2, 3}
B = {4, 5, 6}
C = {3, 4, 5}

print(A.isdisjoint(B))  # Output: True  (No overlap at all)
print(A.isdisjoint(C))  # Output: False (They share the element 3)
```

Frozen Set

```python
# Creating a frozenset
# Use the frozenset() constructor with any iterable (list, tuple, standard set, or string).

# From a list
numbers = frozenset([1, 2, 2, 3, 4])
print(numbers)  
# Output: frozenset({1, 2, 3, 4})  (Duplicates removed)

# Attempting to modify raises an AttributeError
# numbers.add(5)  --> AttributeError: 'frozenset' object has no attribute 'add'
```

```python
# Supported Operations
# frozenset supports all non-modifying mathematical set operations:

fs1 = frozenset([1, 2, 3])
fs2 = frozenset([3, 4, 5])

# Union
print(fs1 | fs2)  # Output: frozenset({1, 2, 3, 4, 5})

# Intersection
print(fs1 & fs2)  # Output: frozenset({3})

# Difference
print(fs1 - fs2)  # Output: frozenset({1, 2})
```

```python
# Storing permissions for specific combinations of roles
permissions = {
    frozenset(["admin", "editor"]): "Full Access",
    frozenset(["guest"]): "Read Only"
}

print(permissions[frozenset(["admin", "editor"])])  
# Output: Full Access
```

```python
# Creating Nested Sets (Set of Sets)
# Standard sets cannot contain other standard sets. Use frozenset for outer/inner elements:

# A set containing sets
nested_set = {frozenset([1, 2]), frozenset([3, 4])}
print(nested_set)  
# Output: {frozenset({1, 2}), frozenset({3, 4})}
```
