# 02_Variables

This notebook introduces Python variables, assignment, naming conventions, object identity, types, reassignment, and expressions.

## 1. What is a Variable?

A **variable** is a name that refers to an object (value) in Python. Variables let us work with data using meaningful names.

````python
age = 25
print(age)
````

Here, `age` is the variable name and `25` is the integer object to which it refers.

## 2. Variable Assignment

Python uses the **assignment operator `=`** to bind a name to an object.

```python
variable_name = value
```

````python
name = "Komal"
age = 25
height = 5.6
is_student = True

print(name)
print(age)
print(height)
print(is_student)
````

## 3. Dynamic Typing

Python is **dynamically typed**. A name does not have a permanently fixed type; it can refer to objects of different types at different times.

````python
value = 10
print(value, type(value))

value = "Python"
print(value, type(value))

value = 3.14
print(value, type(value))
````

## 4. Variable Naming Rules

- Names may contain letters, digits, and underscores.
- A name cannot start with a digit.
- Names are case-sensitive.
- Spaces are not allowed.
- Python keywords such as `if`, `for`, `class`, and `def` cannot be used as ordinary variable names.
- Prefer meaningful names.

````python
student_name = "Komal"
age2 = 25
_private_value = 100
print(student_name, age2, _private_value)
````

## 5. Valid and Invalid Variable Names

### Valid
```python
name
student_name
age2
_count
total_marks
```

### Invalid
```text
2age          # starts with a digit
student name  # contains a space
first-name    # hyphen is not valid in an identifier
class         # Python keyword
```

## 6. Good Variable Naming Practices

- Use descriptive names.
- Prefer `snake_case` for ordinary variables.
- Avoid unnecessary abbreviations.
- Use lowercase for ordinary variables.
- Use uppercase by convention for constants, such as `MAX_SIZE`.
- Avoid shadowing useful built-ins such as `list`, `str`, and `sum`.

````python
student_name = "Komal"
total_marks = 450
MAX_ATTEMPTS = 3
print(student_name, total_marks, MAX_ATTEMPTS)
````

## 7. Multiple Assignment

Python allows several names to be assigned in one statement.

````python
name, age, city = "Komal", 25, "Hyderabad"
print(name, age, city)

x = y = z = 0
print(x, y, z)
````

## 8. Swapping Variables

Python can swap values without a temporary variable.

````python
a = 10
b = 20
print("Before:", a, b)
a, b = b, a
print("After:", a, b)
````

## 9. Variables and Objects

In Python, variables are best understood as **names bound to objects** rather than boxes that permanently contain values.

````python
x = 100
y = x
print(x)
print(y)
````

After `y = x`, both names refer to the same integer object in this example. Assignment binds names; it does not mean that every assignment creates a separate copy of the object.

## 10. Object Identity and `id()`

Every Python object has an identity. The built-in `id()` function returns an integer identifying an object during its lifetime.

````python
x = [1, 2, 3]
y = x
print("id(x):", id(x))
print("id(y):", id(y))
print("Same object:", x is y)
````

Because `y = x`, both names refer to the same list, so `x is y` is `True`. The actual numeric ID is implementation-dependent.

## 11. Checking Type with `type()`

The built-in `type()` function returns the type (class) of an object.

````python
integer_value = 10
float_value = 3.14
text_value = "Python"
list_value = [1, 2, 3]

print(type(integer_value))
print(type(float_value))
print(type(text_value))
print(type(list_value))
````

Typical output:

```text
<class 'int'>
<class 'float'>
<class 'str'>
<class 'list'>
```

## 12. Reassigning Variables

A name can be reassigned to another object at any time.

````python
score = 50
print(score)
score = 75
print(score)
score = 95
print(score)
````

Reassignment changes what the name refers to. If an old object is no longer referenced, Python can eventually reclaim its memory.

## 13. Using Variables in Expressions

Variables can be used in arithmetic, comparisons, function calls, and many other expressions.

````python
length = 10
width = 5
area = length * width
perimeter = 2 * (length + width)
print("Area:", area)
print("Perimeter:", perimeter)
````

````python
marks1 = 85
marks2 = 90
marks3 = 95
total = marks1 + marks2 + marks3
average = total / 3
passed = average >= 40
print("Total:", total)
print("Average:", average)
print("Passed:", passed)
````
