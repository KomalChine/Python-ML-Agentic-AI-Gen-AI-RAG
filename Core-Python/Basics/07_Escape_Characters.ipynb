Escape Characters

New Line - \n

```python
print("Hello\nWorld") # Moves the cursor to next line
```

```python
print("Python\nis\ninteresting")
```

Tab - \t

```python
print("Hello\tWorld") # Adds a tab space
```

```python
print("Name\tAge") # Usweful for formatting
print("Komal\t20")
```

Backslash - \\

```python
# If you want to actually print a backslash, use \\.
print("C:\\Users\\Komal")
```

Single Quote - \'

```python
# Useful when your string is surrounded by single quotes.
print('I don't know') 
# This will cause error
```

```python
print('I don\'t know')
```

Double Quote - \"

```python
# Useful when your string is surrounded by double quotes.
print("He said \"Hello\"")
```

Carriage Return - \r

```python
# Moves the cursor to the beginning of the current line.
print("Hello\rWorld")
# The exact displayed result can depend on the environment, but conceptually \r moves back to the start of the line.
```

Backspace - \b

```python
# Moves one position backward.
print("Hell\bo")
```
