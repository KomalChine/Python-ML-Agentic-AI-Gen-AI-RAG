# 07_Escape_Characters

## Escape Characters & Escape Sequences

An **escape character** is a backslash (`\`) used inside a string to introduce a special meaning.

An **escape sequence** is the combination of the backslash and the character that follows it.

Examples:

```text
\n   → new line
\t   → tab
\\   → literal backslash
\'   → single quote
\"   → double quote
```

The backslash itself is the character that starts an escape sequence.


## 2. New Line (`\n`)

The `\n` escape sequence inserts a new line inside a string.


````python
print("Hello\nWorld")
print("Python\nis\ninteresting")
````

## 3. Tab (`\t`)

The `\t` escape sequence inserts a horizontal tab.

It can be useful for simple text formatting.


````python
print("Hello\tWorld")
print("Name\tAge")
print("Komal\t20")
print("Rahul\t25")
````

## 4. Backslash (`\\`)

A single backslash begins an escape sequence. If you want to include a literal backslash in a normal string, use `\\`.


````python
print("C:\\Users\\Komal")
print("This is a backslash: \\")
````

## 5. Single Quote (`\'`)

A single quote can be escaped with `\'`.

### Problem

The apostrophe in `don't` can accidentally close a string when the string itself uses single quotes:

```python
# This is invalid:
print('I don't know')
```

The second apostrophe closes the string too early, causing a `SyntaxError`.

### Solution

Escape the apostrophe:


````python
print('I don\'t know')
````

You can also avoid the escape by using double quotes:

```python
print("I don't know")
```


````python
print("I don't know")
````

## 6. Double Quote (`\"`)

A double quote can be escaped with `\"` when the surrounding string also uses double quotes.


````python
print("He said \"Hello\"")
````

If you use single quotes around the string, escaping is often unnecessary:

```python
print('He said "Hello"')
```


````python
print('He said "Hello"')
````

## 7. Carriage Return (`\r`)

The `\r` escape sequence represents a **carriage return**. It moves the cursor to the beginning of the current line.

Its visible effect can depend on the terminal or notebook output environment, so it is less commonly used by beginners.


````python
print("Hello\rWorld")
````

## 8. Backspace (`\b`)

The `\b` escape sequence represents a **backspace** character.

Its visible effect can depend on the output environment.


````python
print("Hell\bo")
````

## 9. Raw Strings (`r"..."`)

A raw string is created by placing `r` before the opening quote.

In a raw string, backslashes are generally treated literally instead of starting the usual escape sequences.

Raw strings are especially useful for Windows-style paths and, later, regular expressions.


````python
path = r"C:\Users\Komal\Documents"
print(path)

normal_path = "C:\\Users\\Komal\\Documents"
print(normal_path)
````

### Comparing Normal and Raw Strings

With a normal string, backslashes may need to be escaped:

```python
"C:\\Users\\Komal"
```

With a raw string, the path can be written more naturally:

```python
r"C:\Users\Komal"
```

Raw strings do not make every possible backslash combination valid, but they are a convenient way to write strings containing many literal backslashes.


## 10. Escape Sequence Summary

| Escape sequence | Meaning | Priority |
|---|---|---|
| `\n` | New line | ⭐ Essential |
| `\t` | Tab | ⭐ Essential |
| `\\` | Backslash | ⭐ Essential |
| `\'` | Single quote | ⭐ Essential |
| `\"` | Double quote | ⭐ Essential |
| `\r` | Carriage return | Optional |
| `\b` | Backspace | Optional |

### Quick examples

```python
print("Hello\nWorld")
print("Name\tAge")
print("C:\\Users\\Komal")
print('I don\'t know')
print("He said \"Hello\"")
```

### Key idea

The **essential escape sequences** for beginners are:

```text
\n   \t   \\   \'   \"
```

The sequences `\r` and `\b` are useful to recognize, but they are less commonly needed in everyday beginner Python code.
