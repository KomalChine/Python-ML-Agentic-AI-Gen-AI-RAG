# 02 — Higher-Order Functions Review

This notebook reviews **higher-order functions** as a bridge from functions-as-objects to decorators.

> A higher-order function is a function that **takes another function as an argument, returns a function, or both**.

The goal is practical fluency with the pattern before learning actual decorator syntax.

## 2. What Is a Higher-Order Function?

A function is **higher-order** when it operates on other functions.

Two common forms are:

1. It **takes a function as an argument**.
2. It **returns a function**.

It can also do both.

````python
def execute(func):
    return func()

def create_function():
    def greet():
        print("Hello")
    return greet
````

### A function that takes another function

````python
def greet():
    print("Hello")

def execute(func):
    func()

execute(greet)
````

### A function that returns another function

````python
def create_greeting():
    def greet():
        print("Hello")
    return greet

message = create_greeting()
message()
````

## 3. Functions as Arguments

A function can be passed to another function just like other Python objects.

Remember the distinction:

```python
execute(greet)    # passes the function
execute(greet())  # calls greet first, then passes its result
```

````python
def greet():
    print("Hello")

def execute(func):
    func()

execute(greet)
````

````python
def add(a, b):
    return a + b

def calculate(func, a, b):
    return func(a, b)

result = calculate(add, 10, 20)
print(result)
````

The important pattern is:

```text
calculate()
   ↓
receives a function
   ↓
calls that function
   ↓
returns its result
```

## 4. Calling a Function Passed as an Argument

The same higher-order function can work with completely different operations.

````python
def square(x):
    return x ** 2

def apply_function(func, value):
    return func(value)

print(apply_function(square, 5))
````

````python
def double(x):
    return x * 2

print(apply_function(double, 5))
````

`apply_function()` does not need to know whether the supplied function squares, doubles, or performs some other operation.

## 5. Functions Returning Functions

A higher-order function can create and return another function.

````python
def create_greeting():
    def greet():
        print("Hello!")

    return greet

message = create_greeting()
message()
````

The key concept is:

```text
create_greeting()
       ↓
returns a function
       ↓
message
       ↓
message()
```

Closures are covered separately in the Functions and Scope section, so we will not go deeply into closure mechanics here.

### Returning different functions

A function can choose which function to return.

````python
def get_operation(operation):
    if operation == "add":
        def add(a, b):
            return a + b
        return add

    elif operation == "multiply":
        def multiply(a, b):
            return a * b
        return multiply

operation = get_operation("add")
print(operation(10, 5))
````

## 6. Combining Arguments and Returned Functions

A function can both **accept and return functions**.

````python
def create_multiplier(number):
    def multiply(value):
        return value * number

    return multiply

double = create_multiplier(2)
triple = create_multiplier(3)

print(double(10))
print(triple(10))
````

This is a useful example of the flexibility of functions as objects.

The detailed theory of closures belongs to the earlier **Nested Functions and Closures** notebook.

## 7. Higher-Order Functions with Conditions

Higher-order functions can choose whether to apply a supplied function.

````python
def apply_if(condition, func, value):
    if condition:
        return func(value)

    return value

def square(x):
    return x ** 2

print(apply_if(True, square, 5))
print(apply_if(False, square, 5))
````

This demonstrates how higher-order functions can build reusable behavior.

## 8. Higher-Order Functions with Loops

A higher-order function can receive an operation and apply it repeatedly.

````python
def apply_to_all(func, values):
    results = []

    for value in values:
        results.append(func(value))

    return results

def square(x):
    return x ** 2

numbers = [1, 2, 3, 4, 5]

result = apply_to_all(square, numbers)
print(result)
````

This naturally leads into Python's `map()` function.

## 9. Built-in Higher-Order Functions

Important built-in or standard-library tools to recognize include:

```text
map()

```

Other functions can also accept callable behavior, such as:

```text
sorted()

```

For this review, the main focus is `map()`, `filter()`, and `reduce()`.

## 10. `map()`

Basic syntax:

```python
map(function, iterable)
```

`map()` applies a function to each item.

````python
def square(x):
    return x ** 2

numbers = [1, 2, 3, 4]

result = map(square, numbers)

print(list(result))
````

### `map()` with a lambda

This connects to the earlier Lambda notebook.

````python
numbers = [1, 2, 3, 4]

result = map(lambda x: x ** 2, numbers)

print(list(result))
````

## 11. `filter()`

Basic syntax:

```python
filter(function, iterable)
```

`filter()` keeps values for which the supplied function returns `True`.

````python
def is_even(number):
    return number % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]

result = filter(is_even, numbers)

print(list(result))
````

### `filter()` with a lambda

````python
numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))
````

## 12. `reduce()`

`reduce()` comes from the `functools` module.

```python
from functools import reduce
```

It repeatedly combines values using a supplied function.

````python
from functools import reduce

def add(a, b):
    return a + b

numbers = [1, 2, 3, 4]

result = reduce(add, numbers)

print(result)
````

The calculation proceeds conceptually like this:

```text
1 + 2 → 3
3 + 3 → 6
6 + 4 → 10
```

````python
from functools import reduce

numbers = [1, 2, 3, 4]

result = reduce(lambda a, b: a * b, numbers)

print(result)
````

Keep `reduce()` at an introductory level here. The goal is to recognize it as another example of a function operating with a supplied function.

## 13. Practical Examples

### Example 1 — Convert temperatures

````python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

temperatures = [0, 10, 20, 30]

result = map(celsius_to_fahrenheit, temperatures)

print(list(result))
````

### Example 2 — Filter positive numbers

````python
def is_positive(number):
    return number > 0

numbers = [-5, 3, -2, 8, 0, 4]

result = filter(is_positive, numbers)

print(list(result))
````

### Example 3 — Calculate a total

````python
from functools import reduce

prices = [100, 200, 50, 150]

total = reduce(lambda a, b: a + b, prices)

print(total)
````

### Example 4 — Custom higher-order function

````python
def process(values, operation):
    results = []

    for value in values:
        results.append(operation(value))

    return results

def double(x):
    return x * 2

numbers = [1, 2, 3, 4]

print(process(numbers, double))
````

## 14. Higher-Order Functions and Decorators

This is the most important connection in this notebook.

Do **not** introduce the `@` decorator syntax yet. First understand the mechanism.

Suppose we have:

````python
def greet():
    print("Hello")
````

Now create a higher-order function that receives `greet`, creates another function, and returns it.

````python
def wrapper(func):
    def inner():
        print("Before")
        func()
        print("After")

    return inner

new_greet = wrapper(greet)

new_greet()
````

Output:

```text
Before
Hello
After
```

The mechanism is:

```text
wrapper()
   ↓
takes a function
   ↓
creates another function
   ↓
returns that function
```

That is the fundamental mechanism behind decorators.

The next notebook can introduce the actual decorator pattern and `@` syntax.

## 15. Common Mistakes

### Mistake 1 — Calling the function instead of passing it

Wrong:

```python
apply_function(square(), 5)
```

Correct:

```python
apply_function(square, 5)
```

````python
def square(x):
    return x ** 2

def apply_function(func, value):
    return func(value)

# Correct
print(apply_function(square, 5))
````

### Mistake 2 — Forgetting to return the inner function

Wrong:

```python
def create_function():
    def greet():
        print("Hello")
    greet
```

Correct:

```python
def create_function():
    def greet():
        print("Hello")
    return greet
```

### Mistake 3 — Confusing `map()` with a list

`map()` returns a map object.

```python
result = map(square, [1, 2, 3])
print(result)
```

To materialize the values:

```python
print(list(result))
```

````python
def square(x):
    return x ** 2

result = map(square, [1, 2, 3])

print(result)
print(list(result))
````

### Mistake 4 — Forgetting that `map()` and `filter()` are lazy

Once consumed, the result cannot be replayed from the same object.

````python
result = map(square, [1, 2, 3])

print(list(result))
print(list(result))
````

### Mistake 5 — Using `reduce()` when `sum()` is clearer

This works:

```python
reduce(lambda a, b: a + b, numbers)
```

But for simple addition:

```python
sum(numbers)
```

is generally clearer.

**Practical lesson:** knowing a higher-order function exists does not mean it should always be used.

## 16. Summary

### Key Takeaways

- Higher-order functions work with other functions.
- A function can be passed as an argument.
- A function can return another function.
- A function can both accept and return functions.
- `map()` transforms values using a function.
- `filter()` selects values using a function.
- `reduce()` combines values using a function.
- Higher-order functions enable reusable behavior.
- Decorators are built on these same ideas.
- A decorator typically receives a function, creates or wraps behavior, and returns a function.
