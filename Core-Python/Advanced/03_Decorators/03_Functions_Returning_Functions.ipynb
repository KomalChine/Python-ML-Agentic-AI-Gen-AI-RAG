## 1. Introduction

> A Python function can return another function just like it can return a number, string, list, or any other object.

Simple example:

````python
def create_greeting():
    def greet():
        print("Hello!")

    return greet

greeting = create_greeting()
greeting()
````

The important flow is:

```text
create_greeting()
       ↓
returns greet
       ↓
greeting
       ↓
greeting()
```

## 2. What Does It Mean to Return a Function?

Compare returning a value with returning a function.

````python
def get_number():
    return 10

result = get_number()
print(result)
print(type(result))
````

````python
def get_function():
    def greet():
        print("Hello")

    return greet

result = get_function()

print(type(result))
result()
````

In the second example, `result` contains a function object:

```text
<class 'function'>
```

The function is returned; it is not automatically executed.

## 3. Returning a Named Function

Keep the basic pattern simple.

````python
def create_message():
    def message():
        print("Welcome to Python")

    return message

my_message = create_message()
my_message()
````

You can also call the returned function immediately:

```python
create_message()()
```

The flow is:

```text
create_message()
        ↓
returns message function
        ↓
()
        ↓
calls the returned function
```

Although valid, storing the function in a variable is usually clearer while learning.

## 4. Calling the Returned Function

A returned function can be called multiple times.

````python
def create_greeting():
    def greet():
        print("Good morning!")

    return greet

greet_function = create_greeting()

greet_function()
greet_function()
````

The important distinction is:

> Returning a function does not execute it. It gives the caller the function object.

Compare:

```python
return greet
```

with:

```python
return greet()
```

### `return greet`

Returns the function itself.

### `return greet()`

Calls `greet()` and returns its result.

## 5. Returning Different Functions

A function can choose which function to return.

````python
def get_operation(operation):
    def add(a, b):
        return a + b

    def subtract(a, b):
        return a - b

    if operation == "add":
        return add

    if operation == "subtract":
        return subtract

operation = get_operation("add")
print(operation(10, 5))

operation = get_operation("subtract")
print(operation(10, 5))
````

The idea is:

```text
get_operation("add")
        ↓
returns add

get_operation("subtract")
        ↓
returns subtract
```

## 6. Functions with Parameters

The returned function can accept its own arguments.

````python
def create_multiplier():
    def multiply(number):
        return number * 2

    return multiply

double = create_multiplier()

print(double(5))
print(double(10))
````

The outer function creates the behavior, while the returned function receives the actual value.

## 7. Creating Configurable Functions

One of the most useful applications is creating functions with configurable behavior.

````python
def create_multiplier(multiplier):
    def multiply(number):
        return number * multiplier

    return multiply

double = create_multiplier(2)
triple = create_multiplier(3)
ten_times = create_multiplier(10)

print(double(5))
print(triple(5))
print(ten_times(5))
````

The outer function configures the returned function.

```text
create_multiplier(2)
        ↓
creates double()

create_multiplier(3)
        ↓
creates triple()
```

This pattern is often called a **function factory**.

## 8. Function Factories

A function factory is a function that creates and returns another function configured for a particular purpose.

````python
def create_power(power):
    def calculate(number):
        return number ** power

    return calculate

square = create_power(2)
cube = create_power(3)
fourth_power = create_power(4)

print(square(5))
print(cube(5))
print(fourth_power(2))
````

The same outer function creates different behaviors:

```text
create_power(2) → square
create_power(3) → cube
create_power(4) → fourth_power
```

## 9. Returning Functions with Behavior

Returned functions can provide reusable behavior.

````python
def create_logger(prefix):
    def log(message):
        print(prefix, message)

    return log

info = create_logger("[INFO]")
error = create_logger("[ERROR]")

info("Application started")
error("Something went wrong")
````

Output:

```text
[INFO] Application started
[ERROR] Something went wrong
```

This is a practical pattern and provides a natural bridge toward decorators.

## 10. Returning Inner Functions

Make the structure explicit.

````python
def outer():
    def inner():
        print("Inside inner")

    return inner

function = outer()
function()
````

The roles are:

```text
outer()
  │
  └── creates inner()
          │
          └── returns inner
```

The inner function can access names from the surrounding function. This behavior is related to **closures**.

Do not go deeply into closures here; that topic has already been covered in the Functions and Scope section. The goal here is to understand the function-returning pattern needed for decorators.

## 11. Practical Examples

### Example 1 — Greeting factory

````python
def create_greeting(greeting):
    def greet(name):
        print(greeting, name)

    return greet

morning = create_greeting("Good morning")
evening = create_greeting("Good evening")

morning("Alice")
evening("Bob")
````

### Example 2 — Tax calculator factory

````python
def create_tax_calculator(rate):
    def calculate(price):
        return price + (price * rate)

    return calculate

gst = create_tax_calculator(0.18)

print(gst(1000))
````

Output:

```text
1180.0
```

### Example 3 — Validator factory

````python
def create_validator(minimum):
    def validate(value):
        return value >= minimum

    return validate

is_adult = create_validator(18)

print(is_adult(20))
print(is_adult(15))
````

Output:

```text
True
False
```

### Example 4 — Prefix function

````python
def create_prefixer(prefix):
    def add_prefix(text):
        return prefix + text

    return add_prefix

error_message = create_prefixer("ERROR: ")

print(error_message("File not found"))
````

Output:

```text
ERROR: File not found
```

## 12. Common Mistakes

### Mistake 1 — Calling instead of returning

Wrong:

```python
def create_greeting():
    def greet():
        print("Hello")

    return greet()
```

This executes `greet()` immediately and returns its result.

Correct:

```python
return greet
```

### Mistake 2 — Forgetting the return

```python
def create_greeting():
    def greet():
        print("Hello")
```

Then:

```python
greeting = create_greeting()
print(greeting)
```

`greeting` becomes:

```text
None
```

because the outer function never returned `greet`.

### Mistake 3 — Confusing the outer and inner function

```python
def outer():
    def inner():
        print("Hello")

    return inner
```

This:

```python
outer()
```

returns the function.

This:

```python
outer()()
```

calls the returned function.

### Mistake 4 — Expecting the returned function to execute automatically

This:

```python
greet = create_greeting()
```

does not necessarily print anything.

You need:

```python
greet()
```

## 13. Connection to Decorators

This section prepares you for the next notebook.

Consider:

````python
def wrapper(func):
    def inner():
        print("Before")
        func()
        print("After")

    return inner
````

Notice what is happening:

```text
wrapper(func)
      ↓
receives a function
      ↓
creates another function
      ↓
returns that function
```

This combines the concepts from notebooks 01–03:

```text
Functions as Objects
        +
Higher-Order Functions
        +
Functions Returning Functions
        ↓
     Decorators
```

````python
def greet():
    print("Hello")

new_greet = wrapper(greet)

new_greet()
````

Output:

```text
Before
Hello
After
```

**Do not introduce `@wrapper` yet.** The next decorator-focused notebook should first connect nested functions, closures, and this wrapper pattern directly before introducing decorator syntax.

## 14. Summary

### Key Takeaways

- Functions can return other functions.
- Returning a function does not execute it.
- `return function_name` returns the function object.
- `return function_name()` executes the function first.
- Returned functions can accept their own arguments.
- An outer function can configure the behavior of an inner function.
- A function that creates and returns functions can be called a **function factory**.
- Inner functions can use values from their surrounding scope.
- Returning functions is a fundamental building block of decorators.
