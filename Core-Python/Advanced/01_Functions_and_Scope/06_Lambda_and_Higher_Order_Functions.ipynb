## 1. Introduction to Lambda Functions

A normal function can be defined with `def`:

```python
def square(x):
    return x ** 2
```

A small function can also be written as a lambda:

```python
square = lambda x: x ** 2
```

A **lambda is a small anonymous function written as a single expression**.


````python
def square(x):
    return x ** 2

print(square(5))
````

````python
square = lambda x: x ** 2

print(square(5))
````

Both examples calculate the square of a number.

Lambda functions are especially useful when a small function is needed temporarily, often as an argument to another function.


## 2. Lambda Syntax

The general form is:

```python
lambda arguments: expression
```

Examples:

```python
lambda x: x + 10
```

```python
lambda a, b: a + b
```

```python
lambda x: x ** 2
```

A lambda automatically returns the value of its expression.


````python
add_ten = lambda x: x + 10
add = lambda a, b: a + b
square = lambda x: x ** 2

print(add_ten(5))
print(add(10, 20))
print(square(6))
````

Compare:

```python
def square(x):
    return x ** 2
```

with:

```python
lambda x: x ** 2
```

The lambda version has no explicit `return` statement. The expression's value is returned automatically.


## 3. Lambda vs Normal Functions

| `def` | `lambda` |
|---|---|
| Can contain multiple statements | Contains a single expression |
| Can have a descriptive function name | Often used anonymously |
| Suitable for complex functions | Best for small operations |
| Uses `return` | Expression is returned automatically |

For example:


````python
def add(a, b):
    return a + b

add_lambda = lambda a, b: a + b

print(add(3, 4))
print(add_lambda(3, 4))
````

A lambda is not a replacement for every `def` function. It is most useful when the operation is short and clear.


## 4. Lambda with Multiple Arguments

A lambda can accept multiple arguments.


````python
multiply = lambda a, b: a * b

print(multiply(5, 4))
````

````python
calculate = lambda a, b, c: a + b * c

print(calculate(2, 3, 4))
````

The arguments are separated by commas just like parameters in a normal function.

```python
lambda a, b, c: expression
```


## 5. Lambda with Conditional Expressions

A lambda can contain a **conditional expression**.

```python
value_if_true if condition else value_if_false
```

For example:


````python
check = lambda x: "Even" if x % 2 == 0 else "Odd"

print(check(10))
print(check(7))
````

````python
maximum = lambda a, b: a if a > b else b

print(maximum(10, 25))
````

This is a **conditional expression**, not a normal multi-line `if/else` statement.

Keep the expression simple. If the logic becomes difficult to read, a normal `def` function is usually clearer.


## 6. Using Lambda with `sorted()`

A common use of lambda is supplying a sorting key.

Suppose we have students represented as tuples:


````python
students = [
    ("Alice", 85),
    ("Bob", 92),
    ("Charlie", 78)
]

students_sorted = sorted(
    students,
    key=lambda student: student[1]
)

print(students_sorted)
````

Here:

```python
key=lambda student: student[1]
```

tells `sorted()` to use the student's score as the sorting key.

We do not need to create a separate named function just to specify this simple key.


````python
students = [
    ("Alice", 85),
    ("Bob", 92),
    ("Charlie", 78)
]

sorted_students = sorted(
    students,
    key=lambda student: student[1],
    reverse=True
)

print(sorted_students)
````

### Sorting Dictionaries

Lambda is also useful when sorting dictionaries by a particular value.


````python
students = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 92},
    {"name": "Charlie", "score": 78}
]

result = sorted(
    students,
    key=lambda student: student["score"]
)

print(result)
````

## 7. Using Lambda with `map()`

`map()` applies a function to each item in an iterable.

Conceptually:

```text
map()
  ↓
takes a function
  ↓
applies it to every element
```


````python
numbers = [1, 2, 3, 4, 5]

squares = map(lambda x: x ** 2, numbers)

print(list(squares))
````

Output:

```text
[1, 4, 9, 16, 25]
```

The lambda is applied to every element.


````python
numbers = [10, 20, 30]

result = list(map(lambda x: x + 5, numbers))

print(result)
````

`map()` returns a map object, so `list()` is commonly used when we want to display or work with all the resulting values immediately.


## 8. Using Lambda with `filter()`

`filter()` keeps the elements for which a function produces a truthy result.

For example, select even numbers:


````python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = filter(
    lambda x: x % 2 == 0,
    numbers
)

print(list(even_numbers))
````

Output:

```text
[2, 4, 6]
```

The lambda acts as the condition used to decide which values are kept.


````python
names = ["Alice", "Bob", "Andrew", "Charlie"]

result = filter(
    lambda name: name.startswith("A"),
    names
)

print(list(result))
````

Here, only names beginning with `"A"` are kept.


## 9. Using Lambda with `reduce()`

`reduce()` repeatedly combines values to produce a single result.

It is available from the `functools` module:


````python
from functools import reduce
````

### Sum with `reduce()`



````python
numbers = [1, 2, 3, 4]

result = reduce(
    lambda a, b: a + b,
    numbers
)

print(result)
````

Output:

```text
10
```

The operation progresses conceptually like this:

```text
1, 2, 3, 4

1 + 2 → 3
3 + 3 → 6
6 + 4 → 10
```


### Multiplication with `reduce()`



````python
numbers = [1, 2, 3, 4]

result = reduce(
    lambda a, b: a * b,
    numbers
)

print(result)
````

Output:

```text
24
```

The key difference from `map()` and `filter()` is that `reduce()` combines values repeatedly until one accumulated result remains.


## 10. What Are Higher-Order Functions?

A **higher-order function** is a function that:

- takes another function as an argument,
- returns a function,
- or does both.

Python supports this because functions are **first-class objects**.

A simple example:


````python
def apply_function(function, value):
    return function(value)

def square(x):
    return x ** 2

print(apply_function(square, 5))
````

Output:

```text
25
```

Here, `square` is passed to `apply_function()` as an argument.

The function receives another function and uses it.


## 11. Functions as Arguments

Functions can be passed around just like other Python objects.


````python
def square(x):
    return x ** 2

def cube(x):
    return x ** 3

def calculate(function, value):
    return function(value)

print(calculate(square, 5))
print(calculate(cube, 5))
````

The same `calculate()` function can work with different operations because the operation is supplied as a function.

Functions can be:

- stored in variables
- passed as arguments
- returned from functions
- stored in collections


````python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

operations = [add, multiply]

for operation in operations:
    print(operation(5, 3))
````

## 12. Functions as Return Values

A function can also return another function.

This connects directly to the closures from the previous notebook.


````python
def get_operation(operation):
    if operation == "add":
        return lambda a, b: a + b

    if operation == "multiply":
        return lambda a, b: a * b

add = get_operation("add")
multiply = get_operation("multiply")

print(add(10, 5))
print(multiply(10, 5))
````

The sequence is:

```text
get_operation("add")
        ↓
returns a function
        ↓
add stores that function
        ↓
add(10, 5)
```

The returned function can be called later.


## 13. Practical Higher-Order Function Examples

### Example 1: Reusable Operation

Create a function that applies another function to every number.


````python
def apply_operation(function, numbers):
    return [function(number) for number in numbers]

numbers = [1, 2, 3, 4]

print(apply_operation(lambda x: x ** 2, numbers))
print(apply_operation(lambda x: x * 10, numbers))
````

The same higher-order function supports different operations because the operation is supplied as an argument.

### Example 2: Custom Function Instead of Lambda

Higher-order functions do not require lambda.


````python
def square(x):
    return x ** 2

def apply_operation(function, numbers):
    return [function(number) for number in numbers]

numbers = [1, 2, 3, 4]

print(apply_operation(square, numbers))
````

This is an important distinction:

> **Higher-order functions are about functions being passed or returned. Lambda is simply one convenient way to create a small function.**


### Example 3: Choosing an Operation



````python
def calculate(operation, a, b):
    if operation == "add":
        return a + b
    elif operation == "multiply":
        return a * b
    else:
        return None

print(calculate("add", 10, 5))
print(calculate("multiply", 10, 5))
````

A more function-oriented approach is to pass the operation itself:


````python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def calculate(operation, a, b):
    return operation(a, b)

print(calculate(add, 10, 5))
print(calculate(multiply, 10, 5))
````

This demonstrates the power of treating functions as values: the behavior can be supplied to another function instead of hard-coded into it.


## 14. Lambda Limitations and When Not to Use It

Lambda is best for **small, clear operations**.

### Good use

```python
sorted(data, key=lambda x: x[1])
```

The purpose is immediately understandable.

### Poor use

Trying to put complicated business logic into one lambda can make code difficult to read and maintain.

For example, prefer a normal function for substantial logic:

```python
def calculate_discount(price, percentage):
    # Clear multi-step logic can go here.
    ...
```

rather than forcing complex logic into a single expression.

### Practical rule

> If the lambda becomes difficult to read, give the operation a descriptive name with `def`.
