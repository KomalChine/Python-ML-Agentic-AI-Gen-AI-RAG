# 03 — `*args` and `**kwargs`

## Functions & Scope

This notebook builds on function arguments and introduces **flexible arguments**.

By the end, you will understand how to:

- accept an arbitrary number of positional arguments with `*args`
- accept an arbitrary number of keyword arguments with `**kwargs`
- combine regular parameters with `*args` and `**kwargs`
- unpack lists, tuples, and dictionaries when calling functions

> **Scope:** This notebook focuses only on `*args`, `**kwargs`, and argument unpacking. Topics such as scope, closures, lambda, decorators, and recursion are covered later.


## 1. Why Flexible Arguments Are Needed

A function can be defined with a fixed number of parameters:

```python
def add(a, b):
    return a + b
```

This function expects two values. But sometimes we want a function that can accept **any number of values**.

For example:

- adding 2, 3, or 10 numbers
- displaying an unknown number of values
- accepting optional details about an object

Python provides `*args` and `**kwargs` for this purpose.


````python
def add(a, b):
    return a + b

print(add(10, 20))
````

## 2. `*args` — Variable-Length Positional Arguments

`*args` collects extra **positional arguments** passed to a function.

```python
def function(*args):
    ...
```

Inside the function, `args` is a **tuple**.

The name `args` is conventional. The important part is the `*`.


````python
def add_numbers(*args):
    print(args)

add_numbers(10, 20, 30, 40)
````

The output is a tuple:

```text
(10, 20, 30, 40)
```

So:

- `*` tells Python to collect positional arguments.
- `args` contains those arguments.
- The collected value is a tuple.


````python
def add_numbers(*args):
    return sum(args)

print(add_numbers(10, 20))
print(add_numbers(10, 20, 30, 40))
print(add_numbers())
````

`*args` can therefore handle **zero or more positional arguments**.

### Why is it useful?

Without `*args`, we might need separate functions or increasingly long parameter lists. With `*args`, one function can accept a variable number of positional values.


## 3. Working with `args`

Because `args` is a tuple, you can iterate over it and use indexing just like with other tuples.


````python
def display_values(*args):
    for value in args:
        print(value)

display_values("Python", "Java", "C++")
````

````python
def show_first(*args):
    if args:
        print(args[0])

show_first("Python", "Java", "C++")
show_first()
````

The `if args:` check prevents indexing an empty tuple.

You can use normal tuple operations with `args`, such as:

- indexing
- iteration
- `len()`
- membership testing


````python
def inspect_args(*args):
    print("Type:", type(args))
    print("Count:", len(args))
    print("First:", args[0] if args else "No values")

inspect_args(10, 20, 30)
````

## 4. `**kwargs` — Variable-Length Keyword Arguments

`**kwargs` collects extra **keyword arguments** passed to a function.

```python
def function(**kwargs):
    ...
```

Inside the function, `kwargs` is a **dictionary**.

Again, `kwargs` is only a conventional name. The `**` is what tells Python to collect keyword arguments.


````python
def display_info(**kwargs):
    print(kwargs)

display_info(name="Alice", age=25, city="Delhi")
````

The output is a dictionary:

```python
{'name': 'Alice', 'age': 25, 'city': 'Delhi'}
```

The keyword names become dictionary keys, and their supplied values become dictionary values.


````python
def display_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

display_info(name="Alice", age=25, city="Delhi")
````

Because `kwargs` is a dictionary, you can use dictionary operations such as:

- `.items()`
- `.keys()`
- `.values()`
- indexing by key


````python
def show_name(**kwargs):
    if "name" in kwargs:
        print("Name:", kwargs["name"])

show_name(name="Alice", age=25)
````

## 5. Working with `kwargs`

`**kwargs` can accept zero or more keyword arguments.



````python
def display_settings(**kwargs):
    print("Number of settings:", len(kwargs))
    for key, value in kwargs.items():
        print(f"{key} = {value}")

display_settings()
display_settings(theme="dark", language="English", notifications=True)
````

## 6. Combining Regular Parameters with `*args`

A regular parameter can come before `*args`.

```python
def greet(greeting, *names):
    ...
```

Here:

- `greeting` receives the first positional argument.
- Remaining positional arguments are collected into `names`.


````python
def greet(greeting, *names):
    for name in names:
        print(f"{greeting}, {name}!")

greet("Hello", "Alice", "Bob", "Charlie")
````

This pattern is useful when one value has a specific meaning and the remaining positional values are variable.

Conceptually:

```text
regular parameter → receives its value
        ↓
     *args → collects remaining positional arguments
```


## 7. Combining Regular Parameters with `**kwargs`

A regular parameter can also be combined with `**kwargs`.


````python
def create_user(name, **details):
    print("Name:", name)
    print("Details:", details)

create_user(
    "Alice",
    age=25,
    city="Delhi",
    profession="Developer"
)
````

Here:

- `name` receives `"Alice"`.
- `age`, `city`, and `profession` are collected into `details`.
- `details` is a dictionary.


## 8. Combining `*args` and `**kwargs`

A function can accept both variable-length positional and keyword arguments.


````python
def function(*args, **kwargs):
    print("Positional:", args)
    print("Keyword:", kwargs)

function(10, 20, 30, name="Alice", age=25)
````

Conceptually:

```text
positional arguments → *args   → tuple
keyword arguments    → **kwargs → dictionary
```

### A realistic example


````python
def student_info(name, *subjects, **details):
    print("Name:", name)
    print("Subjects:", subjects)
    print("Details:", details)

student_info(
    "Alice",
    "Python",
    "SQL",
    age=22,
    city="Delhi"
)
````

In this example:

- `name` is a regular parameter.
- `"Python"` and `"SQL"` are collected into `subjects`.
- `age` and `city` are collected into `details`.


## 9. Argument Unpacking

So far, we have seen `*` and `**` used in a **function definition** to collect arguments.

They can also be used in a **function call** to unpack existing collections.

There are two forms:

- `*` → positional unpacking
- `**` → keyword unpacking


### `*` for Positional Unpacking

Suppose we have a list:

```python
numbers = [10, 20, 30]
```

and a function that expects three positional arguments.


````python
numbers = [10, 20, 30]

def add(a, b, c):
    return a + b + c

print(add(*numbers))
````

This:

```python
add(*numbers)
```

is equivalent to:

```python
add(10, 20, 30)
```

The `*` takes the values from the iterable and passes them as separate positional arguments.


### Tuple Unpacking

The same idea works with tuples.


````python
numbers = (10, 20, 30)

def multiply(a, b, c):
    return a * b * c

print(multiply(*numbers))
````

## 10. Passing Lists/Tuples with `*`

`*` unpacking is especially useful when the values are already stored in a list or tuple.

The number and order of unpacked values must match what the function expects.


````python
def introduce(name, age, city):
    print(f"{name} is {age} years old and lives in {city}.")

person_list = ["Alice", 25, "Delhi"]
person_tuple = ("Bob", 30, "Mumbai")

introduce(*person_list)
introduce(*person_tuple)
````

Without unpacking, the list or tuple would be passed as a single argument rather than as three separate arguments.

### Keyword Unpacking with `**`

A dictionary can be unpacked into keyword arguments.


````python
user = {
    "name": "Alice",
    "age": 25
}

def introduce(name, age):
    print(name, age)

introduce(**user)
````

This:

```python
introduce(**user)
```

is equivalent to:

```python
introduce(name="Alice", age=25)
```

The dictionary keys must match the function's parameter names.


## 11. Passing Dictionaries with `**`

Dictionary unpacking is useful when function arguments are already stored as named data.


````python
def create_profile(name, age, city):
    print("Name:", name)
    print("Age:", age)
    print("City:", city)

profile = {
    "name": "Alice",
    "age": 25,
    "city": "Delhi"
}

create_profile(**profile)
````

The keys in `profile` are used as keyword argument names.

So the following two calls are equivalent:

```python
create_profile("Alice", 25, "Delhi")
```

and:

```python
create_profile(name="Alice", age=25, city="Delhi")
```

when using the appropriate unpacking form.


## 12. Practical Examples

### Example 1: Flexible Sum


````python
def calculate_total(*prices):
    return sum(prices)

print(calculate_total(100, 250))
print(calculate_total(100, 250, 75, 50))
````

### Example 2: Flexible Student Details


````python
def student_record(name, *subjects, **details):
    print("Student:", name)
    print("Subjects:", subjects)

    for key, value in details.items():
        print(f"{key}: {value}")

student_record(
    "Alice",
    "Python",
    "SQL",
    age=22,
    city="Delhi"
)
````

### Example 3: Forwarding Flexible Arguments

A function can receive flexible arguments and pass them to another function.


````python
def show_data(*args, **kwargs):
    print("Arguments:", args)
    print("Details:", kwargs)

values = [10, 20, 30]
details = {"name": "Alice", "age": 25}

show_data(*values, **details)
````

This demonstrates both kinds of unpacking in the same function call:

- `*values` → positional arguments
- `**details` → keyword arguments


## 13. Common Mistakes

### Mistake 1: Expecting `*args` to collect keyword arguments

```python
def test(*args):
    print(args)

test(a=10)
```

This does **not** put `a=10` into `args`.

`*args` collects positional arguments. Keyword arguments need `**kwargs`.


````python
def test(*args):
    print(args)

# test(a=10)  # TypeError
````

### Mistake 2: Expecting `**kwargs` to collect positional arguments

```python
def test(**kwargs):
    print(kwargs)

test(10, 20)
```

This does not work because `**kwargs` collects keyword arguments, not positional arguments.


````python
def test(**kwargs):
    print(kwargs)

# test(10, 20)  # TypeError
````

### Mistake 3: Forgetting the distinction between collection and unpacking

The same symbols have related but opposite-looking roles:

| Context | `*` | `**` |
|---|---|---|
| Function definition | Collect positional arguments | Collect keyword arguments |
| Function call | Unpack positional values | Unpack keyword values |



### Mistake 4: Using an incompatible number of unpacked values

If a function expects three positional arguments, the unpacked iterable must provide three values.


````python
def add(a, b, c):
    return a + b + c

numbers = [10, 20]

# add(*numbers)  # TypeError: not enough values
````

### Mistake 5: Dictionary keys do not match parameter names

For `**` unpacking, dictionary keys become keyword names.

```python
def introduce(name, age):
    ...
```

A dictionary such as `{"person": "Alice", "years": 25}` does not match those parameter names.


````python
def introduce(name, age):
    print(name, age)

wrong_data = {
    "person": "Alice",
    "years": 25
}

# introduce(**wrong_data)  # TypeError
````

### Ordering Rule

A common flexible-argument pattern is:

```python
def function(a, *args, **kwargs):
    ...
```

Conceptually:

```text
regular parameters
        ↓
      *args
        ↓
    **kwargs
```

This means regular parameters receive their intended arguments first, remaining positional arguments can be collected by `*args`, and keyword arguments can be collected by `**kwargs`.


````python
def function(a, *args, **kwargs):
    print("a:", a)
    print("args:", args)
    print("kwargs:", kwargs)

function(10, 20, 30, name="Alice", age=25)
````

## 14. Summary

### `*args`

- Accepts a variable number of **positional arguments**.
- Collects them into a **tuple**.
- `args` is a conventional name.

### `**kwargs`

- Accepts a variable number of **keyword arguments**.
- Collects them into a **dictionary**.
- `kwargs` is a conventional name.

### Argument unpacking

- `*` unpacks a list or tuple (or another suitable iterable) into positional arguments.
- `**` unpacks a dictionary into keyword arguments.

### Quick comparison

| Feature | `*args` | `**kwargs` |
|---|---|---|
| Accepts | Positional arguments | Keyword arguments |
| Inside function | Tuple | Dictionary |
| Example call | `func(10, 20)` | `func(a=10, b=20)` |
| Symbol | `*` | `**` |

### Key idea

```text
*args   → many positional arguments → tuple
**kwargs → many keyword arguments → dictionary

*collection  → unpack positional values
**dictionary → unpack keyword values
```

These tools make functions flexible while keeping the function interface clear and reusable.
