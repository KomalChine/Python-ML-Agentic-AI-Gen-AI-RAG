## 1. Built-in Functions vs User-Defined Functions

Python already provides many built-in functions.

For example:

```python
len("Python")
```

We can also create our own functions.

```python
def greet():
    print("Hello!")
```

A **user-defined function** is a reusable block of code created using the `def` keyword.


````python
len("Python")
````

````python
def greet():
    print("Hello!")
````

## 2. Defining a Function

The basic structure is:

```python
def function_name():
    statement
```

The indented statements form the **function body**.

### Important

Defining a function does **not** execute its body. The function runs when we call it.


````python
def greet():
    print("Hello, Python!")

print("Function defined")
````

````python
def greet():
    print("Hello")

print("Before calling")
greet()
print("After calling")
````

## 3. Calling a Function

A function is called by writing its name followed by parentheses.

One advantage of functions is that we can write the logic once and reuse it many times.


````python
def greet():
    print("Hello!")

greet()
greet()
greet()
````

### The basic idea

```text
Write once
    ↓
Call many times
    ↓
Reuse the same logic
```


## 4. Functions with Parameters

A function can receive information through **parameters**.

```python
def greet(name):
    print(f"Hello, {name}!")
```

Here:

- `name` is a **parameter** — a variable listed in the function definition.
- `"Komal"` is an **argument** — a value supplied when the function is called.


````python
def greet(name):
    print(f"Hello, {name}!")

greet("Komal")
greet("Rahul")
````

### Parameter vs Argument

```text
def greet(name):
           ↑
       parameter

greet("Komal")
      ↑
    argument
```

The parameter receives the argument when the function is called.


## 5. Multiple Parameters

A function can have more than one parameter.


````python
def add(a, b):
    print(a + b)

add(10, 20)
````

````python
def introduce(name, age, city):
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")

introduce("Komal", 25, "Hyderabad")
````

At this stage, focus on the basic idea of passing values into a function.

More detailed argument styles—such as positional arguments, keyword arguments, default arguments, positional-only parameters, and keyword-only parameters—will be covered in **`02_Function_Arguments.ipynb`**.


## 6. Return Values

A function can **return** a value to the code that called it.

Use the `return` statement:

```python
def add(a, b):
    return a + b
```


````python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
````

````python
def add(a, b):
    return a + b

result = add(10, 20)

print("Result:", result)
print("Result × 2:", result * 2)
````

The returned value can be stored in a variable, used in a calculation, passed to another function, or used in a condition.


## 7. `print()` vs `return`

These two statements have different purposes.

### Using `print()`

```python
def calculate_area(length, width):
    print(length * width)
```

This displays the result.

### Using `return`

```python
def calculate_area(length, width):
    return length * width
```

This gives the result back to the caller.


````python
def calculate_area(length, width):
    print(length * width)

calculate_area(10, 5)
````

````python
def calculate_area(length, width):
    return length * width

area = calculate_area(10, 5)

print("Area:", area)

if area > 40:
    print("Large area")
````

### Key difference

```text
print()
→ displays a value

return
→ sends a value back to the caller
```

If you need to use a function's result elsewhere in your program, `return` is usually what you need.


## 8. Multiple Return Values

A function can return multiple values.

Python packages multiple returned values into a tuple.


````python
def calculate(a, b):
    return a + b, a - b, a * b

addition, subtraction, multiplication = calculate(10, 5)

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
````

````python
def calculate(a, b):
    return a + b, a - b, a * b

result = calculate(10, 5)

print(result)
print(type(result))
````

The second example shows that the returned object is a tuple.

This connects directly to the earlier **Tuples** material.


## 9. Functions Without `return`

A function does not have to contain a `return` statement.

If a function does not explicitly return a value, Python returns `None`.


````python
def greet(name):
    print(f"Hello, {name}")

result = greet("Komal")

print("Returned value:", result)
````

The function displays:

```text
Hello, Komal
```

and the variable `result` contains:

```python
None
```

So:

```text
No explicit return value
        ↓
      None
```


## 10. Local Variables

Variables created inside a function normally belong to that function's **local scope**.

For now, we only need to recognize the basic idea. Detailed scope rules and the LEGB rule will be covered later in **`04_Function_Scope_and_LEGB.ipynb`**.


````python
def calculate():
    x = 10
    print("Inside function:", x)

calculate()
````

The variable `x` was created inside `calculate()`, so it is a local variable of that function.

Trying to use `x` outside the function does not work because `x` is not defined in the outer scope.


````python
def calculate():
    x = 10
    print("Inside function:", x)

calculate()

# Uncomment the next line to see the error:
# print(x)
````

## 11. Function Documentation with Docstrings

A **docstring** is a string placed at the beginning of a function body to describe what the function does.

Docstrings make functions easier to understand and document.


````python
def calculate_area(length, width):
    """Return the area of a rectangle."""
    return length * width

print(calculate_area(10, 5))
````

````python
help(calculate_area)
````

````python
print(calculate_area.__doc__)
````

This connects with the `help()` function from the Basic Functions notebook.

Keep docstrings short and clear when a function's purpose needs explanation.


## 12. Function Naming and Good Practices

Use descriptive names that communicate what the function does.

### Good

```python
def calculate_average():
    ...
```

### Less clear

```python
def ca():
    ...
```

### Basic guidelines

- Use descriptive function names.
- Use `snake_case` for function names.
- Give a function one clear responsibility when practical.
- Keep functions reasonably small and readable.
- Choose parameter names that clearly describe the values they represent.


````python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

scores = [80, 90, 70, 60]

print(calculate_average(scores))
````

## 13. Putting It Together

Let's combine the main ideas from this notebook in one small example.


````python
def calculate_total(price, quantity):
    return price * quantity

price = 250
quantity = 3

total = calculate_total(price, quantity)

print("Price:", price)
print("Quantity:", quantity)
print("Total:", total)
````
