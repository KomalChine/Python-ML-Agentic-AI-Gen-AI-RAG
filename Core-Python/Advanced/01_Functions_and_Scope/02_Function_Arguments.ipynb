## 1. Parameters vs Arguments

```python
def greet(name):
    print(f"Hello, {name}!")
```

Here, `name` is a **parameter**.

When we call:

```python
greet("Komal")
```

`"Komal"` is an **argument**.

- **Parameter** → variable in the function definition.
- **Argument** → actual value supplied when the function is called.


````python
def greet(name):
    print(f"Hello, {name}!")

greet("Komal")
````

## 2. Positional Arguments

With positional arguments, values are assigned to parameters according to their **position**.


````python
def introduce(name, age):
    print(f"Name: {name}")
    print(f"Age: {age}")

introduce("Alice", 25)
````

The mapping is:

```text
"Alice" → name
25      → age
```

### Order matters

```python
introduce(25, "Alice")
```

This is valid Python because the number of arguments matches the parameters, but the values are assigned to the opposite parameters.

Too few or too many arguments raise `TypeError`.


````python
# Uncomment each line separately to see the TypeError.

# introduce("Alice")
# introduce("Alice", 25, "Delhi")
````

## 3. Keyword Arguments

With keyword arguments, a value is assigned to a parameter by its **parameter name**.


````python
def introduce(name, age, city):
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")

introduce(name="Alice", age=25, city="Delhi")
````

````python
introduce(city="Delhi", name="Alice", age=25)
````

### Positional vs Keyword

```text
Positional argument
→ assignment based on position

Keyword argument
→ assignment based on parameter name
```


## 4. Default Arguments

A parameter can have a default value.

```python
def greet(name, message="Hello"):
    ...
```

If the caller does not provide `message`, Python uses `"Hello"`.


````python
def greet(name, message="Hello"):
    print(f"{message}, {name}!")

greet("Alice")
greet("Alice", "Good morning")
greet(name="Alice", message="Welcome")
````

````python
def create_profile(name, age=18, city="Delhi"):
    print(name, age, city)

create_profile("Alice")
create_profile("Alice", 25)
create_profile("Alice", 25, "Mumbai")
create_profile("Alice", city="Mumbai")
````

### Required Parameters Before Default Parameters

This is invalid:

```python
def greet(message="Hello", name):
    ...
```

A non-default parameter cannot follow a default parameter.

Correct:

```python
def greet(name, message="Hello"):
    ...
```


## 5. Mixing Positional and Keyword Arguments

Positional arguments can be followed by keyword arguments.


````python
def user(name, age, city):
    print(name, age, city)

user("Alice", age=25, city="Delhi")
````

This is invalid:

```python
user(name="Alice", 25, "Delhi")
```

**Rule:** positional arguments must come before keyword arguments in a function call.


## 6. Positional-Only Parameters

The `/` marks the end of positional-only parameters.


````python
def greet(name, /):
    print(f"Hello, {name}!")

greet("Alice")
````

````python
def divide(a, b, /):
    return a / b

print(divide(10, 2))
````

This is not allowed:

```python
greet(name="Alice")
divide(a=10, b=2)
```

Parameters before `/` are **positional-only**.


## 7. Keyword-Only Parameters

The `*` can require parameters after it to be supplied by keyword.


````python
def create_user(name, *, age, city):
    print(name, age, city)

create_user("Alice", age=25, city="Delhi")
````

This is invalid:

```python
create_user("Alice", 25, "Delhi")
```

Parameters after `*` are **keyword-only**.


## 8. Combining Parameter Types

```python
def example(a, b, /, c, d, *, e, f):
    ...
```

```text
a, b
↓
positional-only

c, d
↓
positional or keyword

e, f
↓
keyword-only
```


````python
def example(a, b, /, c, d, *, e, f):
    print(a, b, c, d, e, f)

example(1, 2, 3, 4, e=5, f=6)
example(1, 2, c=3, d=4, e=5, f=6)
````

Everything before `/` is positional-only, parameters between `/` and `*` may be positional or keyword, and everything after `*` is keyword-only.

`*args` and `**kwargs` are intentionally not introduced here; they belong in the next notebook.


## 9. Common Argument Errors

### Missing required argument

```python
def greet(name):
    print(name)

greet()
```

The required parameter `name` was not supplied.

### Too many arguments

```python
greet("Alice", "Bob")
```

The function has one parameter but received two positional arguments.

### Unexpected keyword

```python
greet(username="Alice")
```

The function has `name`, not `username`.

### Multiple values for the same argument

```python
greet("Alice", name="Bob")
```

`name` receives a positional value and then a second keyword value.

### Positional argument after a keyword argument

```python
greet(name="Alice", "Bob")
```

Positional arguments must come before keyword arguments.


## 10. Summary

| Type | Example |
|---|---|
| Positional | `func(10, 20)` |
| Keyword | `func(a=10, b=20)` |
| Default | `def func(a, b=10)` |
| Positional-only | `def func(a, /)` |
| Keyword-only | `def func(*, a)` |

### Overall parameter syntax

```python
def function(
    positional_only,
    /,
    normal_parameter,
    *,
    keyword_only
):
    ...
```

### Key Rules

1. Positional arguments are assigned by position.
2. Keyword arguments are assigned by parameter name.
3. Positional arguments must come before keyword arguments.
4. Required parameters must come before default parameters.
5. Parameters before `/` are positional-only.
6. Parameters after `*` are keyword-only.
7. `*args` and `**kwargs` belong to the next notebook.
