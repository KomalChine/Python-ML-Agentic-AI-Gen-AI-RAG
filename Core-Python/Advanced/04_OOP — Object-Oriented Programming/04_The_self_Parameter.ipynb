# 04 — The `self` Parameter

This notebook focuses **only on `self`** in Python classes.

We have already used `self` in previous examples, but now we will understand what it means, why it is needed, how Python passes it, and how it connects instance methods to individual objects.

> **Core idea:** `self` refers to the **current object (instance)** on which an instance method is being called.

## 1. Introduction

Consider this simple class:

```python
class Student:
    def display(self):
        print("Student")
```

We can create an object and call the method:

```python
student = Student()
student.display()
```

But what exactly is `self`?

When `student.display()` is called, `self` refers to `student`.

````python
class Student:
    def display(self):
        print("Student")

student = Student()
student.display()
````

### Mental model

```text
student.display()
       ↓
self = student
```

The object that calls the instance method becomes the value of `self`.

## 2. What Is `self`?

We can print `self` directly to see that it represents the current object.

````python
class Student:
    def display(self):
        print("Current object:", self)

student = Student()
student.display()
````

If another object calls the same method, `self` refers to that other object.

```text
student1.display()
      ↓
self → student1

student2.display()
      ↓
self → student2
```

So `self` changes depending on **which object calls the method**.

````python
class Student:
    def display(self):
        print("Current object:", self)

student1 = Student()
student2 = Student()

student1.display()
student2.display()
````

## 3. Why Do We Need `self`?

The same method must be able to work with the data belonging to different objects.

For example:

````python
class Student:
    def display(self):
        print(self.name)

student1 = Student()
student2 = Student()

student1.name = "Komal"
student2.name = "Rahul"

student1.display()
student2.display()
````

For the first call:

```text
student1.display()
      ↓
self → student1
      ↓
self.name → student1.name
         → "Komal"
```

For the second call:

```text
student2.display()
      ↓
self → student2
      ↓
self.name → student2.name
         → "Rahul"
```

Therefore, `self` allows the **same method** to work with different objects.

## 4. `self` Refers to the Current Object

The central idea can be demonstrated directly:

````python
class Student:
    def show(self):
        print("Object:", self)

student1 = Student()
student2 = Student()

student1.show()
student2.show()
````

The mental model is:

```text
student1.show()
      ↓
self = student1

student2.show()
      ↓
self = student2
```

`self` is not a separate object. It is the name used inside the method for the object on which the method is currently operating.

## 5. Accessing Attributes Using `self`

One of the most important uses of `self` is accessing an object's attributes.

Inside an instance method:

```python
self.attribute
```

means:

> the attribute belonging to the current object.

````python
class Student:
    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

student1 = Student()
student2 = Student()

student1.name = "Komal"
student1.age = 22

student2.name = "Rahul"
student2.age = 23

student1.display()
student2.display()
````

For `student1.display()`:

```text
self.name → student1.name
self.age  → student1.age
```

For `student2.display()`:

```text
self.name → student2.name
self.age  → student2.age
```

This is how instance methods access the state of the particular object that called them.

## 6. Accessing Methods Using `self`

`self` can also be used to call another instance method belonging to the same object.

````python
class Student:
    def study(self):
        print("Student is studying")

    def start_study(self):
        print("Starting...")
        self.study()

student = Student()
student.start_study()
````

The flow is:

```text
student.start_study()
        ↓
self = student
        ↓
self.study()
        ↓
student.study()
```

This is useful when several methods work together on the same object's state.

## 7. `self` with Multiple Objects

Each object has its own state. `self` makes sure the method works with the correct object's state.

````python
class BankAccount:
    def show_balance(self):
        print("Balance:", self.balance)

account1 = BankAccount()
account2 = BankAccount()

account1.balance = 5000
account2.balance = 10000

account1.show_balance()
account2.show_balance()
````

For the first call:

```text
account1.show_balance()
        ↓
self = account1
        ↓
self.balance = account1.balance
        ↓
5000
```

For the second call:

```text
account2.show_balance()
        ↓
self = account2
        ↓
self.balance = account2.balance
        ↓
10000
```

The same `show_balance()` method works correctly for both accounts because `self` identifies the current object.

## 8. How Python Passes `self`

Now we can look at the underlying mechanism.

When we write:

```python
student.display()
```

Python effectively passes the object to the method:

```python
Student.display(student)
```

So this:

```python
def display(self):
```

receives the object as `self`.

This is the reason the first parameter of an ordinary instance method represents the instance.

````python
class Student:
    def display(self):
        print("Current object:", self)

student = Student()

# Normal method call
student.display()

# Explicit form
Student.display(student)
````

Both calls above operate on the same object.

```text
student.display()
       ↓
Python supplies student
       ↓
Student.display(student)
```

The second form makes the role of `self` especially clear.

## 9. Calling an Instance Method Through the Class

An instance method can also be called explicitly through the class when the object is supplied.

```python
Student.display(student)
```

But if we write:

```python
Student.display()
```

there is no object available to become `self`, so Python raises an error.

````python
class Student:
    def display(self):
        print("Student:", self)

student = Student()

student.display()
Student.display(student)
````

### Missing the object

The following call fails because no value is supplied for `self`:

```python
Student.display()
```

The important distinction is:

```text
student.display()
    → Python supplies the instance automatically

Student.display(student)
    → we supply the instance explicitly
```

````python
class Student:
    def display(self):
        print("Student")

student = Student()

try:
    Student.display()
except TypeError as error:
    print(type(error).__name__ + ":", error)
````

## 10. `self` Is Not a Python Keyword

`self` is the **conventional name** for the first parameter of an instance method. It is not a reserved Python keyword.

Technically, another valid name can be used:

````python
class Student:
    def display(current_object):
        print(current_object)

student = Student()
student.display()
````

However, this is not the normal style.

Prefer:

```python
class Student:
    def display(self):
        print(self)
```

Use `self` because it is the universally accepted Python convention.

```text
self
 ↓
conventional name

not
 ↓
reserved keyword
```

## 11. Why `self` Must Be the First Parameter

Consider a method that also accepts another argument:

````python
class Student:
    def display(self, name):
        print("Object:", self)
        print("Name:", name)

student = Student()
student.display("Komal")
````

Conceptually, Python treats:

```python
student.display("Komal")
```

as:

```python
Student.display(student, "Komal")
```

Therefore:

```text
self → student
name → "Komal"
```

The first parameter receives the automatically supplied instance, while the following parameters receive the arguments passed to the method.

## 12. Common Mistakes with `self`

### Mistake 1 — Forgetting `self`

Incorrect:

```python
class Student:
    def display():
        print("Student")
```

When `student.display()` is called, Python supplies the instance, but the method has no parameter to receive it.

Correct:

```python
class Student:
    def display(self):
        print("Student")
```

````python
class Student:
    def display():
        print("Student")

student = Student()

try:
    student.display()
except TypeError as error:
    print(type(error).__name__ + ":", error)
````

### Mistake 2 — Forgetting `self` before an attribute

Incorrect:

```python
def display(self):
    print(name)
```

Correct:

```python
def display(self):
    print(self.name)
```

`name` by itself is a local/name lookup. `self.name` explicitly accesses the attribute of the current object.

````python
class Student:
    def display(self):
        print(self.name)

student = Student()
student.name = "Komal"
student.display()
````

### Mistake 3 — Using a local variable instead of object state

Here `name` is only a local variable:

```python
class Student:
    def display(self):
        name = "Komal"
        print(name)
```

But here `name` is stored on the object:

```python
class Student:
    def display(self):
        print(self.name)
```

The distinction matters because object state remains associated with the object and can be accessed by its other methods.

````python
class Student:
    def set_name(self, name):
        self.name = name

    def display(self):
        print(self.name)

student = Student()
student.set_name("Komal")
student.display()
````

### Mistake 4 — Calling an instance method through the class without an object

Incorrect:

```python
Student.display()
```

Correct:

```python
student = Student()
student.display()
```

Or explicitly:

```python
Student.display(student)
```

## 13. `self` and Object State

Now combine the ideas.

The methods below use `self` to create and access object state:

````python
class Student:
    def set_data(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

student1 = Student()
student2 = Student()

student1.set_data("Komal", 85)
student2.set_data("Rahul", 90)

student1.display()
student2.display()
````

The important flow for the first object is:

```text
student1.set_data("Komal", 85)
        ↓
self = student1
        ↓
self.name = "Komal"
self.marks = 85
```

For the second object:

```text
student2.set_data("Rahul", 90)
        ↓
self = student2
        ↓
self.name = "Rahul"
self.marks = 90
```

So `self` connects a method to the **particular object whose state it needs to work with**.

## 14. Practical Examples

### Example 1 — Student

````python
class Student:
    def set_details(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"{self.name}: {self.marks}")

student = Student()
student.set_details("Komal", 85)
student.display()
````

### Example 2 — Rectangle

Here `self.length` and `self.width` refer to the dimensions of the current rectangle.

````python
class Rectangle:
    def set_dimensions(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

rectangle = Rectangle()
rectangle.set_dimensions(10, 5)

print("Area:", rectangle.area())
````

### Example 3 — Bank Account

Every method works with the same object's balance through `self`.

````python
class BankAccount:
    def set_balance(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def show_balance(self):
        print("Balance:", self.balance)

account = BankAccount()

account.set_balance(5000)
account.deposit(1000)
account.show_balance()
````

