## 2. Assignment vs Copying

### Assignment

```python
a = [1, 2, 3]
b = a
```

Both references point to the same object:

```text
a ───┐
     ├──► [1, 2, 3]
b ───┘
```

### Copying

A copy creates another object:

```text
a ───► [1, 2, 3]

b ───► [1, 2, 3]
```

The two outer objects are separate.

````python
a = [1, 2, 3]
b = a

print("Assignment:")
print("a is b:", a is b)

b = a.copy()

print("\nAfter copying:")
print("a is b:", a is b)
````

The important distinction is:

```text
Assignment → another reference to the same object

Copying → a new object
```

The rest of this notebook explores different kinds of copying.

## 3. Why Simple Assignment Is Not Copying

Consider:

```python
a = [10, 20, 30]
b = a
```

Then modify `b`.

````python
a = [10, 20, 30]
b = a

b.append(40)

print(a)
print(b)
````

Output:

```text
[10, 20, 30, 40]
[10, 20, 30, 40]
```

Both variables refer to the same list.

```text
a ─────┐
       ├──► [10, 20, 30, 40]
b ─────┘
```

Therefore, changing the list through `b` also changes what is seen through `a`.

# 4. Creating a Shallow Copy

Python's `copy` module provides `copy.copy()` for creating a shallow copy.

First import the module:

```python
import copy
```

Then:

```python
a = [10, 20, 30]
b = copy.copy(a)
```

A new outer list is created.

````python
import copy

a = [10, 20, 30]
b = copy.copy(a)

print(a)
print(b)
print(a is b)
````

The result of:

```python
a is b
```

is:

```text
False
```

because a new outer object was created.

## 5. Shallow Copy with a Simple List

For a flat list, a shallow copy behaves as we usually expect.

````python
import copy

a = [10, 20, 30]
b = copy.copy(a)

b.append(40)

print(a)
print(b)
````

Output:

```text
[10, 20, 30]
[10, 20, 30, 40]
```

The outer lists are different objects, so appending an item to `b` does not change `a`.

For a simple flat list, a shallow copy is often sufficient.

## 6. Different Ways to Create a Shallow Copy

For common containers such as lists, several techniques can create a shallow copy.

### `copy.copy()`

```python
b = copy.copy(a)
```

### List slicing

```python
b = a[:]
```

### `list()`

```python
b = list(a)
```

### `copy()` method

```python
b = a.copy()
```

For a simple flat list, these create a separate outer list.

````python
import copy

a = [1, 2, 3]

b1 = copy.copy(a)
b2 = a[:]
b3 = list(a)
b4 = a.copy()

print(b1)
print(b2)
print(b3)
print(b4)

print(b1 is a)
print(b2 is a)
print(b3 is a)
print(b4 is a)
````

These techniques create separate outer list objects.

However, the important limitation appears when the container contains **nested mutable objects**.

## 7. Checking Identity After Copying

Remember the difference between `==` and `is`.

````python
a = [1, 2, 3]
b = a.copy()

print(a == b)
print(a is b)
````

Output:

```text
True
False
```

This reinforces the previous notebook:

- `==` → compares equality/value
- `is` → checks whether the references point to the same object

The two lists have equal contents but are different objects.

# 8. The Important Problem: Nested Objects

The real difference between shallow and deep copy becomes visible when a container contains other objects.

Consider:

```python
a = [[1, 2], [3, 4]]
b = copy.copy(a)
```

At first glance:

```text
a → [[1, 2], [3, 4]]
b → [[1, 2], [3, 4]]
```

But internally the structure is:

```text
a ─────► Outer List A
           │
           ├──► [1, 2]
           │
           └──► [3, 4]

b ─────► Outer List B
           │
           ├──► [1, 2]
           │
           └──► [3, 4]
```

The outer list is different, but the nested lists are shared.

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)

print("Outer objects:", a is b)
print("First nested objects:", a[0] is b[0])
print("Second nested objects:", a[1] is b[1])
````

This is the central idea of a shallow copy:

```text
Outer object → copied
Nested objects → references may be shared
```

## 9. Shallow Copy and Nested Mutable Objects

Now modify one of the nested lists.

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)

b[0].append(100)

print(a)
print(b)
````

Output:

```text
[[1, 2, 100], [3, 4]]
[[1, 2, 100], [3, 4]]
```

Why did the original change?

Because:

```text
b[0] ───► same nested list as a[0]
```

The outer list was copied, but the nested list was not.

# 10. Outer Object vs Inner Objects

Check both levels directly:

```python
print(a is b)
print(a[0] is b[0])
```

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)

print(a is b)
print(a[0] is b[0])
````

Expected results:

```text
False
True
```

Therefore:

```text
Outer object → different
Nested object → same
```

This is the defining idea of a **shallow copy**.

# 11. Modifying the Outer Object vs Nested Object

Compare two different operations.

### Outer modification

```python
b.append([5, 6])
```

This modifies only `b` because `b` has its own outer list.

### Nested modification

```python
b[0].append(100)
```

This can affect both `a` and `b` because `a[0]` and `b[0]` refer to the same nested list.

This distinction is extremely important when working with nested data.

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)

# Modify only the copied outer list
b.append([5, 6])

print("After outer modification:")
print("a:", a)
print("b:", b)

# Modify a shared nested list
b[0].append(100)

print("\nAfter nested modification:")
print("a:", a)
print("b:", b)
````

The two operations behave differently because one changes the copied outer object while the other changes a nested object that is shared.

# 12. What Is a Shallow Copy?

A **shallow copy** creates a new outer object but does not recursively copy the objects contained inside it.

Mental model:

```text
Shallow Copy

Original
   │
   ▼
Outer Object ──► Inner Object
                    ▲
                    │
Copied Outer ───────┘
```

So:

- The outer object is new.
- Nested objects can still be shared.
- Changes to shared mutable nested objects can be visible through both structures.

# 13. What Is a Deep Copy?

A **deep copy** recursively copies the objects contained inside the original structure.

Use:

```python
b = copy.deepcopy(a)
```

For example:

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
```

Now the nested lists are also separate objects.

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)

print("Outer objects:", a is b)
print("First nested objects:", a[0] is b[0])
print("Second nested objects:", a[1] is b[1])
````

The important difference is:

```text
Deep Copy

Original
   │
   ├──► [1, 2]
   └──► [3, 4]

Copied
   │
   ├──► [1, 2]
   └──► [3, 4]
```

The displayed values may be the same, but the nested objects are separate.

# 14. Deep Copy with Nested Lists

Now modify a nested list in the deep copy.

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)

b[0].append(100)

print(a)
print(b)
````

Output:

```text
[[1, 2], [3, 4]]
[[1, 2, 100], [3, 4]]
```

The original remains unchanged because the nested list was copied as well.

# 15. Checking Deep-Copy Identity

Check the outer and nested objects:

```python
print(a is b)
print(a[0] is b[0])
```

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)

print(a is b)
print(a[0] is b[0])
````

Both results are:

```text
False
False
```

So:

```text
Deep Copy
│
├── Outer object → different
└── Nested objects → different
```

# 16. Shallow Copy vs Deep Copy

| Feature | Shallow Copy | Deep Copy |
|---|---|---|
| New outer object | Yes | Yes |
| Nested objects copied | No | Yes |
| Nested references shared | Often yes | No |
| `copy.copy()` | Yes | No |
| `copy.deepcopy()` | No | Yes |
| Memory required | Less | More |
| Processing required | Less | More |

The important distinction is whether nested objects are independently copied.

# 17. Visual Comparison

### Shallow copy

```text
Original ─────► Outer A
                  │
                  └──► Inner List ◄────┐
                                       │
Copied ────────► Outer B ──────────────┘
```

The outer objects are different, but the inner list is shared.

### Deep copy

```text
Original ─────► Outer A
                  │
                  └──► Inner List A

Copied ────────► Outer B
                  │
                  └──► Inner List B
```

The outer and nested objects are separate.

# 18. Shallow Copy with Dictionaries

Shallow copying is not limited to lists.

Consider a dictionary containing a nested list:

````python
import copy

original = {
    "name": "Komal",
    "skills": ["Python", "SQL"]
}

shallow = copy.copy(original)

shallow["skills"].append("Excel")

print("Original:", original)
print("Shallow:", shallow)
````

The original dictionary's `skills` list is also affected.

Why?

```text
original["skills"] ──┐
                     ├──► same list
shallow["skills"] ───┘
```

The dictionaries are separate, but the nested list is shared.

# 19. Deep Copy with Dictionaries

Now use `deepcopy()`:

````python
import copy

original = {
    "name": "Komal",
    "skills": ["Python", "SQL"]
}

deep = copy.deepcopy(original)

deep["skills"].append("Excel")

print("Original:", original)
print("Deep:", deep)
````

The original remains unchanged because the nested `skills` list was also copied.

# 20. Copying Custom Objects

Shallow and deep copying also apply to objects created from our own classes.

Consider:

```python
class Student:
    def __init__(self, name, subjects):
        self.name = name
        self.subjects = subjects
```

````python
import copy

class Student:
    def __init__(self, name, subjects):
        self.name = name
        self.subjects = subjects

student1 = Student(
    "Komal",
    ["Python", "SQL"]
)

student2 = copy.copy(student1)

print(student1.name)
print(student2.name)
print(student1 is student2)
````

A shallow copy creates a new `Student` object.

However, the `subjects` attribute refers to a list, which is a nested mutable object.

# 21. Shallow Copy of a Custom Object

Check the object and its nested list:

````python
import copy

class Student:
    def __init__(self, name, subjects):
        self.name = name
        self.subjects = subjects

student1 = Student("Komal", ["Python", "SQL"])
student2 = copy.copy(student1)

print("Student objects:", student1 is student2)
print("Subjects lists:", student1.subjects is student2.subjects)
````

Expected result:

```text
Student objects: False
Subjects lists: True
```

The `Student` objects are different, but their `subjects` lists are shared.

Therefore, changing the subjects through the shallow copy can affect the original.

````python
student2.subjects.append("Excel")

print("student1 subjects:", student1.subjects)
print("student2 subjects:", student2.subjects)
````

This is the same shallow-copy rule we saw with nested lists and dictionaries:

```text
Outer object → copied
Nested mutable object → shared
```

# 22. Deep Copy of a Custom Object

Now create a deep copy:

````python
import copy

class Student:
    def __init__(self, name, subjects):
        self.name = name
        self.subjects = subjects

student1 = Student("Komal", ["Python", "SQL"])
student3 = copy.deepcopy(student1)

print("Student objects:", student1 is student3)
print("Subjects lists:", student1.subjects is student3.subjects)
````

Expected result:

```text
Student objects: False
Subjects lists: False
```

The nested `subjects` list is also independently copied.

````python
student3.subjects.append("Excel")

print("student1 subjects:", student1.subjects)
print("student3 subjects:", student3.subjects)
````

The original student's subject list remains unchanged.

# 23. Copying Objects with Nested Objects

Copying becomes even more useful when one custom object contains another custom object.

Consider:

````python
class Address:
    def __init__(self, city):
        self.city = city


class Student:
    def __init__(self, name, address):
        self.name = name
        self.address = address


address = Address("Hyderabad")
student1 = Student("Komal", address)

print(student1.name)
print(student1.address.city)
````

Now create both kinds of copies:

```python
student2 = copy.copy(student1)
student3 = copy.deepcopy(student1)
```

The shallow copy creates a new `Student` object but can share the same `Address` object.

The deep copy creates a new `Student` object and recursively copies its `Address` object.

````python
import copy

class Address:
    def __init__(self, city):
        self.city = city


class Student:
    def __init__(self, name, address):
        self.name = name
        self.address = address


address = Address("Hyderabad")
student1 = Student("Komal", address)

student2 = copy.copy(student1)
student3 = copy.deepcopy(student1)

print("student1 is student2:", student1 is student2)
print("student1.address is student2.address:",
      student1.address is student2.address)

print("student1 is student3:", student1 is student3)
print("student1.address is student3.address:",
      student1.address is student3.address)
````

The reference structure explains the results.

# 24. Shallow Copy vs Deep Copy with Nested Objects

### Shallow copy

```text
student1 ─────► Student A
                   │
                   └──► Address A ◄──── student2
```

The two `Student` objects are different, but both reference the same `Address`.

### Deep copy

```text
student1 ─────► Student A
                   │
                   └──► Address A

student3 ─────► Student B
                   │
                   └──► Address B
```

The `Student` objects and their nested `Address` objects are separate.

This is why deep copy is useful when complete independence is required.

````python
import copy

class Address:
    def __init__(self, city):
        self.city = city


class Student:
    def __init__(self, name, address):
        self.name = name
        self.address = address


address = Address("Hyderabad")
student1 = Student("Komal", address)

student2 = copy.copy(student1)
student3 = copy.deepcopy(student1)

student2.address.city = "Mumbai"

print("Original:", student1.address.city)
print("Shallow copy:", student2.address.city)
print("Deep copy:", student3.address.city)
````

The shallow copy shares the original address, so changing it through `student2` also affects `student1`.

The deep copy has an independent address, so `student3` remains unchanged.

# 25. When Should You Use Shallow Copy?

Shallow copy can be useful when:

- You need a new outer container.
- Nested objects can safely be shared.
- The structure is flat.
- You do not need independent nested mutable objects.
- Lower memory usage is desirable.

For example:

```python
original = [1, 2, 3]
new_list = copy.copy(original)
```

Because the list is flat, there are no nested mutable objects to worry about.

# 26. When Should You Use Deep Copy?

Deep copy can be useful when:

- Objects contain nested mutable objects.
- You need complete independence.
- Changes to nested objects should not affect the original.
- You are working with complex object structures.

Example:

```python
new_object = copy.deepcopy(original)
```

The decision should depend on the behavior you need rather than simply choosing deep copy by default.

# 27. Performance and Memory Considerations

Deep copy is not automatically "better."

A deep copy:

- creates more objects,
- uses more memory,
- can require more processing,
- may be unnecessary for simple or flat data.

A shallow copy generally does less work because it does not recursively copy nested objects.

Therefore, choose the copying strategy according to the structure of the data and the required independence.

# 28. Common Mistakes

### Mistake 1 — Thinking assignment is a shallow copy

```python
b = a
```

is not a shallow copy.

It creates another reference to the same object.

### Mistake 2 — Thinking shallow copy duplicates everything

A shallow copy creates a new outer object, but nested objects can still be shared.

### Mistake 3 — Assuming `a.copy()` always creates completely independent data

For a nested structure, `a.copy()` is a shallow copy.

Nested mutable objects can remain shared.

### Mistake 4 — Using deep copy unnecessarily

Deep copying can consume additional memory and processing time.

Use it when independent nested objects are actually needed.

### Mistake 5 — Confusing equality and identity

```python
a == b
```

checks equality, while:

```python
a is b
```

checks identity.
