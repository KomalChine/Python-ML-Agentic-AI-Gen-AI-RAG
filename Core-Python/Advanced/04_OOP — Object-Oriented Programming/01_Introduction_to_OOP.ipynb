# 01 — Introduction to OOP

This notebook introduces the core ideas of **Object-Oriented Programming (OOP)** in Python.

The focus is on objects, classes, attributes, methods, and why OOP is useful. Detailed topics such as `self`, `__init__`, inheritance, polymorphism, properties, and dunder methods are covered later.


## 1. Introduction

**Object-Oriented Programming (OOP)** is a programming approach where programs are designed around **objects** that contain data and behavior.

```text
Object
 ├── Data
 └── Behavior
```

For example, a `Student` object might contain:

```text
Data:
    name
    age
    marks

Behavior:
    study()
    attend_class()
    display_result()
```


## 2. What Is OOP?

Python supports multiple programming styles, including:

- Procedural programming
- Object-oriented programming
- Functional programming

The focus here is OOP. Higher-order functions and related functional concepts were covered earlier.

A procedural example:


````python
student_name = "Komal"
student_age = 22
student_marks = 85

def display_student(name, age, marks):
    print(name)
    print(age)
    print(marks)

display_student(student_name, student_age, student_marks)
````

With OOP, related data and behavior can be grouped together into objects, making larger programs easier to organize.


## 3. Why Do We Need OOP?

Imagine managing 100 students.

Separate variables could lead to:

```text
student1_name
student1_age
student1_marks

student2_name
student2_age
student2_marks
```

This becomes difficult to maintain.

With OOP, we can work with objects such as:

```text
student1
student2
student3
```

Each object can contain its own data and methods.

```text
Related data + related behavior
              ↓
            Object
```


## 4. Procedural Programming vs OOP

### Procedural approach

```text
Data
 ↓
Functions
 ↓
Operations
```

Example:


````python
balance = 1000

def deposit(balance, amount):
    return balance + amount
````

### OOP approach

```text
Object
 ├── Data
 └── Methods
```

Conceptually, this can eventually allow operations such as:

```python
account.deposit(500)
```

The goal here is to understand the programming style, not implement the complete class yet.

## 6. Classes and Objects — Basic Idea

The most important relationship is:

```text
Class
  ↓
Blueprint
  ↓
Objects
```

- **Class** → blueprint or template
- **Object** → actual instance created from a class

Detailed class syntax and object creation come in the next notebook.


## 7. Attributes and Methods — Basic Idea

### Attributes

Attributes represent an object's data or state.

```text
Student
 ├── name
 ├── age
 └── marks
```

### Methods

Methods represent behavior.

```text
Student
 ├── study()
 ├── display()
 └── calculate_grade()
```

Mental model:

```text
Object
│
├── Attributes → What the object has
│
└── Methods    → What the object can do
```


## 8. The Four Main OOP Concepts

The four major OOP concepts are:

```text
Encapsulation
Inheritance
Polymorphism
Abstraction
```

### Encapsulation
Bundling data and behavior together and controlling access to internal details.

### Inheritance
Creating a new class based on an existing class.

### Polymorphism
Allowing the same interface or operation to behave differently for different objects.

### Abstraction
Exposing essential behavior while hiding unnecessary implementation details.

These will be taught in dedicated later sections.


## 9. OOP in Python

Python is object-oriented, and many things we have already used are objects.

For example:


````python
numbers = [1, 2, 3]
numbers.append(4)
print(numbers)
````

`numbers` is a list object, and `append()` is a method of that object.

Strings are objects too:


````python
text = "Python"
print(text.upper())
````

This connects OOP with concepts already encountered while learning Python.


## 10. Simple OOP Example

Here is a very small class:


````python
class Student:
    def display(self):
        print("I am a student.")
````

````python
student = Student()
student.display()
````

The basic flow is:

```text
class Student
      ↓
   blueprint
      ↓
student = Student()
      ↓
    object
      ↓
student.display()
      ↓
    method
```

The detailed meaning of `self` will be covered in `04_The_self_Parameter.ipynb`.


## 11. When Should We Use OOP?

OOP can be particularly useful when:

- the program has many related entities
- objects have both data and behavior
- the application is large
- code needs reusable components
- multiple objects share the same structure
- maintaining state is important

Examples include:

```text
Banking systems
E-commerce applications
Games
GUI applications
Data-processing systems
Web applications
Large software projects
```

For small scripts, OOP is not always necessary.


## 12. Advantages of OOP

### 1. Organization
Related data and behavior can be grouped together.

### 2. Reusability
A class can be used to create many objects.

### 3. Maintainability
Large applications can be divided into logical components.

### 4. Scalability
New functionality can be added through new classes and relationships.

### 5. Modeling
Real-world entities can be represented naturally.

OOP is not automatically better than procedural programming; the appropriate approach depends on the problem.


## 13. Common Misconceptions

### Misconception 1 — Every Python program must use classes

False. Simple scripts can often be written perfectly well using functions.

### Misconception 2 — A class and an object are the same thing

They are different:

```text
Class  → blueprint
Object → instance
```

### Misconception 3 — OOP means only classes and objects

OOP also involves encapsulation, inheritance, polymorphism, abstraction, and composition.

### Misconception 4 — Everything needs to be modeled as a real-world object

Not necessarily. The purpose is useful organization and modeling, not forcing every piece of code into a class.
