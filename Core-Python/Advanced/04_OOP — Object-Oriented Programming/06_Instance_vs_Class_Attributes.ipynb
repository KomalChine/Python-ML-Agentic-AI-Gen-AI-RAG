# 06 — Instance vs Class Attributes

This notebook explains the difference between **instance attributes** and **class attributes**.

Before this notebook, we learned that `__init__()` commonly creates attributes using `self`:

```python
class Student:
    def __init__(self, name):
        self.name = name
```

Now we will understand **where that attribute belongs** and what happens when an attribute is defined directly inside the class.

## 1. Introduction

```python
class Student:
    school = "ABC College"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
```

Here:

```text
school → Class Attribute
name, marks → Instance Attributes
```

The key question is: what is the difference between an attribute belonging to an individual object and one belonging to the class?

````python
class Student:
    school = "ABC College"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student1 = Student("Komal", 85)
student2 = Student("Rahul", 92)

print(student1.name, student1.marks)
print(student2.name, student2.marks)
print(student1.school)
print(student2.school)
````

## 2. What Are Instance Attributes?

An **instance attribute** is associated with a particular object.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
```

Each object gets its own values.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student1 = Student("Komal", 85)
student2 = Student("Rahul", 92)

print(student1.name, student1.marks)
print(student2.name, student2.marks)
````

## 3. Creating Instance Attributes

The common pattern is:

```python
self.attribute = value
```

This creates/stores an attribute on the particular instance.

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

employee1 = Employee("Komal", 50000)
employee2 = Employee("Rahul", 60000)

print(employee1.name)
print(employee2.name)
print(employee1.salary)
print(employee2.salary)
````

## 4. Instance Attributes Belong to Individual Objects

Changing one object's instance attribute does not change another object's attribute.

````python
class Student:
    def __init__(self, name):
        self.name = name

student1 = Student("Komal")
student2 = Student("Rahul")

student1.name = "Priya"

print(student1.name)
print(student2.name)
````

Output:

```text
Priya
Rahul
```

Each object has its own instance attribute.

## 5. What Are Class Attributes?

A **class attribute** is defined directly inside the class body rather than through `self`.

```python
class Student:
    school = "ABC College"
```

It is associated with the class and can be accessed through the class or through instances.

````python
class Student:
    school = "ABC College"

student1 = Student()
student2 = Student()

print(student1.school)
print(student2.school)
print(Student.school)
````

All three accesses produce `ABC College`.

Conceptually, `school` is defined on the class rather than separately initialized in every object.

## 6. Creating Class Attributes

Class attributes are written directly in the class body.

````python
class Employee:
    company = "Tech Solutions"

    def __init__(self, name):
        self.name = name

employee1 = Employee("Komal")
employee2 = Employee("Rahul")

print(employee1.company)
print(employee2.company)
print(Employee.company)
````

Here:

```text
company → Class attribute
name → Instance attribute
```

## 7. Accessing Class Attributes

A class attribute can be accessed through:

```python
Student.school
```

or:

```python
student.school
```

The attribute is conceptually defined on the class.

````python
class Student:
    school = "ABC College"

student = Student()

print(Student.school)
print(student.school)
````

## 8. Instance vs Class Attributes

This is the central comparison.

```text
                   Student class
                        │
                        └── school = "ABC College"
                              │
                     ┌────────┴────────┐
                     ▼                 ▼
                 student1          student2
                 ├─ name           ├─ name
                 │  Komal          │  Rahul
                 └─ marks          └─ marks
                    85                92
```

Instance attributes normally contain data that can differ between objects.

Class attributes represent information associated with the class and commonly shared as a common/default value.

````python
class Student:
    school = "ABC College"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student1 = Student("Komal", 85)
student2 = Student("Rahul", 92)

print(student1.name, student1.marks, student1.school)
print(student2.name, student2.marks, student2.school)
````

## 9. How Python Looks Up Attributes

When you write:

```python
student.school
```

Python can look for `school` on the instance and, if it is not found there, find it on the class.

Conceptually:

```text
student.school
      ↓
check instance
      ↓
if not found
      ↓
check class
      ↓
return class value
```

````python
class Student:
    school = "ABC College"

student = Student()
print(student.school)
````

## 10. Shadowing a Class Attribute

If an instance is assigned an attribute with the same name as a class attribute, the instance value shadows the class value for that instance.

````python
class Student:
    school = "ABC College"

student1 = Student()
student2 = Student()

student1.school = "XYZ College"

print(student1.school)
print(student2.school)
print(Student.school)
````

The result is:

```text
student1.school → XYZ College
student2.school → ABC College
Student.school  → ABC College
```

`student1` now has its own instance-level `school`; it did not change `Student.school`.

## 11. Modifying Class Attributes

Use the class when you intend to change the class attribute:

```python
Student.school = "XYZ College"
```

Instances without their own `school` attribute will then see the new value.

````python
class Student:
    school = "ABC College"

student1 = Student()
student2 = Student()

Student.school = "XYZ College"

print(student1.school)
print(student2.school)
print(Student.school)
````

## 12. Modifying Instance Attributes

Instance attributes can be changed independently.

```python
student1.name = "Priya"
```

Only `student1.name` changes.

````python
class Student:
    school = "ABC College"

    def __init__(self, name):
        self.name = name

student1 = Student("Komal")
student2 = Student("Rahul")

student1.name = "Priya"

print(student1.name)
print(student2.name)
````

## 13. Class Attributes with Multiple Objects

A realistic example:

```python
class Employee:
    company = "DataTech"

    def __init__(self, name, department):
        self.name = name
        self.department = department
```

`company` is common information, while `name` and `department` are individual information.

````python
class Employee:
    company = "DataTech"

    def __init__(self, name, department):
        self.name = name
        self.department = department

employee1 = Employee("Komal", "Data Science")
employee2 = Employee("Rahul", "Development")
employee3 = Employee("Priya", "Testing")

print(employee1.company)
print(employee2.company)
print(employee3.company)

print(employee1.department)
print(employee2.department)
print(employee3.department)
````

## 14. When to Use Instance Attributes

Use an instance attribute when the value can differ between objects.

Examples for a student:

```text
name
age
marks
email
```

Examples for an employee:

```text
name
salary
department
employee_id
```

Typical pattern:

```python
self.name = name
self.salary = salary
```

## 15. When to Use Class Attributes

Use a class attribute when the value represents something associated with the class or a common/default value across instances.

Examples:

```python
class Employee:
    company = "DataTech"

class Student:
    school = "ABC College"
```

Other possible examples include:

```text
maximum_speed
organization_name
tax_rate
country
species
```

Do not make something class-level merely because several objects currently happen to have the same value. The design should reflect whether it is genuinely class-level/common state.

## 16. Practical Examples

### Example 1 — Student

````python
class Student:
    school = "ABC College"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student1 = Student("Komal", 85)
student2 = Student("Rahul", 92)

print(student1.name)
print(student2.name)
print(student1.school)
print(student2.school)
````

### Example 2 — Employee

````python
class Employee:
    company = "DataTech"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print(self.name)
        print(self.salary)
        print(self.company)

employee = Employee("Komal", 50000)
employee.display()
````

### Example 3 — Car

Here `brand` and `model` describe an individual car, while `wheels` is common class-level information.

````python
class Car:
    wheels = 4

    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

car1 = Car("Toyota", "Camry")
car2 = Car("Honda", "City")

print(car1.brand, car1.model, car1.wheels)
print(car2.brand, car2.model, car2.wheels)
````

## 17. Common Mistakes

### Mistake 1 — Thinking `self.x` creates a class attribute

```python
class Student:
    def __init__(self, name):
        self.name = name
```

`name` is an **instance attribute**, not a class attribute.

### Mistake 2 — Thinking class attributes automatically create copies

```python
class Student:
    school = "ABC College"
```

This does not mean every object necessarily has its own independent `school` attribute. The class provides the value through attribute lookup.

### Mistake 3 — Accidentally shadowing a class attribute

```python
student.school = "XYZ College"
```

creates an instance-level `school` for that object and does not modify `Student.school`.

### Mistake 4 — Changing the class when you intended to change one object

```python
Student.school = "XYZ College"
```

changes the class attribute. If only one object should differ, use:

```python
student.school = "XYZ College"
```

### Mistake 5 — Using mutable class attributes carelessly

This can cause shared mutable state:

```python
class Student:
    subjects = []
```

For per-object mutable data, initialize it inside `__init__()`:

```python
class Student:
    def __init__(self):
        self.subjects = []
```

````python
class Student:
    subjects = []

student1 = Student()
student2 = Student()

student1.subjects.append("Python")

print(student1.subjects)
print(student2.subjects)
````

````python
class Student:
    def __init__(self):
        self.subjects = []

student1 = Student()
student2 = Student()

student1.subjects.append("Python")

print(student1.subjects)
print(student2.subjects)
````

        ↓
