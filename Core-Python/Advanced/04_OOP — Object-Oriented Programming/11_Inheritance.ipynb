# 11 — Inheritance

Inheritance allows a new class to **reuse and extend** the attributes and methods of an existing class.

This notebook covers inheritance from fundamentals through practical use. Polymorphism and deep MRO concepts are reserved for later OOP notebooks.

## 1. Introduction to Inheritance

```text
Parent Class
     ↓
Child Class
     ↓
inherits + extends
```

**Parent / Base / Superclass** is the existing class.  
**Child / Derived / Subclass** is the class that inherits from it.

````python
class Animal:
    def eat(self):
        print("Animal is eating")


class Dog(Animal):
    pass


dog = Dog()
dog.eat()
````

## 2. Why Do We Need Inheritance?

Without inheritance, common behavior may be duplicated across related classes.

````python
class Dog:
    def eat(self):
        print("Eating")

    def sleep(self):
        print("Sleeping")


class Cat:
    def eat(self):
        print("Eating")

    def sleep(self):
        print("Sleeping")
````

````python
class Animal:
    def eat(self):
        print("Eating")

    def sleep(self):
        print("Sleeping")


class Dog(Animal):
    pass


class Cat(Animal):
    pass


dog = Dog()
cat = Cat()

dog.eat()
cat.sleep()
````

### Benefits

- Code reuse
- Less duplication
- Easier maintenance
- Logical relationships
- Extending existing classes
- Better organization

class ChildClass(ParentClass):

````python
class Person:
    def introduce(self):
        print("I am a person")


class Student(Person):
    pass


student = Student()
student.introduce()
````

## 5. Single Inheritance

```text
Parent
  ↓
Child
```

A child can have inherited behavior and its own behavior.

````python
class Animal:
    def eat(self):
        print("Eating")


class Dog(Animal):
    def bark(self):
        print("Barking")


dog = Dog()
dog.eat()
dog.bark()
````

## 6. Child Class Can Add New Attributes

A child can add its own instance attributes. Repeating the parent's initialization is possible, but `super()` provides a cleaner approach.

````python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks


student = Student("Komal", 85)

print(student.name)
print(student.marks)
````

## 7. Using `super()`

`super()` provides access to parent-class functionality.

```python
super().__init__(name)
```

calls the parent class's `__init__()`.

````python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks


student = Student("Komal", 85)

print(student.name)
print(student.marks)
````

## 8. Inheritance of Attributes and Methods

````python
class Employee:
    company = "DataTech"

    def work(self):
        print("Employee is working")


class Developer(Employee):
    def code(self):
        print("Developer is coding")


developer = Developer()

print(developer.company)
developer.work()
developer.code()
````

A child can access:

```text
Parent class attribute
Parent class method
Own attribute
Own method
```

## 9. Method Overriding

A child can provide its own implementation of a parent method.

> When a child defines a method with the same name as an inherited method, the child implementation overrides the parent implementation.

````python
class Animal:
    def sound(self):
        print("Some animal sound")


class Dog(Animal):
    def sound(self):
        print("Bark")


dog = Dog()
dog.sound()
````

## 10. Overriding `__init__()`

A child can define its own `__init__()` while still using the parent's initialization through `super()`.

````python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks


student = Student("Komal", 85)

print(student.name)
print(student.marks)
````

## 11. Parent Method + Child Extension

Sometimes a child should extend rather than completely replace a parent method.

````python
class Person:
    def introduce(self):
        print("I am a person")


class Student(Person):
    def introduce(self):
        super().introduce()
        print("I am also a student")


student = Student()
student.introduce()
````

## 12. Accessing Parent Attributes

A child can access inherited class attributes. More advanced attribute lookup will be connected to MRO later.

````python
class Vehicle:
    category = "Vehicle"


class Car(Vehicle):
    def show_category(self):
        print(super().category)


car = Car()
print(car.category)
car.show_category()
````

## 13. `isinstance()` with Inheritance

````python
class Animal:
    pass


class Dog(Animal):
    pass


dog = Dog()

print(isinstance(dog, Dog))
print(isinstance(dog, Animal))
````

Both results are `True` because a `Dog` object is also an `Animal` through inheritance.

## 14. `issubclass()`

`issubclass()` checks a relationship between classes.

```text
isinstance() → checks an object
issubclass() → checks a class relationship
```

````python
class Animal:
    pass


class Dog(Animal):
    pass


print(issubclass(Dog, Animal))
print(issubclass(Animal, Dog))
````

## 15. Single Inheritance — Practical Employee Example

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)


class Developer(Employee):
    def code(self):
        print(f"{self.name} is writing code")


developer = Developer("Komal", 50000)

developer.display()
developer.code()
````

## 16. Multilevel Inheritance

```text
Grandparent
     ↓
   Parent
     ↓

```

````python
class Animal:
    def eat(self):
        print("Eating")


class Mammal(Animal):
    def walk(self):
        print("Walking")


class Dog(Mammal):
    def bark(self):
        print("Barking")


dog = Dog()

dog.eat()
dog.walk()
dog.bark()
````

## 17. Multilevel Inheritance with `super()`

````python
class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary


class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department


manager = Manager("Komal", 70000, "Data Science")

print(manager.name)
print(manager.salary)
print(manager.department)
````

## 18. Hierarchical Inheritance

```text
       Parent
       /    \
 Child A   Child B
```

````python
class Animal:
    def eat(self):
        print("Eating")


class Dog(Animal):
    def bark(self):
        print("Barking")


class Cat(Animal):
    def meow(self):
        print("Meowing")


dog = Dog()
cat = Cat()

dog.eat()
dog.bark()

cat.eat()
cat.meow()
````

## 19. Multiple Inheritance

A class can inherit from more than one parent.

````python
class Father:
    def skills(self):
        print("Driving")


class Mother:
    def hobbies(self):
        print("Painting")


class Child(Father, Mother):
    pass


child = Child()

child.skills()
child.hobbies()
````

## 20. Multiple Inheritance — Important Caution

If multiple parents define the same method, Python needs a rule for deciding which implementation is selected.

````python
class A:
    def show(self):
        print("A")


class B:
    def show(self):
        print("B")


class C(A, B):
    pass


obj = C()
obj.show()
````

This leads to **Method Resolution Order (MRO)**. We introduce it here without going deeply into its algorithm.

## 21. Checking MRO

Use either:

```python
C.mro()
```

or:

```python
C.__mro__
```

For `class C(A, B)`, the order begins conceptually as:

```text
C → A → B → object
```

````python
class A:
    pass


class B:
    pass


class C(A, B):
    pass


print(C.mro())
print(C.__mro__)
````

## 22. `super()` and Multiple Inheritance

`super()` does not simply mean "my direct parent" in every multiple-inheritance situation. It follows Python's method resolution order.

````python
class A:
    def show(self):
        print("A")


class B(A):
    def show(self):
        super().show()
        print("B")


obj = B()
obj.show()
````

## 23. Hybrid Inheritance

Hybrid inheritance combines multiple inheritance patterns.

A common diamond structure is:

```text
       A
      / \
     B   C
      \ /
       D
```

````python
class A:
    pass


class B(A):
    pass


class C(A):
    pass


class D(B, C):
    pass


print(D.mro())
````

Complex inheritance relationships make MRO important. C3 linearization is intentionally not covered here.

## 24. The `object` Base Class

Python classes ultimately inherit from `object`.

````python
class Person:
    pass


print(Person.__mro__)
````

The result is equivalent to:

```text
(Person, object)
```

## 25. Inheritance and Class Attributes

````python
class Employee:
    company = "DataTech"


class Developer(Employee):
    pass


developer = Developer()

print(developer.company)

Developer.company = "TechCorp"

print(developer.company)
print(Employee.company)
````

An inherited class attribute can be overridden at the child-class level.

## 26. Inheritance and Instance Attributes

````python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks


student = Student("Komal", 85)

print(student.name)
print(student.marks)
````

The child object can contain both parent-defined and child-defined instance attributes:

```text
student
├── name
└── marks
```

## 27. When Should You Use Inheritance?

Inheritance is useful when there is a genuine **IS-A** relationship.

Examples:

```text
Dog is an Animal
Car is a Vehicle
Manager is an Employee
Student is a Person
```

By contrast:

```text
Car has an Engine
Student has an Address
Employee has a Laptop
```

are generally HAS-A relationships and are better represented through composition.

## 28. Inheritance vs Composition

### Inheritance

```python
class Dog(Animal):
    pass
```

means:

```text
Dog IS-A Animal
```

### Composition

```python
class Car:
    def __init__(self):
        self.engine = Engine()
```

means:

```text
Car HAS-A Engine
```

| Inheritance | Composition |
|---|---|
| IS-A | HAS-A |
| Class hierarchy | Object-to-object relationship |
| Reuse through inheritance | Reuse through contained objects |
| Can create tighter coupling | Often more flexible |

## 29. Common Mistakes

### 1. Forgetting `super().__init__()`

If the parent initializes important attributes, a child initializer may need to call the parent initializer.

### 2. Assuming inheritance copies source code

Inheritance does not literally copy the parent's source code into the child.

### 3. Confusing overriding with overloading

Defining the same method in a child class is **overriding**.

### 4. Excessive inheritance

Do not create inheritance relationships just to avoid writing a few lines.

### 5. Using inheritance for HAS-A relationships

`Car → Engine` is generally composition, not:

```python
class Car(Engine):
    pass
```

## 30. Practical Project — Employee Management System

Hierarchy:

```text
Employee
   │
   ├── Developer
   ├── Designer
   └── Manager
```

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)


class Developer(Employee):
    def code(self):
        print(self.name, "is coding.")


class Designer(Employee):
    def design(self):
        print(self.name, "is designing.")


class Manager(Employee):
    def manage(self):
        print(self.name, "is managing.")


developer = Developer("Komal", 60000)
designer = Designer("Asha", 55000)
manager = Manager("Rahul", 75000)

developer.display()
developer.code()

designer.display()
designer.design()

manager.display()
manager.manage()
````

## 31. Practical Project — Vehicle System

Hierarchy:

```text
Vehicle
   │
   ├── Car
   ├── Bike
   └── Truck
```

The parent contains common vehicle information and behavior. Child classes add their own behavior.

````python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def display(self):
        print("Brand:", self.brand)
        print("Speed:", self.speed)


class Car(Vehicle):
    def drive(self):
        print(self.brand, "car is driving.")


class Bike(Vehicle):
    def ride(self):
        print(self.brand, "bike is riding.")


class Truck(Vehicle):
    def load(self):
        print(self.brand, "truck is loading.")


car = Car("Toyota", 120)
bike = Bike("Yamaha", 100)
truck = Truck("Volvo", 90)

car.display()
car.drive()

bike.display()
bike.ride()

truck.display()
truck.load()
````

This project prepares you for polymorphism in the next notebook.

## 32. Practical Project — Student System

Hierarchy:

```text
Person
  ↓
Student
```

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)


student = Student("Komal", 22, 85)
student.display()
````
