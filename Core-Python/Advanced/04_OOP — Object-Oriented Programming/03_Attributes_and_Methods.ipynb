# 03 — Attributes and Methods

This notebook moves one step deeper into OOP: **how objects store data and how they perform behavior**.

The previous notebook introduced classes and objects. Here we focus specifically on **attributes and methods**. `__init__()`, class attributes, properties, and the detailed mechanics of `self` are reserved for their dedicated notebooks.


## 1. Introduction

The fundamental model is:

```text
Object
│
├── Attributes → Data / State
│
└── Methods    → Behavior / Actions
```

For example:

```text
Student object
├── name
├── age
├── marks
├── study()
├── attend_class()
└── display_result()
```

- **Attribute** → information associated with an object
- **Method** → function associated with an object


## 2. What Are Attributes?

An attribute is a value associated with an object.

Example:


````python
class Student:
    pass

student = Student()

student.name = "Komal"
student.age = 22
student.marks = 85
````

The object now has:

```text
student
├── name → "Komal"
├── age → 22
└── marks → 85
```

At this stage, the attributes are being created directly on the object.


## 3. Creating Object Attributes

Attributes can store different Python data types.


````python
class Student:
    pass

student = Student()

student.name = "Komal"
student.age = 22
student.marks = 85.5
student.is_passed = True

print(student.name)
print(student.age)
print(student.marks)
print(student.is_passed)
````

Attributes can also contain collections:


````python
student.subjects = ["Python", "SQL", "Machine Learning"]

print(student.subjects)
````

A well-structured class will normally initialize its expected attributes consistently. The proper initialization technique will be covered with `__init__()` later.


## 4. Accessing Object Attributes

Use dot notation:

```text
object.attribute
```

Examples:

```python
student.name
student.age
student.marks
```


````python
print(student.name)
print(student.age)
print(student.marks)
````

The same pattern applies to other objects:

```text
student.name
car.brand
account.balance
```


## 5. Modifying Attributes

An object's attributes can be changed during its lifetime.


````python
print(student.marks)

student.marks = 90

print(student.marks)
````

Output:

```text
85.5
90
```

This demonstrates that an object's state can change during its lifetime.


## 6. Adding New Attributes

Python allows attributes to be added to an existing object.


````python
student.city = "Hyderabad"

print(student.city)
````

The object can now be viewed as:

```text
student
├── name
├── age
├── marks
├── is_passed
├── subjects
└── city
```

Another object may have a different set of attributes:


````python
student2 = Student()

student2.name = "Rahul"
student2.age = 23

print(student2.name)
print(student2.age)
````

This flexibility is useful, but consistent classes should normally establish their expected object structure in one place. That will be handled properly with `__init__()`.


## 7. Deleting Attributes

Use `del` to remove an attribute.


````python
del student.city

print(hasattr(student, "city"))
print(hasattr(student, "name"))
````

The results are:

```text
False
True
```

Trying to access the deleted attribute directly would raise `AttributeError`.


````python
# Uncomment to see the error:
# print(student.city)
````

Python also provides supporting tools such as `getattr()` and `setattr()`:


````python
print(getattr(student, "name"))

setattr(student, "city", "Hyderabad")
print(student.city)
````

These are useful tools, but they are supporting concepts here rather than the main focus.


## 8. What Are Methods?

A method is a function defined inside a class and used through an object.


````python
class Student:

    def study(self):
        print("Student is studying")

student = Student()
student.study()
````

Output:

```text
Student is studying
```

Mental model:

```text
student.study()
      │
      ▼
    method
      │
      ▼
    action
```


## 9. Defining Methods

A class can define multiple methods.


````python
class Student:

    def study(self):
        print("Studying")

    def attend_class(self):
        print("Attending class")

    def take_exam(self):
        print("Taking exam")

student = Student()

student.study()
student.attend_class()
student.take_exam()
````

Methods describe what an object **can do**.


## 10. Calling Methods

The general syntax is:

```python
object.method()
```

For example:


````python
student.study()
````

The same method can be called through different objects:


````python
student1.study()
student2.study()
````

Python supplies the current object when an instance method is called. The detailed mechanics of this process are covered in `04_The_self_Parameter.ipynb`.


## 11. Methods Working with Object Data

Now combine attributes and methods.

A method can access an attribute belonging to the current object:


````python
class Student:

    def display_name(self):
        print(self.name)

student = Student()
student.name = "Komal"

student.display_name()
````

Output:

```text
Komal
```

The relationship is:

```text
student
│
├── name → "Komal"
│
└── display_name()
          │
          └── accesses the object's name
```

This is the transition from simply having attributes and methods to having **methods that operate on object state**.


## 12. Methods That Return Values

Methods do not have to print results. They can return values.


````python
class Calculator:

    def add(self, a, b):
        return a + b

calculator = Calculator()

result = calculator.add(10, 20)

print(result)
````

Output:

```text
30
```

Remember:

```text
print()
    → displays something

return
    → sends a value back to the caller
```

This connects directly with the functions section you have already studied.


## 13. Methods That Modify Object State

Methods can also change an object's state.


````python
class BankAccount:

    def deposit(self, amount):
        self.balance += amount

account = BankAccount()
account.balance = 1000

account.deposit(500)

print(account.balance)
````

Output:

```text
1500
```

The flow is:

```text
Before:
balance = 1000

deposit(500)
     ↓
balance changes

After:
balance = 1500
```

`__init__()` is intentionally not introduced here; it will be used properly in its dedicated notebook.


## 14. Attributes and Methods Together

Build a slightly more complete example.


````python
class Employee:

    def display_info(self):
        print("Name:", self.name)
        print("Department:", self.department)

    def change_department(self, new_department):
        self.department = new_department

employee = Employee()

employee.name = "Komal"
employee.department = "Data Science"

employee.display_info()

employee.change_department("Machine Learning")

employee.display_info()
````

This demonstrates a key OOP idea:

> Methods can read and modify the object's attributes.


## 15. Multiple Objects and Independent State

The same method can operate on different objects with different state.


````python
class Student:

    def display(self):
        print(self.name)

student1 = Student()
student2 = Student()

student1.name = "Komal"
student2.name = "Rahul"

student1.display()
student2.display()
````

Output:

```text
Komal
Rahul
```

Mental model:

```text
Student class
      │
  ┌───┴───┐
  ▼       ▼
student1 student2
   │         │
 name       name
 Komal      Rahul
```

The method is shared by the class, but each object has its own state.


## 16. Attribute vs Method

| Attribute | Method |
|---|---|
| Stores data/state | Performs behavior |
| Usually accessed without `()` | Called using `()` |
| `student.name` | `student.study()` |
| `car.speed` | `car.accelerate()` |
| `account.balance` | `account.deposit()` |

Example:

```python
student.name
```

is an attribute.

```python
student.study()
```

is a method.

A common mistake is writing `student.name()` when `name` contains a string rather than a callable object.


## 17. Practical Examples

### Example 1 — Rectangle


````python
class Rectangle:

    def area(self, length, width):
        return length * width

    def perimeter(self, length, width):
        return 2 * (length + width)

rectangle = Rectangle()

print(rectangle.area(10, 5))
print(rectangle.perimeter(10, 5))
````

### Example 2 — Student


````python
class Student:

    def display_result(self):
        print("Student:", self.name)
        print("Marks:", self.marks)

student = Student()

student.name = "Komal"
student.marks = 85

student.display_result()
````

### Example 3 — Bank Account


````python
class BankAccount:

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance")

account = BankAccount()
account.balance = 5000

account.deposit(1000)
account.withdraw(2000)

print(account.balance)
````

These examples prepare the learner for later improvements using `__init__()` and encapsulation.


## 18. Common Mistakes

### Mistake 1 — Forgetting parentheses when calling a method

Incorrect:

```python
student.study
```

This refers to the method object.

Correct:

```python
student.study()
```

### Mistake 2 — Confusing an attribute and a method

Incorrect:

```python
student.name()
```

if `name` is a string.

Correct:

```python
student.name
```

### Mistake 3 — Forgetting `self` when accessing object data

Inside an instance method:

```python
def display(self):
    print(self.name)
```

not:

```python
def display(self):
    print(name)
```

The detailed meaning of `self` comes next.

### Mistake 4 — Assuming objects share instance data

```python
student1.name = "Komal"
student2.name = "Rahul"
```

These are separate object states.

### Mistake 5 — Creating inconsistent attributes manually

For example:

```python
student1.name = "Komal"
student2.age = 22
student3.city = "Hyderabad"
```

Python allows this flexibility, but it is usually not how a well-designed class should be structured. Later, `__init__()` will provide a consistent initialization approach.
