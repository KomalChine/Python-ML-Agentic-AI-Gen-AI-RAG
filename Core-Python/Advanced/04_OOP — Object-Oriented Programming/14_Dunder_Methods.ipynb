## Learning Goal

Dunder means **double underscore**. Dunder methods are special methods Python uses to define how custom objects behave with built-in operations such as `print()`, `len()`, comparisons, arithmetic operators, indexing, iteration, and function-call syntax.

This notebook shows how to make your own classes behave naturally like Python's built-in types.

# 1. Introduction to Dunder Methods

Dunder methods are special methods whose names begin and end with double underscores. They are also called special methods or, informally, magic methods.

Examples: `__init__()`, `__str__()`, `__repr__()`, `__len__()`, `__add__()`, and `__eq__()`.

Python uses these methods to define how objects respond to built-in operations.

````python
numbers = [10, 20, 30]
print(len(numbers))
````

# 2. Ordinary Methods vs Dunder Methods

An ordinary method is normally called directly. A dunder method is often called by Python when an operation is performed.

````python
class Student:
    def study(self):
        print("Student is studying")

    def __str__(self):
        return "Student object"


student = Student()
student.study()
print(student)
````

- `study()` is an ordinary method that we call directly.
- `__str__()` is called when Python converts the object to a user-friendly string, such as through `str(student)` or `print(student)`.

You can implement supported dunder methods yourself, but you generally should not invent new dunder names. Python gives special meaning to particular names.

# 3. `__str__()` — User-Friendly Object Representation

Without `__str__()`, printing a custom object usually produces a default representation that is not meaningful to a user.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks


student = Student("Komal", 85)
print(student)
````

Now define `__str__()` to provide a readable display.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"Student: {self.name}, Marks: {self.marks}"


student = Student("Komal", 85)
print(student)
````

Output:

```text
Student: Komal, Marks: 85
```

Important: `__str__()` should **return** a string, not print one.

````python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name


print(Student("Komal"))
````

# 4. `__repr__()` — Developer-Friendly Representation

`__repr__()` provides an informative representation useful for debugging and inspecting objects.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __repr__(self):
        return f"Student(name={self.name!r}, marks={self.marks!r})"


student = Student("Komal", 85)
print(repr(student))
````

Output:

```text
Student(name='Komal', marks=85)
```

The `!r` formatting conversion uses the `repr()` representation of each value.

| Method | Main purpose |
|---|---|
| `__str__()` | Readable representation for users |
| `__repr__()` | Informative representation for debugging and developers |

If `__str__()` is not defined, Python can fall back to `__repr__()` when one is available.

# 5. `__len__()` — Custom Length

Implement `__len__()` to define how `len(object)` behaves. The result should be a non-negative integer.

````python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)


team = Team(["Amit", "Priya", "Komal"])
print(len(team))
````

# 6. `__bool__()` — Truth-Value Testing

`__bool__()` customizes `bool(object)` and how an object behaves in conditions.

````python
class ShoppingCart:
    def __init__(self, items):
        self.items = items

    def __bool__(self):
        return len(self.items) > 0


cart1 = ShoppingCart(["Laptop"])
cart2 = ShoppingCart([])

print(bool(cart1))
print(bool(cart2))

if cart1:
    print("Cart has items")
````

If a class defines neither `__bool__()` nor `__len__()`, its instances are normally truthy.

# 7. `__eq__()` — Equality Comparison

Separate instances of a custom class are generally compared by identity unless equality behavior is customized.

````python
class Student:
    def __init__(self, student_id):
        self.student_id = student_id

    def __eq__(self, other):
        if not isinstance(other, Student):
            return NotImplemented
        return self.student_id == other.student_id


student1 = Student(101)
student2 = Student(101)
student3 = Student(102)

print(student1 == student2)
print(student1 == student3)
````

Here, students are considered equal when their IDs match.

Returning `NotImplemented` for an unsupported comparison allows Python to try the appropriate alternative comparison behavior.

# 8. `__add__()` — Operator Overloading

Operator overloading lets a class define what an operator means for its objects. The `+` operator can add numbers or concatenate strings; a custom class can define its own addition behavior through `__add__()`.

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        return Point(
            self.x + other.x,
            self.y + other.y
        )

    def __repr__(self):
        return f"Point({self.x}, {self.y})"


p1 = Point(2, 3)
p2 = Point(4, 5)

print(p1 + p2)
````

Conceptually, `p1 + p2` invokes Python's addition protocol, which normally tries `p1.__add__(p2)`.

# 9. Other Comparison Methods

| Operator | Special method |
|---|---|
| `==` | `__eq__()` |
| `!=` | `__ne__()` |
| `<` | `__lt__()` |
| `<=` | `__le__()` |
| `>` | `__gt__()` |
| `>=` | `__ge__()` |

Implement `__lt__()` to compare products by price.

````python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __lt__(self, other):
        if not isinstance(other, Product):
            return NotImplemented
        return self.price < other.price


p1 = Product("Mouse", 500)
p2 = Product("Keyboard", 1200)

print(p1 < p2)
````

This can also help when sorting objects, provided the comparison behavior is defined appropriately.

````python
products = [
    Product("Monitor", 8000),
    Product("Mouse", 500),
    Product("Keyboard", 1200)
]

products.sort()
print([product.name for product in products])
````

# 10. `__getitem__()` — Indexing

This method allows an object to support indexing such as `object[0]`.

````python
class DataCollection:
    def __init__(self, data):
        self.data = data

    def __getitem__(self, index):
        return self.data[index]


data = DataCollection([10, 20, 30])

print(data[0])
print(data[1])
print(data[0:2])
````

Slices work here because the internal list supports slicing.

# 11. `__iter__()` and `__next__()` — Iteration

You have already studied iterators and generators, so this section connects to that knowledge rather than repeating the entire iterator chapter.

- `__iter__()` obtains or returns an iterator.
- `__next__()` obtains the next item from an iterator.
- `StopIteration` signals that iteration has finished.

````python
class NumberCollection:
    def __init__(self, numbers):
        self.numbers = numbers

    def __iter__(self):
        return iter(self.numbers)


collection = NumberCollection([10, 20, 30])

for number in collection:
    print(number)
````

This class delegates iteration to its internal list. It does not need to implement `__next__()` itself because the list's iterator already implements the iterator protocol.

Keep deeper iterator implementation details in your Iterators and Generators notebooks.

# 12. `__call__()` — Making Objects Callable

`__call__()` allows an object to be called like a function. This is useful for stateful function-like objects.

````python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, number):
        return number * self.factor


double = Multiplier(2)

print(double(5))
print(double(10))
````

The expression `double(5)` invokes the object's call behavior. This connects to functions as objects and decorators.

# 13. `__contains__()` — Membership Testing

This method customizes the `in` operator.

````python
class NumberGroup:
    def __init__(self, numbers):
        self.numbers = numbers

    def __contains__(self, number):
        return number in self.numbers


group = NumberGroup([10, 20, 30])

print(20 in group)
print(50 in group)
````

Python can also perform membership testing through iteration when a suitable `__contains__()` implementation is absent.

# 14. `__del__()` — Object Finalization

`__del__()` may be called when an object is finalized, but its timing is not guaranteed across Python implementations or program situations.

It is **not** a reliable way to manage files or other important resources. Prefer context managers and `with` statements for resource cleanup; those belong in your Context Managers section.

````python
class Demo:
    def __del__(self):
        print("Finalizer called")


demo = Demo()
del demo
````

The output may depend on when the object is finalized. Treat this only as a demonstration, not as a recommended cleanup pattern.

# 16. Practical Project — Custom Shopping Cart

Build a `ShoppingCart` class that supports:

- `__init__()` to initialize items
- `__len__()` to return the number of items
- `__bool__()` to check whether the cart has items
- `__str__()` to display the cart
- `__add__()` to combine two carts
- `__contains__()` to check whether a product is present

````python
class ShoppingCart:
    def __init__(self, items=None):
        self.items = list(items) if items is not None else []

    def __len__(self):
        return len(self.items)

    def __bool__(self):
        return len(self.items) > 0

    def __str__(self):
        return f"ShoppingCart({self.items})"

    def __contains__(self, item):
        return item in self.items

    def __add__(self, other):
        if not isinstance(other, ShoppingCart):
            return NotImplemented
        return ShoppingCart(self.items + other.items)


cart1 = ShoppingCart(["Mouse", "Keyboard"])
cart2 = ShoppingCart(["Monitor"])

print(cart1)
print(len(cart1))
print("Mouse" in cart1)
print(bool(cart2))

combined_cart = cart1 + cart2
print(combined_cart)
print(len(combined_cart))
````

The same object can now work naturally with `print()`, `len()`, `bool()`, `in`, and `+` because those behaviors are defined by special methods.
