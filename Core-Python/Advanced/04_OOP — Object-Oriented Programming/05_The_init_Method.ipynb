# 05 — The `__init__()` Method

So far, we have created objects and then manually assigned their attributes. The `__init__()` method provides a cleaner way to initialize an object's attributes automatically when the object is created.

> **Core idea:** `__init__()` is a special method that Python calls automatically when an object is initialized.

## 1. Introduction

Without `__init__()`, object data has to be assigned manually:

```python
class Student:
    pass

student = Student()

student.name = "Komal"
student.age = 22
student.marks = 85
```

This works, but it is inconvenient.

With `__init__()`:

```python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks
```

we can write:

```python
student = Student("Komal", 22, 85)
```

The object is initialized immediately.

## 2. What Is `__init__()`?

`__init__()` is a special method that Python calls automatically when an object is initialized.

````python
class Student:
    def __init__(self):
        print("Student object initialized")

student = Student()
````

Notice that we did not explicitly call:

```python
student.__init__()
```

Python invokes `__init__()` as part of normal object initialization.

## 3. Why Do We Need `__init__()`?

Compare the two approaches.

### Without `__init__()`

```python
student = Student()

student.name = "Komal"
student.age = 22
student.marks = 85
```

### With `__init__()`

```python
student = Student("Komal", 22, 85)
```

Using `__init__()`:

- makes object creation cleaner
- ensures required data is supplied
- creates a consistent object structure
- reduces repetitive code
- makes classes easier to use

### Mental model

```text
Student(...)
    ↓
__init__()
    ↓
initialize attributes
    ↓
ready-to-use object
```

## 4. Basic `__init__()` Syntax

Start with the simplest form:

````python
class Student:
    def __init__(self):
        print("Object created")

student = Student()
````

We can then use `__init__()` to initialize attributes with default values.

````python
class Student:
    def __init__(self):
        self.name = "Unknown"
        self.age = 0

student = Student()

print(student.name)
print(student.age)
````

Output:

```text
Unknown
0
```

## 5. `self` and `__init__()`

This connects directly with the previous notebook.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

Here:

```text
name
    ↓
value supplied by the caller

self.name
    ↓
attribute stored in the current object
```

For:

```python
student = Student("Komal", 22)
```

conceptually:

```text
self → student
name → "Komal"
age  → 22
```

Therefore:

```python
self.name = name
```

means:

> Store `"Komal"` inside the current student's `name` attribute.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student = Student("Komal", 22)

print(student.name)
print(student.age)
````

## 6. Initializing Object Attributes

A complete example can initialize several attributes at once.

````python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

student = Student("Komal", 22, 85)

print(student.name)
print(student.age)
print(student.marks)
````

The object now has its own initialized state:

```text
name  → Komal
age   → 22
marks → 85
```

## 7. `__init__()` with Multiple Parameters

`__init__()` can accept multiple parameters just like a normal function.

````python
class Employee:
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.salary = salary

employee = Employee("Komal", "Data Science", 50000)

print(employee.name)
print(employee.department)
print(employee.salary)
````

The mapping is:

```text
Employee("Komal", "Data Science", 50000)
        ↓
name       → "Komal"
department → "Data Science"
salary     → 50000
```

The corresponding attributes:

```text
self.name
self.department
self.salary
```

belong to that particular object.

## 8. Creating Multiple Objects with Different Data

One major benefit of `__init__()` is that the same class can initialize many objects with different data.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student1 = Student("Komal", 85)
student2 = Student("Rahul", 92)
student3 = Student("Priya", 78)

print(student1.name, student1.marks)
print(student2.name, student2.marks)
print(student3.name, student3.marks)
````

Mental model:

```text
Student class
    │
    ├── student1
    │     ├── name = Komal
    │     └── marks = 85
    │
    ├── student2
    │     ├── name = Rahul
    │     └── marks = 92
    │
    └── student3
          ├── name = Priya
          └── marks = 78
```

Same class, different object state.

## 9. Default Values in `__init__()`

Parameters in `__init__()` can have default values, following the same function-argument rules studied earlier.

````python
class Student:
    def __init__(self, name, age=18):
        self.name = name
        self.age = age

student1 = Student("Komal", 22)
student2 = Student("Rahul")

print(student1.age)
print(student2.age)
````

Output:

```text
22
18
```

The rule is:

```text
Required parameter
    ↓
must be supplied

Parameter with default
    ↓
can be omitted
```

## 10. `__init__()` vs Normal Methods

This distinction is important.

### `__init__()`

Used primarily to initialize the object's initial state:

```python
def __init__(self, name):
    self.name = name
```

### Normal method

Used to perform an action or operation:

```python
def display(self):
    print(self.name)
```

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(self.name, self.marks)

student = Student("Komal", 85)
student.display()
````

Mental model:

```text
Student(...)
      ↓
__init__()
      ↓
initialize object

student.display()
      ↓
normal method
      ↓
perform action
```

## 11. What Happens When an Object Is Created?

When we write:

```python
student = Student("Komal", 85)
```

the conceptual sequence is:

```text
1. Python creates a new Student instance
              ↓
2. Python invokes __init__()
              ↓
3. The new object is supplied as self
              ↓
4. "Komal" is supplied as name
              ↓
5. 85 is supplied as marks
              ↓
6. self.name and self.marks are initialized
              ↓
7. student refers to the initialized object
```

So:

```python
self.name = name
```

stores the supplied value in that object's attribute.

### Important distinction

`__init__()` **initializes** an already-created object; it is not technically the method that creates the object itself.

Object creation is associated with `__new__()`, but `__new__()` is outside the scope of this notebook.

## 12. `__init__()` and Return Values

`__init__()` should not explicitly return a value.

Incorrect:

```python
class Student:
    def __init__(self, name):
        self.name = name
        return self.name
```

This raises an error because `__init__()` must return `None`.

Correct:

```python
class Student:
    def __init__(self, name):
        self.name = name
```

If a value needs to be returned later, use a normal method.

````python
class Student:
    def __init__(self, name):
        self.name = name

    def get_name(self):
        return self.name

student = Student("Komal")
print(student.get_name())
````

## 13. Practical Classes

### Example 1 — Student

````python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)

student = Student("Komal", 22, 85)
student.display()
````

### Example 2 — Rectangle

The object remembers its dimensions through its initialized attributes.

````python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rectangle = Rectangle(10, 5)

print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
````

### Example 3 — Bank Account

This example combines `__init__()`, `self`, attributes, and normal methods.

````python
class BankAccount:
    def __init__(self, holder, balance):
        self.holder = holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance")

    def show_balance(self):
        print("Balance:", self.balance)

account = BankAccount("Komal", 5000)

account.deposit(1000)
account.withdraw(2000)
account.show_balance()
````

Output:

```text
Balance: 4000
```

This class combines:

```text
__init__()
    +
self
    +
attributes
    +
methods
```

## 14. Common Mistakes

### Mistake 1 — Forgetting `self`

Incorrect:

```python
class Student:
    def __init__(name, age):
        self.name = name
        self.age = age
```

Correct:

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

### Mistake 2 — Forgetting `self.` when creating attributes

Incorrect:

```python
def __init__(self, name):
    name = name
```

This only reassigns the local parameter.

Correct:

```python
def __init__(self, name):
    self.name = name
```

Now the value is stored on the object.

````python
class Student:
    def __init__(self, name):
        self.name = name

student = Student("Komal")
print(student.name)
````

### Mistake 3 — Wrong number of arguments

If:

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

then:

```python
student = Student("Komal")
```

fails because `age` is required.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

try:
    student = Student("Komal")
except TypeError as error:
    print(type(error).__name__ + ":", error)
````

### Mistake 4 — Calling `__init__()` manually unnecessarily

Normally:

```python
student = Student("Komal", 22)
```

is sufficient.

You generally should not use:

```python
student.__init__("Komal", 22)
```

as a normal way to initialize an object.

### Mistake 5 — Returning a value from `__init__()`

Incorrect:

```python
def __init__(self):
    return 10
```

`__init__()` must return `None`.
