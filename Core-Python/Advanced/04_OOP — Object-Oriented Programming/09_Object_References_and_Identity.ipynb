# 09 — Object References and Identity

## Advanced Python — Object-Oriented Programming

The main goal of this notebook is to understand an important Python concept:

> **Variables do not contain objects directly; they hold references to objects.**

Understanding references is essential before learning shallow and deep copying.

## 1. Introduction

In Python, variables and objects are separate concepts.

A variable gives us a name through which we can access an object. The variable refers to an object rather than acting as a box that permanently contains the object.

Understanding this relationship helps explain:

- why assignment does not automatically create a copy,
- why changing a mutable object through one reference can affect another reference,
- the difference between `is` and `==`,
- mutation versus reassignment,
- how objects can reference other objects.

## 2. Objects Live in Memory

Start with a simple class:

```python
class Student:
    def __init__(self, name):
        self.name = name

student1 = Student("Komal")
```

Conceptually:

```text
student1
   │
   ▼
┌───────────────┐
│ Student object│
│ name = Komal  │
└───────────────┘
```

`student1` is a reference to the `Student` object.

````python
class Student:
    def __init__(self, name):
        self.name = name

student1 = Student("Komal")

print(student1.name)
print(type(student1))
````

The object and the variable name are different concepts:

```text
variable/reference ─────► object
```

This distinction becomes especially important when more than one variable refers to the same object.

## 3. Variables as References

Consider a list:

```python
x = [10, 20, 30]
```

Conceptually:

```text
x ───────► [10, 20, 30]
```

The variable `x` refers to the list object.

Now assign:

```python
y = x
```

Conceptually:

```text
x ───────┐
         ▼
       [10, 20, 30]
         ▲
         │
y ───────┘
```

Both variables refer to the **same object**.

````python
x = [10, 20, 30]
y = x

print(x)
print(y)
print(x is y)
````

The important idea is:

```python
y = x
```

does not create another list. It creates another reference to the existing list.

## 4. Assignment Does Not Create a New Object

This is one of the most important ideas in this notebook.

````python
x = [10, 20, 30]
y = x

y.append(40)

print(x)
print(y)
````

Output:

```text
[10, 20, 30, 40]
[10, 20, 30, 40]
```

Why did changing `y` also change `x`?

Because:

```text
x ─────┐
       ├──► same list object
y ─────┘
```

There is only one list.

Therefore:

```python
y.append(40)
```

mutates that shared object.

## 5. Object Identity

Every object has an identity during its lifetime.

Python provides the built-in `id()` function to obtain an identity value for an object.

````python
x = [10, 20, 30]

print(id(x))
````

If two references point to the same object, they have the same identity.

````python
x = [10, 20, 30]
y = x

print(id(x))
print(id(y))
print(id(x) == id(y))
````

The identities are the same because `x` and `y` refer to the same object.

Think of:

```text
x ─────► Object A
y ─────► Object A
```

as two references to one object.

## 6. The `is` Operator

The `is` operator checks **object identity**.

Example:

```python
x = [10, 20, 30]
y = x

print(x is y)
```

Because both references point to the same object, the result is `True`.

````python
x = [10, 20, 30]
y = x

print(x is y)
````

The meaning is:

> `is` checks whether two references point to the **same object**.

It is an identity comparison, not a value comparison.

## 7. `is` vs `==`

This distinction is extremely important.

Consider two separately created lists:

```python
x = [10, 20, 30]
y = [10, 20, 30]
```

They contain the same values, but they are different list objects.

````python
x = [10, 20, 30]
y = [10, 20, 30]

print(x == y)
print(x is y)
````

The expected results are:

```text
True
False
```

The difference is:

| Operator | Checks |
|---|---|
| `==` | Value/equality |
| `is` | Object identity |

Conceptually:

```text
x ───► [10, 20, 30]

y ───► [10, 20, 30]
```

The contents are equal, but the objects are different.

Therefore:

```python
x == y     # True
x is y     # False
```

## 8. References with Objects

The same reference behavior applies to objects created from classes.

````python
class Student:
    def __init__(self, name):
        self.name = name

student1 = Student("Komal")
student2 = student1

print(student1 is student2)
````

The result is:

```text
True
```

Both names refer to the same `Student` object.

```text
student1 ─────┐
              ▼
        Student object
              ▲
              │
student2 ─────┘
```

## 9. Modifying an Object Through Another Reference

Because `student1` and `student2` refer to the same object, changing the object through either reference changes the same object.

````python
class Student:
    def __init__(self, name):
        self.name = name

student1 = Student("Komal")
student2 = student1

student2.name = "Rahul"

print(student1.name)
print(student2.name)
````

Both print statements show:

```text
Rahul
```

There is only **one `Student` object**.

Changing its `name` through `student2` changes the same object that `student1` refers to.

## 10. Creating Separate Objects

Now compare that with creating two objects separately.

````python
class Student:
    def __init__(self, name):
        self.name = name

student1 = Student("Komal")
student2 = Student("Komal")

print(student1 is student2)
print(student1.name)
print(student2.name)
````

The result of:

```python
student1 is student2
```

is:

```text
False
```

Even though both objects contain the same data, they are two different objects.

```text
student1 ───► Student("Komal")

student2 ───► Student("Komal")
```

## 11. Object Identity vs Object Equality

Consider two separate `Student` objects with the same attribute values.

````python
class Student:
    def __init__(self, name):
        self.name = name

student1 = Student("Komal")
student2 = Student("Komal")

print(student1 == student2)
print(student1 is student2)
````

By default, separately created objects are not considered equal merely because their attributes contain the same values.

So the important distinction is:

```text
Identity
   ↓
Are these the same object?

Equality
   ↓
Are these considered equal in value?
```

Object equality can later be customized using **dunder methods**, which will be studied in the OOP section on dunder methods.

## 12. References in Function Arguments

References also matter when objects are passed to functions.

Consider:

````python
class Student:
    def __init__(self, name):
        self.name = name

def change_name(student):
    student.name = "Rahul"

student = Student("Komal")

change_name(student)

print(student.name)
````

The output is:

```text
Rahul
```

The function receives a reference to the same object.

Conceptually:

```text
student ─────────────────► Student object
                            ▲
                            │
function parameter ────────┘
```

Therefore, changing the object's attribute through the function parameter changes the same object.

## 13. Mutable Objects and References

Lists are mutable objects.

If a list is passed to a function and the function mutates it, the change is visible through the original reference.

````python
numbers = [1, 2, 3]

def add_number(values):
    values.append(4)

add_number(numbers)

print(numbers)
````

Output:

```text
[1, 2, 3, 4]
```

The function parameter `values` refers to the same list object as `numbers`.

```text
numbers ───────────► [1, 2, 3, 4]
                       ▲
                       │
values ────────────────┘
```

## 14. Immutable Objects and References

Now consider an integer.

Integers are immutable.

````python
x = 10
y = x

y = 20

print(x)
print(y)
````

Output:

```text
10
20
```

The assignment:

```python
y = 20
```

does not modify the integer object `10`.

Instead, `y` is reassigned so that it refers to another integer object.

Conceptually:

```text
Before:

x ─────┐
       ├──► 10
y ─────┘


After y = 20:

x ─────► 10

y ─────► 20
```

## 15. Mutable vs Immutable Reference Behavior

Some common Python types are:

| Type | Mutable? | Example |
|---|---|---|
| `int` | No | `10` |
| `float` | No | `10.5` |
| `str` | No | `"Python"` |
| `tuple` | No | `(1, 2)` |
| `list` | Yes | `[1, 2]` |
| `dict` | Yes | `{"a": 1}` |
| `set` | Yes | `{1, 2}` |

The important point here is not simply memorizing the table.

You should understand how **mutation** and **reassignment** interact with references.

## 16. Reassignment vs Mutation

These two operations are different.

### Mutation

```python
x = [1, 2, 3]
y = x

y.append(4)
```

The existing list object changes.

```text
x ─────┐
       ├──► [1, 2, 3, 4]
y ─────┘
```

### Reassignment

```python
x = [1, 2, 3]
y = x

y = [10, 20]
```

Now `y` refers to a different object.

```text
x ─────► [1, 2, 3]

y ─────► [10, 20]
```

This distinction is fundamental when working with object references.

````python
# Mutation
x = [1, 2, 3]
y = x

y.append(4)

print("After mutation:")
print("x:", x)
print("y:", y)

# Reassignment
x = [1, 2, 3]
y = x

y = [10, 20]

print("\nAfter reassignment:")
print("x:", x)
print("y:", y)
````

## 17. References Between Object Attributes

An object attribute can itself contain a reference to another object.

For example, a `Student` can have an `Address`.

````python
class Address:
    def __init__(self, city):
        self.city = city


class Student:
    def __init__(self, name, address):
        self.name = name
        self.address = address


address = Address("Hyderabad")
student = Student("Komal", address)

print(student.name)
print(student.address.city)
````

Conceptually:

```text
student
   │
   ▼
Student object
   │
   └── address ───► Address object
                         │
                         └── city = Hyderabad
```

This is an example of objects being connected through references.

## 18. Multiple References to the Same Object

Different objects can contain references to the same object.

For example:

````python
class Address:
    def __init__(self, city):
        self.city = city


class Student:
    def __init__(self, name, address):
        self.name = name
        self.address = address


address = Address("Hyderabad")

student1 = Student("A", address)
student2 = Student("B", address)

student1.address.city = "Mumbai"

print(student1.address.city)
print(student2.address.city)
````

Both students now show:

```text
Mumbai
```

because both `address` attributes refer to the same `Address` object.

```text
student1 ──► Student A ──┐
                          │
                          ▼
                      Address
                          │
student2 ──► Student B ──┘
```

This is a useful real-world example of shared object references.

## 19. Reference Diagrams

Reference diagrams are useful for understanding what Python is doing.

For one reference:

```text
student1 ─────► Student object
```

For two references to one object:

```text
student1 ──────┐
               │
student2 ──────┼──► Student object
               │
student3 ──────┘
```

For two separate objects:

```text
student1 ─────► Student object A

student2 ─────► Student object B
```

When debugging reference-related behavior, drawing these relationships can make the situation much easier to understand.

## 20. Garbage Collection — Basic Idea

Only the basic idea is needed here.

When an object no longer has references pointing to it, it becomes unreachable.

For example:

```python
x = [1, 2, 3]
y = x

x = None
y = None
```

After both references are removed, the list is no longer reachable through those variables.

Python has memory-management mechanisms that can eventually reclaim memory occupied by objects that are no longer reachable.

A detailed study of garbage collection is outside the scope of this notebook.

````python
x = [1, 2, 3]
y = x

print("Before removing references:")
print(x)
print(y)

x = None
y = None

print("\nReferences have been removed.")
````

The important idea is simply:

```text
References exist
      ↓
Object is reachable

No references remain
      ↓
Object becomes unreachable
```

## 21. Connection to Copying Objects

This concept prepares you for the later topic of **shallow and deep copying**.

Consider:

```python
student2 = student1
```

This is **not copying**.

It creates another reference to the same object.

A simplified comparison is:

```text
Assignment
    ↓
same object

Shallow copy
    ↓
new outer object
but references may be shared

Deep copy
    ↓
new object
and nested objects are copied
```

The detailed `copy()` and `deepcopy()` operations belong to the later copying notebook.

## 22. Common Mistakes

### Mistake 1 — Thinking assignment creates a copy

```python
y = x
```

does not automatically create a copy.

It creates another reference.

### Mistake 2 — Using `is` to compare values

If the intention is value comparison, use:

```python
x == y
```

rather than:

```python
x is y
```

`is` is for identity.

### Mistake 3 — Not understanding mutation vs reassignment

These are different:

```python
y.append(4)
```

and:

```python
y = [10, 20]
```

The first mutates an existing object. The second reassigns the variable.

### Mistake 4 — Assuming equal data means the same object

Two separately created objects can contain identical data while having different identities.

### Mistake 5 — Forgetting that attributes can contain references

An attribute such as:

```python
student.address
```

can itself refer to another object.

This means several objects can potentially share a reference to the same nested object.

## 23. Summary

You should now understand:

- Objects exist independently from variable names.
- Variables hold references to objects.
- Assignment creates another reference; it does not automatically copy an object.
- `id()` can show an object's identity.
- `is` checks object identity.
- `==` checks equality.
- Multiple variables can reference one object.
- Multiple objects can contain identical values while still being different objects.
- Mutation and reassignment are different operations.
- Mutable and immutable objects behave differently when references are involved.
- Function parameters can refer to the same object.
- Object attributes can reference other objects.
- Multiple objects can share references to the same nested object.
- Understanding references is essential before learning shallow and deep copying.
