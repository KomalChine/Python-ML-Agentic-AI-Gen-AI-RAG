## 1. Introduction

Start with a simple class:

```python
class Student:
    school = "ABC College"

    def __init__(self, name):
        self.name = name

    def display(self):
        print(self.name)
```

`display()` is an **instance method**.

Python also provides:

```python
@classmethod
```

and:

```python
@staticmethod
```

These allow methods to have different relationships with the class and its objects.

## 2. Review: Methods and `self`

An instance method normally looks like:

```python
class Student:
    def display(self):
        print(self.name)
```

Calling:

```python
student.display()
```

conceptually supplies:

```text
student → self
```

Therefore:

```python
self.name
```

means the `name` attribute of the current object.

```text

```

## 3. Instance Methods

An **instance method** works with a particular object.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

student = Student("Komal", 85)
student.display()
````

The method receives:

```text
self → student
```

Instance methods are appropriate when the operation depends on the **specific object's state**.

## 4. What Instance Methods Can Access

Instance methods can access:

- instance attributes
- class attributes
- other instance methods
- class-level functionality

Example:

````python
class Student:
    school = "ABC College"

    def __init__(self, name):
        self.name = name

    def display(self):
        print(self.name)
        print(self.school)

student = Student("Komal")
student.display()
````

Here:

```text
self.name
    ↓
instance attribute

self.school
    ↓
class attribute found through the instance
```

Examples of instance-method operations include:

```text
student.display()
account.withdraw()
employee.calculate_salary()
```

## 5. Class Methods

A **class method** receives the class itself rather than an individual object.

Syntax:

```python
class Student:
    @classmethod
    def method_name(cls):
        ...
```

The conventional first parameter is:

```python
cls
```

where:

```text
cls → current class
```

## 6. The `@classmethod` Decorator

Example:

````python
class Student:
    school = "ABC College"

    @classmethod
    def show_school(cls):
        print(cls.school)

Student.show_school()
````

The flow is:

```text
Student.show_school()
        ↓
cls → Student
        ↓
cls.school
        ↓
Student.school
```

## 7. The `cls` Parameter

Just as `self` conventionally refers to an instance, `cls` conventionally refers to the class.

```text
Instance method
       ↓
self → object
```

versus:

```text
Class method
       ↓
cls → class
```

````python
class Employee:
    company = "DataTech"

    def show_employee(self):
        print(self)

    @classmethod
    def show_company(cls):
        print(cls)

employee = Employee()

employee.show_employee()
Employee.show_company()
````

Conceptually:

```text
employee.show_employee()
        ↓
self → employee

Employee.show_company()
        ↓
cls → Employee
```

## 8. What Class Methods Can Access

Class methods can directly work with class-level data.

Example:

````python
class Employee:
    company = "DataTech"

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

print(Employee.company)

Employee.change_company("NewTech")

print(Employee.company)
````

This is a good use of a class method because the operation changes **class-level state**, rather than one particular employee.

## 9. Alternative Constructors

One of the most useful applications of class methods is creating **alternative constructors**.

Suppose the normal constructor is:

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

Normally:

```python
student = Student("Komal", 22)
```

But suppose the data arrives as:

```text
"Komal,22"
```

A class method can provide another way to construct the object.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, data):
        name, age = data.split(",")
        return cls(name, int(age))

student = Student.from_string("Komal,22")

print(student.name)
print(student.age)
````

The flow is:

```text
Student.from_string(...)
        ↓
cls → Student
        ↓
cls(name, age)
        ↓
Student(name, age)
        ↓
new object
```

This is called an **alternative constructor**.

Other examples could include:

```python
Student.from_dict(...)
Student.from_csv(...)
Employee.from_json(...)
```

They belong naturally in this topic because they demonstrate why class methods receive `cls`.

## 10. Static Methods

A **static method** does not automatically receive:

```text
self
```

or:

```text
cls
```

Syntax:

```python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b
```

````python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
````

Output:

```text
30
```

There is no automatically supplied object or class reference because the operation does not need one.

## 11. The `@staticmethod` Decorator

Example:

````python
class MathUtils:
    @staticmethod
    def is_even(number):
        return number % 2 == 0

print(MathUtils.is_even(10))
print(MathUtils.is_even(7))
````

Output:

```text
True
False
```

There is no `self` and no `cls` automatically supplied.

## 12. What Static Methods Can Access

A static method works using explicitly supplied arguments.

Example:

```python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b
```

Static methods are useful for utility operations that logically belong to a class but don't need object or class state.

Examples:

```text
validate_email()
convert_temperature()
is_even()
calculate_percentage()
format_name()
```

## 13. Comparing the Three Method Types

| Feature | Instance Method | Class Method | Static Method |
|---|---|---|---|
| Decorator | None | `@classmethod` | `@staticmethod` |
| First parameter | `self` | `cls` | None automatically |
| Receives object automatically | Yes | No | No |
| Receives class automatically | Through instance/class lookup | Yes | No |
| Usually works with | Object state | Class state | Independent inputs |
| Common use | Object behavior | Class behavior / alternate constructors | Utility behavior |

### Mental model

```text
Instance method
    ↓
"What should THIS object do?"

Class method
    ↓
"What should THIS class do?"

Static method
    ↓
"What useful operation belongs logically here,
 but doesn't need object/class state?"
```

## 14. Choosing the Right Method

### Use an instance method when

The operation needs data from a particular object.

Example:

```python
account.withdraw(500)
```

because the operation needs that account's balance.

### Use a class method when

The operation needs class-level information or creates objects in an alternative way.

Example:

```python
Student.from_string("Komal,22")
```

### Use a static method when

The operation doesn't need `self` or `cls`.

Example:

```python
MathUtils.is_even(10)
```

## 15. Practical Example — Employee

Create a class combining all three method types.

````python
class Employee:
    company = "DataTech"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    # Instance method
    def display(self):
        print(self.name, self.salary)

    # Class method
    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

    # Static method
    @staticmethod
    def is_valid_salary(salary):
        return salary >= 0

employee = Employee("Komal", 50000)

employee.display()
Employee.change_company("NewTech")
print(Employee.company)
print(Employee.is_valid_salary(50000))
````

This one class demonstrates three different relationships:

```text
Instance method → works with employee data
Class method    → works with company/class data
Static method   → validates an independent input
```

## 16. Another Practical Example — Student

This example includes an instance method, class methods, an alternative constructor, and a static method.

````python
class Student:
    school = "ABC College"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Instance method
    def display(self):
        print(self.name, self.age)

    # Class method
    @classmethod
    def change_school(cls, school):
        cls.school = school

    # Alternative constructor
    @classmethod
    def from_string(cls, data):
        name, age = data.split(",")
        return cls(name, int(age))

    # Static method
    @staticmethod
    def is_adult(age):
        return age >= 18

student = Student("Komal", 22)
student.display()

Student.change_school("XYZ College")

student2 = Student.from_string("Rahul,23")

print(student2.name, student2.age)
print(Student.is_adult(22))
````

## 17. Common Mistakes

### Mistake 1 — Forgetting `@classmethod`

Incorrect:

```python
class Student:
    def show_school(cls):
        print(cls.school)
```

Without `@classmethod`, this is an ordinary method from Python's binding perspective.

Correct:

```python
class Student:
    @classmethod
    def show_school(cls):
        print(cls.school)
```

### Mistake 2 — Using `self` in a class method

The conventional parameter is `cls`:

```python
@classmethod
def change_school(cls, school):
    ...
```

### Mistake 3 — Assuming `@staticmethod` supplies `cls`

A static method does not automatically receive the class.

If you write:

```python
@staticmethod
def show(cls):
    ...
```

then `cls` is simply an explicitly supplied argument. It is not automatically provided by `@staticmethod`.

### Mistake 4 — Using a static method when object state is required

If an operation needs:

```python
self.balance
```

an instance method is generally appropriate.

### Mistake 5 — Using an instance method when no instance state is needed

For a utility such as:

```python
def is_even(number):
    return number % 2 == 0
```

a static method may be appropriate if keeping the utility logically grouped under the class is useful.

```text

       Method       Method      Method
          │           │           │
        self         cls       neither
          │           │           │
          ▼           ▼           ▼
       Object        Class       Inputs
        state         state      supplied
```

### Instance method

```python
def display(self):
    ...
```

Use when working with an object's state.

### Class method

```python
@classmethod
def change_company(cls):
    ...
```

Use when working with class-level state or creating alternative constructors.

### Static method

```python
@staticmethod
def is_even(number):
    ...
```

Use for operations that don't need an automatically supplied object or class.
