## 1. Introduction

The basic relationship is:

```text
Class → blueprint
Object → instance created from that blueprint
```

In this notebook we learn:

- how to define a class
- how to create objects
- how objects access attributes and methods
- how multiple objects can be created from one class
- how each object can maintain its own state


## 2. What Is a Class?

A class is a structure used to define the common attributes and behavior that its objects can have.

Example:


````python
class Student:
    pass
````

Here, `Student` is a class.

```text
Student
   ↓
 class
```

`pass` means that the class currently has no implementation.

The class itself is not yet a particular student.


## 3. Creating a Class

Basic class syntax:


````python
class Student:
    def study(self):
        print("Student is studying")
````

Breakdown:

```text
class
  ↓
Student
  ↓
class name
```

The `study()` function is a method belonging to the class.

For now, do not worry deeply about `self`. It refers to the current object and will be covered properly in `04_The_self_Parameter.ipynb`.


## 4. What Is an Object?

An object is an **instance of a class**.

For:

```python
class Student:
    pass
```

we can create an object:


````python
class Student:
    pass

student1 = Student()
````

The process is:

```text
Student
   ↓
 class
   ↓
Student()
   ↓
object created
   ↓
student1
```

The object is now stored in the variable `student1`.


## 5. Creating Objects

Create multiple objects from the same class:


````python
class Student:
    pass

student1 = Student()
student2 = Student()
````

The relationship is:

```text
Student class
      │
  ┌───┴───┐
  ▼       ▼
student1 student2
 object    object
```

Both objects come from the same class.


## 6. Multiple Objects from One Class

A class can define behavior that can be used by many objects.


````python
class Student:
    def study(self):
        print("Studying")

student1 = Student()
student2 = Student()

student1.study()
student2.study()
````

Both objects can use the same method because the method is defined by their common class.

> One class can be used to create many independent objects.


## 7. Accessing Class Members

We can define simple class-level attributes:


````python
class Student:
    name = "Unknown"
    course = "CSE"

student1 = Student()

print(student1.name)
print(student1.course)
````

Output:

```text
Unknown
CSE
```

These are class-level attributes for now.

A detailed comparison of **instance attributes and class attributes** belongs in `06_Instance_vs_Class_Attributes.ipynb`.


## 8. Adding Methods to a Class

A class can contain methods representing behavior.


````python
class Calculator:

    def add(self):
        print(10 + 20)

    def subtract(self):
        print(20 - 10)

calculator = Calculator()

calculator.add()
calculator.subtract()
````

The object can be viewed as:

```text
calculator
    │
    ├── add()
    └── subtract()
```

Methods represent the behavior of an object.


## 9. Calling Methods Through Objects

Example:


````python
class Dog:

    def bark(self):
        print("Woof!")

dog1 = Dog()

dog1.bark()
````

Conceptually:

```text
dog1.bark()
    ↓
Python finds bark()
    ↓
dog1 is the current object
    ↓
method executes
```

The detailed mechanics of `self` are intentionally reserved for the dedicated `self` notebook.


## 10. Understanding Object State

Different objects can have different data.

For example:


````python
class Student:
    pass

student1 = Student()
student2 = Student()

student1.name = "Komal"
student2.name = "Rahul"

print(student1.name)
print(student2.name)
````

Output:

```text
Komal
Rahul
```

The objects are separate:

```text
student1
   ↓
name = "Komal"

student2
   ↓
name = "Rahul"
```

This is an introductory demonstration. The proper way to initialize instance data will be taught with `__init__()` later.


## 11. Class as a Blueprint

Use this mental model:

```text
                 Student
                  CLASS
             ┌─────────────┐
             │ name        │
             │ age         │
             │ study()     │
             └─────────────┘
                    │
          ┌─────────┼─────────┐
          ▼         ▼         ▼
      student1  student2  student3
       OBJECT     OBJECT    OBJECT
```

The class defines common structure and behavior.

Each object represents an individual instance.


## 12. Object Identity

Different objects are separate objects.

Example:


````python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1 is student2)
````

Output:

```text
False
```

The two variables refer to different objects.

You can also inspect their identity values:


````python
print(id(student1))
print(id(student2))
````

`id()` gives an identity value for an object during its lifetime. Do not rely on the exact numeric value itself.


## 13. Checking Object Type

Use `type()` to inspect an object's type:


````python
class Student:
    pass

student = Student()

print(type(student))
````

The output will look similar to:

```text
<class '__main__.Student'>
```


````python
print(type(student) == Student)
````

Output:

```text
True
```

We can also use `isinstance()`:


````python
print(isinstance(student, Student))
````

Output:

```text
True
```

Briefly:

```text
type()
    → tells you the object's type

isinstance()
    → checks whether an object is an instance of a class
```

`isinstance()` becomes especially useful when inheritance is introduced later.


## 14. Creating Objects with Different Data

Before learning `__init__()`, we can temporarily assign attributes after creating objects.


````python
class Employee:
    pass

employee1 = Employee()
employee2 = Employee()

employee1.name = "Komal"
employee1.department = "Data Science"

employee2.name = "Rahul"
employee2.department = "Development"

print(employee1.name)
print(employee1.department)

print(employee2.name)
print(employee2.department)
````

Mental model:

```text
Employee class
      │
      ├───────────────┐
      ▼               ▼
 employee1         employee2
      │               │
      ├─ name         ├─ name
      └─ department   └─ department
```

This works, but manually assigning attributes after object creation is inconvenient. Better initialization techniques will be introduced later.


## 15. Practical Examples

### Example 1 — Student


````python
class Student:

    def study(self):
        print("Student is studying")

student = Student()
student.study()
````

### Example 2 — Car


````python
class Car:

    def start(self):
        print("Car started")

    def stop(self):
        print("Car stopped")

car = Car()

car.start()
car.stop()
````

````python
class BankAccount:

    def show_message(self):
        print("This is a bank account")

account = BankAccount()
account.show_message()
````

These examples focus only on class/object syntax. Real deposit and withdrawal logic will be introduced when the relevant OOP concepts are covered.


## 16. Common Mistakes

### Mistake 1 — Forgetting parentheses when creating an object

Incorrect:

```python
student = Student
```

This stores the **class itself**.

Correct:

```python
student = Student()
```

This creates an **object**.

### Mistake 2 — Confusing the class and object

```text
Student  → class
student1 → object
```

### Mistake 3 — Thinking two objects are the same

```python
student1 = Student()
student2 = Student()

print(student1 is student2)
```

The result is:

```text
False
```

They are separate instances.

### Mistake 4 — Treating `self` as a special keyword

`self` is the conventional parameter name used for the current instance. Its detailed behavior belongs in `04_The_self_Parameter.ipynb`.
