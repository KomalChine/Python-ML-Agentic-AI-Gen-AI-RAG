# 08 — Properties and Getters/Setters

## Advanced Python — Object-Oriented Programming

Properties provide a controlled, attribute-like interface for accessing and modifying object data.

In this notebook, we will learn:

- Traditional getters and setters
- The `@property` decorator
- Property getters, setters, and deleters
- Read-only properties
- Computed properties
- Validation through setters
- Internal storage using `_attribute`
- Properties and encapsulation

## 1. Introduction

When working with objects, we often want to control how an attribute is accessed or modified.

For example, suppose a student's marks must always be between `0` and `100`.

With direct attribute access, Python allows:

```python
student.marks = -20
```

A **property** lets us keep simple attribute-style syntax while adding controlled behavior such as validation, calculation, or deletion.

## 2. Direct Attribute Access — The Problem

Consider this class:

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
```

We can create and modify an object directly.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student = Student("Komal", 85)

print(student.marks)

student.marks = -20
print(student.marks)
````

Python does not automatically know that `-20` is an invalid mark.

This motivates **controlled access** to attributes.

Instead of allowing every assignment directly, we can place rules around reading and changing the value.

## 3. What Is a Getter?

A **getter** is a method used to retrieve an attribute's value.

Example:

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def get_marks(self):
        return self.marks
```

The method `get_marks()` retrieves the marks.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def get_marks(self):
        return self.marks

student = Student("Komal", 85)

print(student.get_marks())
````

The important idea is:

```text
get_marks()
     ↓
retrieve the stored value
```

A getter is especially useful when reading an attribute requires some controlled behavior.

## 4. What Is a Setter?

A **setter** is a method used to modify an attribute's value, usually with validation.

For example, we can allow marks to change only when they are between `0` and `100`.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.marks = marks
        else:
            print("Invalid marks")

student = Student("Komal", 85)

student.set_marks(90)
print(student.marks)

student.set_marks(150)
print(student.marks)
````

Validation is useful because it prevents an object from entering an invalid state.

The setter acts as a controlled point through which a value must pass before it is stored.

## 5. The Problem with Traditional Getters and Setters

The traditional approach requires method calls:

```python
student.get_marks()
student.set_marks(90)
```

Python provides a cleaner approach where controlled behavior can still look like normal attribute access.

This leads to **properties**.

The goal is to let the user write:

```python
student.marks
student.marks = 90
```

while Python performs additional logic internally.

## 6. What Is `@property`?

The `@property` decorator turns a method into a property.

Example:

````python
class Student:
    def __init__(self, marks):
        self._marks = marks

    @property
    def marks(self):
        return self._marks

student = Student(85)

print(student.marks)
````

Notice that we wrote:

```python
student.marks
```

and **not**:

```python
student.marks()
```

Although `marks` is defined using a method, `@property` makes it behave like an attribute when accessed.

Conceptually:

```text
student.marks
      ↓
property getter
      ↓
return self._marks
```

## 7. Getter with `@property`

A property getter commonly looks like this:

```python
@property
def marks(self):
    return self._marks
```

### Parts of the getter

- `@property` — converts the method into a property
- `marks` — public property name
- `self._marks` — internal storage attribute
- `return` — supplies the value to the caller

We use `_marks` as the internal attribute so that the property and its storage have different names.

````python
class Student:
    def __init__(self, marks):
        self._marks = marks

    @property
    def marks(self):
        return self._marks

student = Student(85)

print(student.marks)
print(student._marks)
````

Here:

```text
Public property
      ↓
    marks
      ↓
   getter
      ↓
  _marks
```

The public interface is `marks`, while `_marks` is the internal storage.

## 8. Setter with `@marks.setter`

A getter allows reading. To control assignment, we can add a setter.

The setter is connected to the property using the same property name:

```python
@marks.setter
def marks(self, value):
    ...
```

````python
class Student:
    def __init__(self, marks):
        self._marks = marks

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):
        if 0 <= value <= 100:
            self._marks = value
        else:
            print("Invalid marks")

student = Student(85)

print(student.marks)

student.marks = 95
print(student.marks)

student.marks = 150
print(student.marks)
````

The important syntax is:

```python
@marks.setter
```

It must use the **same name as the property**:

```python
@property
def marks(...):
    ...

@marks.setter
def marks(...):
    ...
```

When we write:

```python
student.marks = 95
```

Python invokes the setter.

## 9. Why Do We Use `_marks`?

The internal attribute is usually given a leading underscore:

```python
self._marks
```

This is important because the property itself is called `marks`.

A common mistake is:

```python
@marks.setter
def marks(self, value):
    self.marks = value
```

This makes the setter assign to the same property again. The setter calls itself repeatedly and leads to recursive behavior.

Instead, store the value in the internal attribute:

```python
self._marks = value
```

### Mental model

```text
Public property
      ↓
    marks
      ↓
 getter / setter
      ↓
Internal attribute
      ↓
   _marks
```

The leading underscore convention communicates that `_marks` is intended for internal use.

## 10. Read-Only Properties

A property can have only a getter.

For example, a circle's diameter can be calculated from its radius.

````python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def diameter(self):
        return 2 * self.radius

circle = Circle(5)

print(circle.diameter)
````

Because there is no setter for `diameter`, this is a read-only property.

Trying to assign a value:

```python
circle.diameter = 20
```

raises an error because the property does not define how assignment should be handled.

Read-only properties are useful for values that should be calculated or exposed but not directly changed by the caller.

## 11. Computed Properties

A property does not have to store a separate value.

It can calculate a value whenever it is accessed.

````python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    @property
    def area(self):
        return self.length * self.width

rectangle = Rectangle(10, 5)

print(rectangle.area)
````

`area` is a **computed property**.

We write:

```python
rectangle.area
```

not:

```python
rectangle.area()
```

because `area` is exposed as a property.

The calculation happens when the property is accessed.

## 12. Property with Validation

Properties are especially useful when an attribute needs validation.

Example: a bank account balance should not be negative.

````python
class BankAccount:
    def __init__(self, balance):
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, amount):
        if amount >= 0:
            self._balance = amount
        else:
            print("Balance cannot be negative")

account = BankAccount(5000)

print(account.balance)

account.balance = 7000
print(account.balance)

account.balance = -1000
print(account.balance)
````

The setter creates a controlled point for validation:

```text
account.balance = value
          ↓
       setter
          ↓
    validation
       ↙   ↘
 valid     invalid
   ↓          ↓
store      reject
```

## 13. Property Deleter

A property can also define what happens when the property is deleted.

The third part of the property system is:

```python
@marks.deleter
def marks(self):
    ...
```

````python
class Student:
    def __init__(self, marks):
        self._marks = marks

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):
        if 0 <= value <= 100:
            self._marks = value
        else:
            print("Invalid marks")

    @marks.deleter
    def marks(self):
        del self._marks

student = Student(85)

print(student.marks)

del student.marks

print(hasattr(student, "_marks"))
````

The three property components are:

```text
@property       → getter
@marks.setter   → setter
@marks.deleter  → deleter
```

The deleter controls what happens when we write:

```python
del student.marks
```

## 14. Complete Property Structure

Here is the complete pattern together:

```python
class Student:
    def __init__(self, marks):
        self._marks = marks

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):
        if 0 <= value <= 100:
            self._marks = value
        else:
            raise ValueError("Marks must be between 0 and 100")

    @marks.deleter
    def marks(self):
        del self._marks
```

This gives us a single public property with three controlled operations:

```text
student.marks
      ↓
    getter

student.marks = 90
      ↓
    setter

del student.marks
      ↓
    deleter
```

````python
class Student:
    def __init__(self, marks):
        self._marks = marks

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):
        if 0 <= value <= 100:
            self._marks = value
        else:
            raise ValueError("Marks must be between 0 and 100")

    @marks.deleter
    def marks(self):
        del self._marks

student = Student(85)

print(student.marks)

student.marks = 95
print(student.marks)

del student.marks
print(hasattr(student, "_marks"))
````

## 15. Property vs Normal Attribute

| Normal Attribute | Property |
|---|---|
| Direct access | Controlled access |
| No automatic validation | Can validate |
| Usually stores a value directly | Can calculate a value |
| Simple | More control |
| `obj.value` | `obj.value` |

An important point is that **the syntax for the user can remain the same**.

For example:

```python
student.marks
```

can look like ordinary attribute access while internally running a getter.

## 16. Getter/Setter Methods vs `@property`

### Traditional approach

```python
student.get_marks()
student.set_marks(90)
```

### Property approach

```python
student.marks
student.marks = 90
```

The property approach is useful when you want an attribute-like interface while still controlling access.

Python commonly uses properties for this purpose.

## 17. Practical Example — Employee Salary

Suppose an employee's salary should never be negative.

A property can provide validation while keeping normal attribute-style syntax.

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self._salary = salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value >= 0:
            self._salary = value
        else:
            raise ValueError("Salary cannot be negative")

employee = Employee("Rahul", 50000)

print(employee.salary)

employee.salary = 60000
print(employee.salary)
````

If an invalid salary is assigned:

```python
employee.salary = -5000
```

the setter rejects it by raising `ValueError`.

The caller still uses:

```python
employee.salary
```

and:

```python
employee.salary = value
```

rather than explicit getter/setter method calls.

## 18. Practical Example — Temperature

Properties can also expose related calculated values.

````python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        self._celsius = value

    @property
    def fahrenheit(self):
        return (self._celsius * 9 / 5) + 32

temp = Temperature(25)

print(temp.celsius)
print(temp.fahrenheit)

temp.celsius = 30

print(temp.celsius)
print(temp.fahrenheit)
````

Here:

- `celsius` is a stored property with a getter and setter.
- `fahrenheit` is a computed property.
- `fahrenheit` does not need separate storage because its value can be calculated from Celsius.

This is a common use of read-only computed properties.

## 19. Properties and Encapsulation

Properties connect naturally with the OOP concept of **encapsulation**.

A simplified mental model is:

```text
Object
│
├── Internal data

│    _marks
│
└── Controlled interface
       ↓
     marks
       ↓
   getter/setter
```

Properties provide a way to control how internal object data is accessed and modified.

## 20. Common Mistakes

### Mistake 1 — Forgetting `@property`

If we write:

```python
def marks(self):
    return self._marks
```

without `@property`, then `marks` remains a normal method.

Calling:

```python
student.marks
```

will not behave like a property.

---

### Mistake 2 — Recursive setter

Avoid:

```python
self.marks = value
```

inside the `marks` setter.

Use:

```python
self._marks = value
```

for internal storage.

---

### Mistake 3 — Wrong setter name

The setter must be attached to the same property:

```python
@property
def marks(self):
    ...

@marks.setter
def marks(self, value):
    ...
```

Using a different name such as `@score.setter` does not attach the setter to `marks`.

---

### Mistake 4 — Calling a property like a method

Incorrect:

```python
student.marks()
```

Correct:

```python
student.marks
```

A property is accessed like an attribute.

---

### Mistake 5 — Forgetting `_` for internal storage

Using a separate internal name such as:

```python
self._marks
```

helps prevent the property setter from assigning to the property itself.

The underscore also communicates that the attribute is intended as an internal implementation detail.
