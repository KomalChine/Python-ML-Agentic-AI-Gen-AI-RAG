# 1. Introduction to Polymorphism

**Polymorphism** means:

> One interface, many forms.

The word can be understood as:

```text
Poly       → many
Morphism   → forms
```

In Python, polymorphism allows the same operation, function, or method name to work with different types or objects.

````python
print(len("Python"))
print(len([10, 20, 30]))
print(len({"a": 1, "b": 2}))
````

The same `len()` function works with different objects.

The object determines what the operation means.

# 2. Real-World Analogy

Suppose there is one action:

```text
make_sound()
```

Different animals perform it differently:

```text
Dog  → Bark
Cat  → Meow
Cow  → Moo
```

The action is the same, but the implementation is different.

That is the basic idea behind polymorphism.

# 3. Polymorphism with Built-in Functions

Python's built-in functions often demonstrate polymorphic behavior.

## `len()`

````python
print(len("Hello"))
print(len([1, 2, 3]))
print(len((10, 20)))
````

## `sum()`

````python
print(sum([10, 20, 30]))
print(sum((1, 2, 3)))
````

## `max()`

````python
print(max([10, 50, 20]))
print(max("Python"))
````

The same built-in function can operate on different kinds of objects when those objects support the required operation.

# 5. Method Overriding as Polymorphism

Now connect polymorphism directly to inheritance.

A parent class can define a method, and child classes can override that method with their own behavior.

````python
class Animal:
    def sound(self):
        print("Some animal sound")


class Dog(Animal):
    def sound(self):
        print("Bark")


class Cat(Animal):
    def sound(self):
        print("Meow")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
````

The same method:

```text
sound()
```

has different behavior depending on the object.

This is one of the most important forms of polymorphism in Python.

# 6. Same Method, Different Objects

Polymorphism does not always require inheritance.

Here, each class provides the same method name.

````python
class Dog:
    def speak(self):
        print("Bark")


class Cat:
    def speak(self):
        print("Meow")


class Cow:
    def speak(self):
        print("Moo")


dog = Dog()
cat = Cat()
cow = Cow()

dog.speak()
cat.speak()
cow.speak()
````

The method name is the same:

```text
speak()
```

but each object provides different behavior.

# 7. Polymorphism Without Inheritance

In Python, classes do not necessarily need to inherit from a common parent to participate in polymorphic behavior.

````python
class Dog:
    def speak(self):
        print("Bark")


class Robot:
    def speak(self):
        print("Beep")


class Human:
    def speak(self):
        print("Hello")


def make_speak(obj):
    obj.speak()


make_speak(Dog())
make_speak(Robot())
make_speak(Human())
````

All three objects work with the same function because they provide the required `speak()` behavior.

Python often cares about what an object **can do**, rather than only what class it belongs to.

# 8. Duck Typing

A famous Python principle is:

> "If it walks like a duck and quacks like a duck, treat it like a duck."

In programming terms:

> An object can be used based on the behavior it provides rather than its exact class.

````python
class Dog:
    def speak(self):
        print("Bark")


class Person:
    def speak(self):
        print("Hello")


def make_speak(obj):
    obj.speak()


make_speak(Dog())
make_speak(Person())
````

No common parent is required.

The function simply expects the object to provide `speak()`.

# 9. Duck Typing vs Inheritance

### Inheritance

```python
class Animal:
    def speak(self):
        pass


class Dog(Animal):
    def speak(self):
        print("Bark")
```

Inheritance expresses a **relationship between classes**.

### Duck typing

```python
class Dog:
    def speak(self):
        print("Bark")


class Robot:
    def speak(self):
        print("Beep")
```

Duck typing focuses on the **required behavior**.

```text
Inheritance
→ relationship between classes

Duck typing
→ required behavior
```

# 10. Polymorphic Functions

A polymorphic function can accept different objects as long as they provide the operation the function needs.

````python
def start_engine(vehicle):
    vehicle.start()


class Car:
    def start(self):
        print("Car engine started")


class Bike:
    def start(self):
        print("Bike engine started")


start_engine(Car())
start_engine(Bike())
````

The same function:

```text
start_engine()
```

works with different object types.

# 11. Polymorphism with Inheritance

A common design is a parent class that defines a common interface and child classes that provide different implementations.

````python
class Vehicle:
    def start(self):
        print("Vehicle starting")


class Car(Vehicle):
    def start(self):
        print("Car engine starting")


class Bike(Vehicle):
    def start(self):
        print("Bike engine starting")


vehicles = [
    Car(),
    Bike()
]

for vehicle in vehicles:
    vehicle.start()
````

This gives us an important pattern:

```text
Common interface
      ↓
Different implementations
      ↓
One loop/function
```

# 12. Polymorphism with Lists

Polymorphism becomes especially useful when different objects are stored together in a collection.

````python
class Dog:
    def speak(self):
        print("Bark")


class Cat:
    def speak(self):
        print("Meow")


class Cow:
    def speak(self):
        print("Moo")


animals = [
    Dog(),
    Cat(),
    Cow()
]

for animal in animals:
    animal.speak()
````

The loop does not need to know whether the current object is a `Dog`, `Cat`, or `Cow`.

It only needs each object to provide `speak()`.

# 13. Polymorphism with Functions

````python
def describe_animal(animal):
    animal.speak()


describe_animal(Dog())
describe_animal(Cat())
describe_animal(Cow())
````

The function does not need separate checks such as:

```python
if isinstance(animal, Dog):
    ...
elif isinstance(animal, Cat):
    ...
```

A common interface keeps the function simple.

# 14. Avoid Unnecessary `isinstance()` Checks

A less flexible approach is to make the processing function responsible for every concrete class.

````python
def make_sound_with_checks(animal):
    if isinstance(animal, Dog):
        animal.speak()
    elif isinstance(animal, Cat):
        animal.speak()
    elif isinstance(animal, Cow):
        animal.speak()


make_sound_with_checks(Dog())
make_sound_with_checks(Cat())
````

A more polymorphic approach is:

```python
def make_sound(animal):
    animal.speak()
```

Each class owns its own behavior.

This makes the processing code easier to extend when new classes are introduced.

````python
def make_sound(animal):
    animal.speak()


make_sound(Dog())
make_sound(Cat())
make_sound(Cow())
````

# 15. Method Overriding vs Polymorphism

These concepts are related, but they are not identical.

### Method overriding

A child class replaces a parent method:

```python
class Animal:
    def speak(self):
        print("Animal")


class Dog(Animal):
    def speak(self):
        print("Bark")
```

### Polymorphism

The same interface can produce different behavior:

```python
dog.speak()
animal.speak()
```

So:

```text
Method overriding
→ technique

Polymorphism
→ behavior/design principle
```

Method overriding is commonly used to achieve runtime polymorphism.

# 16. Polymorphism with Abstract-Like Interfaces

Before learning formal abstraction in the next notebook, it is useful to see the idea of a common interface.

A parent method can indicate that subclasses are expected to provide their own implementation.

````python
class Shape:
    def area(self):
        raise NotImplementedError


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2


shapes = [
    Rectangle(10, 5),
    Circle(7)
]

for shape in shapes:
    print(shape.area())
````

The same interface:

```text
area()
```

has different implementations.

We are only introducing the idea here. Formal `abc.ABC` and `@abstractmethod` belong in **13_Encapsulation_and_Abstraction.ipynb**.

# 17. Real-World Example — Payment System

Different payment methods can expose the same `pay()` operation.

````python
class CreditCard:
    def pay(self, amount):
        print(f"Paid ₹{amount} using credit card")


class UPI:
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")


class Cash:
    def pay(self, amount):
        print(f"Paid ₹{amount} using cash")


def process_payment(payment_method, amount):
    payment_method.pay(amount)


process_payment(CreditCard(), 1000)
process_payment(UPI(), 500)
process_payment(Cash(), 200)
````

The processing function does not need to know which payment method it receives.

# 18. Real-World Example — Notification System

````python
class EmailNotification:
    def send(self, message):
        print("Sending email:", message)


class SMSNotification:
    def send(self, message):
        print("Sending SMS:", message)


class PushNotification:
    def send(self, message):
        print("Sending push notification:", message)


def notify(service, message):
    service.send(message)


notify(
    EmailNotification(),
    "Your order has been shipped."
)

notify(
    SMSNotification(),
    "Your OTP is 1234."
)

notify(
    PushNotification(),
    "You have a new message."
)
````

The same `notify()` function works with multiple notification implementations.

# 19. Real-World Example — Data Processing

Polymorphism is especially useful for systems that process different kinds of data.

This example is relevant to data-processing workflows.

````python
class CSVProcessor:
    def process(self):
        print("Processing CSV data")


class JSONProcessor:
    def process(self):
        print("Processing JSON data")


class ExcelProcessor:
    def process(self):
        print("Processing Excel data")


def run_processor(processor):
    processor.process()


processors = [
    CSVProcessor(),
    JSONProcessor(),
    ExcelProcessor()
]

for processor in processors:
    run_processor(processor)
````

The processing pipeline can work with different data sources through the same `process()` interface.

# 22. Polymorphism and `super()`

Polymorphism also connects with the `super()` concept from inheritance.

An overridden method can extend the parent implementation.

````python
class Animal:
    def speak(self):
        print("Animal speaks")


class Dog(Animal):
    def speak(self):
        super().speak()
        print("Dog barks")


dog = Dog()
dog.speak()
````

Output:

```text
Animal speaks
Dog barks
```

Here, `super()` lets the child method reuse parent behavior before adding its own behavior.

# 23. Polymorphism with Class Methods

This is an advanced connection to the class methods already studied.

Because `cls` refers to the class through which a class method is called, the same class method can create objects of different subclasses.

````python
class Animal:
    @classmethod
    def create(cls):
        return cls()


class Dog(Animal):
    pass


dog = Dog.create()

print(type(dog))
````

Because `cls` refers to the class through which `create()` was called, `Dog.create()` creates a `Dog` object.

This is useful, but the main focus of this notebook remains method/interface polymorphism.

# 24. Benefits of Polymorphism

### 1. Less conditional logic

Instead of checking every concrete type:

```python
if type == "car":
    ...
elif type == "bike":
    ...
```

we can use:

```python
vehicle.start()
```

### 2. Easier extension

A new class can provide the same interface without requiring changes to the processing function.

### 3. Code reuse

One function can work with many objects.

### 4. Maintainability

Each class controls its own behavior.

### 5. Flexibility

The caller does not need to know the object's internal implementation.

# 25. Common Mistakes

## Mistake 1 — Confusing inheritance with polymorphism

Inheritance:

```text
class Dog(Animal)
```

Polymorphism:

```text
dog.speak()
cat.speak()
```

with the same interface producing different behavior.

---

## Mistake 2 — Too many `isinstance()` checks

Avoid long chains of type checks when a common interface can solve the problem.

---

## Mistake 3 — Different method names

For a shared interface, use the same method name:

```text
Dog.speak()
Cat.speak()
Cow.speak()
```

If one class uses `bark()` and another uses `meow()`, they do not expose the same interface.

---

## Mistake 4 — Forgetting the required method

If a function expects:

```python
def make_sound(animal):
    animal.speak()
```

the object must provide `speak()`.

Otherwise the call fails.

---

## Mistake 5 — Overcomplicating simple programs

Do not introduce inheritance or polymorphism when a simple function or straightforward design is enough.

# 26. Inheritance + Polymorphism Together

This is one of the most important patterns to understand.

````python
class Animal:
    def speak(self):
        raise NotImplementedError


class Dog(Animal):
    def speak(self):
        print("Bark")


class Cat(Animal):
    def speak(self):
        print("Meow")


class Cow(Animal):
    def speak(self):
        print("Moo")


animals = [
    Dog(),
    Cat(),
    Cow()
]

for animal in animals:
    animal.speak()
````

The structure is:

```text
               Animal
              /  |  \
             /   |   \
          Dog   Cat   Cow
           ↓     ↓     ↓
         Bark   Meow   Moo
```

This combines:

```text
Inheritance
    +
Polymorphism
```

The parent establishes the common interface, while each child provides its own implementation.

# 27. Practical Project — Shape Calculator

Build a shape system in which different shapes provide the same `area()` method.

```text
Shape
 ├── Rectangle
 ├── Circle
 └── Triangle
```

The processing code should be able to work with all shape objects through the common interface.

````python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2


shapes = [
    Rectangle(10, 5),
    Circle(7)
]

for shape in shapes:
    print(shape.area())
````

The important part is not the individual formula.

The important part is that every shape exposes the same:

```python
area()
```

interface.

A `Triangle` can later be added using the same pattern without changing the loop.
