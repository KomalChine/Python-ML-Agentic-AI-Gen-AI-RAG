# 06 — Generator Expressions

Generator expressions provide a compact way to create generators using expression syntax. They connect the familiar list comprehension syntax from Basic Python with the generators learned in the previous notebooks.

In this notebook, we will learn:

- how generator expressions are written
- how they differ from list comprehensions
- lazy evaluation and on-demand values
- using `for`, `next()`, conditions, and functions
- using generator expressions with `any()`, `all()`, `sum()`, `min()`, and `max()`
- when a generator expression is preferable to a generator function

## 1. Introduction to Generator Expressions

From Basic Python, you already know list comprehensions:

```python
squares = [number ** 2 for number in range(1, 6)]
```

This creates a **list**.

Now change the square brackets to parentheses:

```python
squares = (number ** 2 for number in range(1, 6))
```

This creates a **generator expression**.

The important difference is:

```text
[ ... ]  → list comprehension
( ... )  → generator expression
```

A generator expression creates a generator object rather than a list.

````python
squares = [number ** 2 for number in range(1, 6)]
print(squares)

squares = (number ** 2 for number in range(1, 6))
print(squares)
````

## 2. Generator Expression Syntax

The basic syntax is:

```python
(expression for item in iterable)
```

Example:

```python
numbers = (number for number in range(5))
```

A condition can be added:

```python
numbers = (number for number in range(10) if number % 2 == 0)
```

The parts can be viewed as:

```text
(expression    for item    in iterable    if condition)
     ↓             ↓            ↓              ↓
   value        variable      source       optional
```

The `if condition` part is optional.

````python
numbers = (number for number in range(5))
print(numbers)

even_numbers = (
    number
    for number in range(10)
    if number % 2 == 0
)

print(even_numbers)
````

## 3. Generator Expression vs List Comprehension

List comprehensions and generator expressions can look almost identical, but they create different kinds of objects.

| List Comprehension | Generator Expression |
|---|---|
| Uses `[]` | Uses `()` |
| Creates a list | Creates a generator |
| Values are created eagerly | Values are produced lazily |
| Can be indexed | Cannot be directly indexed |
| Can be iterated over repeatedly | Becomes exhausted |
| Stores the generated values | Produces values on demand |

Compare the two:

````python
squares_list = [number ** 2 for number in range(5)]
squares_generator = (number ** 2 for number in range(5))

print(squares_list)
print(squares_generator)

print(squares_list[2])

# A generator does not support direct indexing.
# print(squares_generator[2])
````

A list is immediately populated with its values.

A generator expression instead creates a generator object. Its values are produced as the generator is consumed.

This is the first major idea to remember:

> A generator expression describes how to produce values; it does not create all of them immediately.

## 4. Lazy Evaluation

Lazy evaluation means that values are produced when they are requested rather than all at once.

Consider:

```python
numbers = (number ** 2 for number in range(1, 6))
```

The generator can produce values one at a time:

```python
print(next(numbers))
print(next(numbers))
print(next(numbers))
```

The first three values are:

```text
1
4
9
```

The process is:

```text
Generator Expression
        ↓
does not create everything immediately
        ↓
next() requests a value
        ↓
value is calculated
        ↓
value is produced
```

````python
numbers = (number ** 2 for number in range(1, 6))

print(next(numbers))
print(next(numbers))
print(next(numbers))
````

Lazy evaluation is especially useful when the source contains many values or when you only need to consume values as they are needed.

The important point here is behavioral: the generator produces values on demand.

## 5. Creating Generator Expressions

Generator expressions can be used with many familiar iterables.

### Numbers

````python
numbers = (number for number in range(5))

for number in numbers:
    print(number)
````

### Squares

````python
squares = (number ** 2 for number in range(1, 6))

for square in squares:
    print(square)
````

### Cubes

````python
cubes = (number ** 3 for number in range(1, 6))

for cube in cubes:
    print(cube)
````

### Characters

````python
characters = (character for character in "Python")

for character in characters:
    print(character)
````

## 6. Using Generator Expressions with `for` Loops

A `for` loop is one of the most common ways to consume a generator expression.

```python
squares = (number ** 2 for number in range(1, 6))

for square in squares:
    print(square)
```

````python
squares = (number ** 2 for number in range(1, 6))

for square in squares:
    print(square)
````

Generator expressions can also transform values from an existing iterable.

````python
names = ("Name: " + name for name in ["Alice", "Bob", "Charlie"])

for name in names:
    print(name)
````

The `for` loop requests values from the generator until the generator is exhausted.

## 7. Using `next()` with Generator Expressions

Because a generator expression creates a generator object, it can be consumed with `next()`.

````python
numbers = (number * 10 for number in range(1, 4))

print(next(numbers))
print(next(numbers))
print(next(numbers))
````

After all available values have been produced, another `next()` call raises `StopIteration`.

This connects directly to the earlier iterator notebooks: a generator is an iterator and follows the same `next()` / `StopIteration` behavior.

````python
numbers = (number * 10 for number in range(1, 4))

print(next(numbers))
print(next(numbers))
print(next(numbers))

# This raises StopIteration because the generator is exhausted.
# print(next(numbers))
````

## 8. Generator Expressions with Conditions

An `if` condition can be included to select values.

Example:

```python
even_numbers = (
    number
    for number in range(10)
    if number % 2 == 0
)
```

````python
even_numbers = (
    number
    for number in range(10)
    if number % 2 == 0
)

for number in even_numbers:
    print(number)
````

Another example filters out negative values:

````python
positive_numbers = (
    number
    for number in [-5, 3, -2, 8, -1, 10]
    if number > 0
)

for number in positive_numbers:
    print(number)
````

The condition does not turn the generator into a list. The selected values are still produced lazily as the generator is consumed.

## 9. Generator Expressions with Functions

A generator expression can call functions or methods while producing each value.

For example:

````python
names = ["alice", "bob", "charlie"]

uppercase_names = (name.upper() for name in names)

for name in uppercase_names:
    print(name)
````

Generator expressions can also be chained.

The first generator produces values that become the input to the second generator:

````python
numbers = (number for number in range(1, 6))
squares = (number ** 2 for number in numbers)

for square in squares:
    print(square)
````

The chain remains lazy: values move through the generators as they are requested.

## 10. Generator Expressions with Built-in Functions

Generator expressions are often passed directly to functions.

Instead of creating a list first:

```python
numbers = [number ** 2 for number in range(1, 6)]
result = sum(numbers)
```

you can write:

```python
result = sum(number ** 2 for number in range(1, 6))
```

When a generator expression is the sole argument to a function, its extra outer parentheses are unnecessary.

````python
numbers = [number ** 2 for number in range(1, 6)]
result = sum(numbers)

print(result)
````

````python
result = sum(number ** 2 for number in range(1, 6))

print(result)
````

This pattern is common because the function can consume the generated values directly without first creating a separate list.

## 11. Generator Expressions with `any()` and `all()`

Generator expressions work particularly well with `any()` and `all()`.

### `all()`

`all()` returns `True` when every consumed condition is true.

````python
numbers = [2, 4, 6, 8]

result = all(number % 2 == 0 for number in numbers)

print(result)
````

### `any()`

`any()` returns `True` when at least one consumed condition is true.

````python
numbers = [1, 3, 5, 8]

result = any(number % 2 == 0 for number in numbers)

print(result)
````

`any()` and `all()` can consume a generator expression lazily. They may stop before examining every value when the final result is already known.

This can make generator expressions a natural fit for checking conditions over an iterable.

## 12. Generator Expressions with `sum()`, `min()`, `max()`

Generator expressions are also useful when calculating results from generated values.

### Sum

````python
total = sum(number for number in range(1, 6))

print(total)
````

### Sum of squares

````python
total = sum(number ** 2 for number in range(1, 6))

print(total)
````

### Maximum

````python
maximum = max(number ** 2 for number in range(1, 6))

print(maximum)
````

### Minimum

````python
minimum = min(number ** 2 for number in range(1, 6))

print(minimum)
````

However, do not use a generator expression automatically when it adds no value.

For example:

```python
sum(number for number in range(1, 6))
```

can simply be:

```python
sum(range(1, 6))
```

Prefer the simpler expression when the built-in function can already consume the source directly.

````python
print(sum(number for number in range(1, 6)))
print(sum(range(1, 6)))
````

## 13. Nested Generator Expressions

Generator expressions can contain more than one `for` clause.

Keep the logic simple while learning this pattern.

Example:

````python
pairs = (
    (x, y)
    for x in range(3)
    for y in range(3)
)

for pair in pairs:
    print(pair)
````

A more practical example flattens a small matrix:

````python
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

values = (
    value
    for row in matrix
    for value in row
)

for value in values:
    print(value)
````

Nested generator expressions are useful, but readability matters. If the expression becomes difficult to understand, a normal loop or generator function may be clearer.

## 14. Generator Expression vs Generator Function

A generator function uses `def` and `yield`:

```python
def squares(n):
    for number in range(1, n + 1):
        yield number ** 2
```

A generator expression can express the same simple transformation compactly:

```python
squares = (
    number ** 2
    for number in range(1, 6)
)
```

Comparison:

| Generator Function | Generator Expression |
|---|---|
| Uses `def` | Uses expression syntax |
| Uses `yield` | Does not use `yield` |
| Can contain complex logic | Best for simple transformations/filtering |
| Can contain multiple statements | Uses one expression |
| More flexible | More concise |

An important rule of thumb:

> Use a generator expression for simple iteration logic. Use a generator function when the logic needs more structure or multiple statements.

````python
def squares(n):
    for number in range(1, n + 1):
        yield number ** 2


generator_from_function = squares(5)

for value in generator_from_function:
    print(value)
````

````python
generator_from_expression = (
    number ** 2
    for number in range(1, 6)
)

for value in generator_from_expression:
    print(value)
````

Both approaches produce values lazily. The main difference is how much logic each form can express clearly.

## 16. Common Mistakes

### Mistake 1 — Expecting a list

```python
numbers = (number for number in range(5))
print(numbers)
```

This prints a generator object, not the generated values.

To materialize all values into a list:

```python
print(list(numbers))
```

````python
numbers = (number for number in range(5))

print(numbers)
print(list(numbers))
````

### Mistake 2 — Trying to index a generator

A generator does not support direct indexing:

```python
numbers = (number for number in range(5))
print(numbers[0])
```

This raises `TypeError`.

If direct indexing is required, use a list instead.

````python
numbers = (number for number in range(5))

# This raises TypeError:
# print(numbers[0])

numbers = list(number for number in range(5))
print(numbers[0])
````

### Mistake 3 — Forgetting that generators are consumed

Once a generator has been consumed, its values are gone.

````python
numbers = (number for number in range(3))

print(list(numbers))
print(list(numbers))
````

The second result is:

```python
[]
```

because the generator has already been exhausted.

### Mistake 4 — Using a generator expression when a list is actually needed

If you need to do something such as:

```python
numbers[2]
```

or need to iterate over the same values repeatedly, a list may be more appropriate.

Choose the data structure based on what the program needs.

### Mistake 5 — Making the expression unnecessarily complicated

Generator expressions are best for compact, readable logic.

If the logic requires multiple statements or complicated control flow, a **generator function** is usually clearer.

For example, move to a generator function when you need several steps of processing rather than forcing everything into one expression.

## 17. Summary

### Key Takeaways

- Generator expressions use parentheses: `(expression for item in iterable)`.
- They create generator objects.
- Generator expressions are lazy.
- Values are produced only when requested.
- They can be consumed with `for` loops and `next()`.
- Conditions can be added with `if`.
- They work especially well with `sum()`, `min()`, `max()`, `any()`, and `all()`.
- Generator expressions are concise alternatives to simple generator functions.
- Generator expressions are not lists.
- Generators are consumed and become exhausted.
- Use generator expressions for simple logic.
- Use generator functions when the logic becomes more complex.
