# 01 — Iterables and Iterators

This notebook builds the foundation for understanding Python's iteration mechanism.

You already know how to use `for` loops. Here, we will look at the concepts behind them:

```text
Iterable → iter() → Iterator → next() → values → StopIteration
```

We will focus on the conceptual difference between **iterables** and **iterators**. Custom iterator protocols and generators come later.

## 1. Introduction to Iteration

You have already used `for` loops to process values one at a time:

````python
numbers = [10, 20, 30, 40]

for number in numbers:
    print(number)
````

Output:

```text
10
20
30
40
```

**Iteration** means going through the elements of a collection or sequence one at a time.

The important question for this Advanced Python section is:

> **What actually happens behind the `for` loop?**

That leads us to **iterables** and **iterators**.

## 2. What Is an Iterable?

An **iterable** is an object whose elements can be accessed one at a time during iteration.

Common examples include lists, tuples, strings, sets, dictionaries, and ranges.

For example:

````python
numbers = [10, 20, 30]
name = "Python"
values = (1, 2, 3)
unique = {10, 20, 30}
data = {"a": 1, "b": 2}

for item in numbers:
    print(item)
````

The same idea works with other iterables:

````python
for character in name:
    print(character)
````

````python
for value in values:
    print(value)
````

## 3. Common Python Iterables

### List

````python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
````

### Tuple

````python
numbers = (10, 20, 30)

for number in numbers:
    print(number)
````

### String

````python
word = "Python"

for character in word:
    print(character)
````

### Set

````python
numbers = {10, 20, 30}

for number in numbers:
    print(number)
````

### Dictionary

When you iterate over a dictionary normally, you get its **keys**.

````python
student = {
    "name": "Alice",
    "age": 25
}

for key in student:
    print(key)
````

### Range

`range()` is also an iterable:

````python
for number in range(5):
    print(number)
````

## 4. What Is an Iterator?

An **iterator** is an object that produces values one at a time and remembers where it is during the iteration.

For example, a list is iterable. We can create an iterator from that list using `iter()`:

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

Output:

```text
10
20
30
```

The iterator remembers its current position. After producing `10`, the next call produces `20`, rather than starting again from the beginning.

## 5. Iterable vs Iterator

This is the most important distinction in this notebook.

| Iterable | Iterator |
|---|---|
| Can be iterated over | Produces values one at a time |
| Can be used to create an iterator | Keeps track of iteration state |
| Works with `iter()` | Works with `next()` |
| Example: a list | Example: `iter(list)` |
| Does not represent a current iteration position in the same way | Maintains its current position |

A useful way to remember the relationship is:

```text
numbers
   ↓
iter()
   ↓
iterator
   ↓
next()
   ↓
10
   ↓
next()
   ↓
20
   ↓
next()
   ↓
30
```

> **An iterable is something you can iterate over; an iterator is the object that performs and remembers that iteration.**

## 6. The `iter()` Function

The `iter()` function creates an iterator from an iterable.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(iterator)
````

The important operation is:

```python
iter(numbers)
```

It creates an iterator that can produce the values from `numbers`.

You can then request those values with `next()`.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

## 7. The `next()` Function

`next()` asks an iterator for its **next value**.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

Output:

```text
10
20
30
```

The iterator moves forward after every `next()` call.

For example:

````python
numbers = [10, 20, 30, 40]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

Output:

```text
10
20
30
```

It does not restart from `10` each time. The iterator remembers its position.

## 8. Manually Iterating Over Data

Normally, you use a `for` loop:

````python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
````

We can manually request the same values using an iterator:

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

Output:

```text
10
20
30
```

This demonstrates an important part of the mechanism behind iteration.

We are intentionally **not** going into the full `__iter__()` / `__next__()` protocol yet. That belongs in `02_Iterator_Protocol.ipynb`.

## 9. Iterator State

An iterator has state: it remembers where it currently is in the sequence of values.

````python
numbers = [10, 20, 30, 40]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
````

Output:

```text
10
20
```

The iterator's progress can be pictured as:

```text
Initial
  ↓
10
  ↓
20
  ↓
30
  ↓
40
  ↓
Finished
```

Two iterators created from the same iterable have independent state:

````python
numbers = [10, 20, 30]

iterator1 = iter(numbers)
iterator2 = iter(numbers)

print(next(iterator1))
print(next(iterator1))

print(next(iterator2))
````

Output:

```text
10
20
10
```

`iterator1` has moved forward twice, while `iterator2` is still at its first value.

## 10. `StopIteration`

Once an iterator has no more values, calling `next()` again raises `StopIteration`.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))

# The next call would raise StopIteration:
# next(iterator)
````

Conceptually:

```text
next() → 10
next() → 20
next() → 30
next() → StopIteration
```

`StopIteration` tells Python that the iterator has no more values to produce.

Later, in `02_Iterator_Protocol.ipynb`, you will learn how this behavior fits into Python's iterator protocol.

## 11. How `for` Loops Use Iterators

Consider this familiar code:

````python
numbers = [10, 20, 30]

for number in numbers:
    print(number)
````

Conceptually, the iteration works like this:

```text
numbers
   ↓
iter(numbers)
   ↓
iterator
   ↓
next()
   ↓
10
   ↓
next()
   ↓
20
   ↓
next()
   ↓
30
   ↓
next()
   ↓
StopIteration
   ↓
loop ends
```

A simplified manual recreation is:

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

while True:
    try:
        number = next(iterator)
        print(number)
    except StopIteration:
        break
````

This is a **conceptual/manual recreation** of what the `for` loop accomplishes. The actual interpreter implementation has additional details.

The important learning sequence is:

```text
iterable
    ↓
iter()
    ↓
iterator
    ↓
next()
    ↓
values one at a time
    ↓
StopIteration
```

## 13. Common Mistakes

### Mistake 1: Calling `next()` on a List

A list is **iterable**, but it is not itself an iterator.

This is invalid:

````python
numbers = [10, 20, 30]

# next(numbers)  # TypeError: list is not an iterator
````

Correct approach:

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
````

### Mistake 2: Forgetting That an Iterator Gets Consumed

An iterator moves forward and can become exhausted.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(list(iterator))
print(list(iterator))
````

Output:

```text
[10, 20, 30]
[]
```

The first `list()` consumes the iterator. There are no values left for the second `list()`.

If you need to iterate over the original data again, create a new iterator:

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(list(iterator))

iterator = iter(numbers)

print(list(iterator))
````

### Mistake 3: Confusing an Iterable with an Iterator

Do not think:

```text
list = iterator
```

Instead, remember:

```text
list → iterable
iter(list) → iterator
```

The basic iteration mechanism can be pictured as:

```text
    ↓
Iterator

StopIteration
```

### Key Points

- **Iteration** means processing values one at a time.
- An **iterable** is something that can be iterated over.
- `iter()` creates an iterator from an iterable.
- An **iterator** produces values one at a time.
- `next()` retrieves the next value.
- An iterator remembers its current state.
- When no values remain, the iterator raises `StopIteration`.
- `for` loops rely on the iterator mechanism.
