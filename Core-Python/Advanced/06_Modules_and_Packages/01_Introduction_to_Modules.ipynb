A module is a Python source file (`.py`) containing code that can be reused, such as variables, functions, and classes.

## 1. What Is a Module?

A module is a Python file that groups related code in one place. Other Python files can import that module and use its definitions.

For example, imagine these two files in the same folder:

```python
# calculator.py
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
```

```python
# main.py
import calculator

print(calculator.add(10, 5))
print(calculator.subtract(10, 5))
```

Output:

```text
15
5
```

Here, `calculator.py` is the module. `main.py` imports it and calls its functions through the module name.

### Demonstration in this notebook

A notebook is not itself a `.py` module file, but we can demonstrate the idea by defining reusable functions in one place and calling them.

````python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

print(add(10, 5))
print(subtract(10, 5))
````

The example above demonstrates reusable functions. In a real multi-file project, those definitions could be saved in `calculator.py` and imported from `main.py`.

## 2. Why Do We Need Modules?

Modules help organize code as programs grow.

- **Reusability:** use the same functions or classes from multiple files.
- **Organization:** keep related functionality together.
- **Maintainability:** update shared logic in one place.
- **Readability:** make a project easier to understand.
- **Collaboration:** different people can work on separate files or areas.
- **Testing:** reusable units can be tested more independently.

For example, a data analysis project might separate file loading, cleaning, visualization, and reporting into different modules rather than placing every function in one large file.

## 3. A Module Can Contain More Than Functions

A module may contain variables, functions, classes, and executable top-level statements.

Consider a file named `student.py`:

````python
# Example contents of student.py

college = "ABC College"

def calculate_average(marks):
    return sum(marks) / len(marks)

class Student:
    def __init__(self, name):
        self.name = name
````

In this example:
- `college` is a module-level variable.
- `calculate_average()` is a function.
- `Student` is a class.

The code cell shows what the file could contain. To make it an importable module in a multi-file project, save the code in a file named `student.py`.

````python
college = "ABC College"

def calculate_average(marks):
    return sum(marks) / len(marks)

class Student:
    def __init__(self, name):
        self.name = name

print(college)
print(calculate_average([80, 90, 100]))
student = Student("Asha")
print(student.name)
````

Output:

```text
ABC College
90.0
Asha
```

## 4. Module Versus Script

A Python file can be used as an importable module, executed directly as a script, or serve both roles.

- **Module:** a file imported so another piece of code can use its definitions.
- **Script:** a file executed directly to perform a task.

The distinction depends on how the file is used, not only on its name. When a file is imported for the first time in a Python process, its top-level statements are executed to initialize the module. Python generally caches the loaded module for later imports in that process.

For example, `report.py` might define a reusable `create_report()` function. It could be imported by another file, or it could also be run directly to create a report. The `if __name__ == "__main__":` pattern is commonly used to separate direct-run behavior from import-time definitions; that pattern will be covered in Notebook 05.

## 5. Module Versus Package Versus Library

| Concept | Meaning | Example |
|---|---|---|
| **Module** | A Python source file | `calculator.py` |
| **Package** | A way to organize related modules under a package name | `math_tools/` |
| **Library** | A collection of reusable code, often containing modules and packages | Python standard library |

A package might have this structure:

```text
math_tools/
├── __init__.py
├── arithmetic.py
└── geometry.py
```

- `arithmetic.py` and `geometry.py` are modules.
- `math_tools` is the package.
- `__init__.py` is commonly used to mark and initialize a regular package.

Python also supports namespace packages that do not require `__init__.py`; package details will be covered later.

## 6. Advantages and Common Mistakes

### Advantages
- Reusable code
- Clear project structure
- Easier debugging and testing
- Better separation of responsibilities

### Common mistakes

1. **Naming a file after a standard-library module.** A local file named `random.py` or `math.py` can interfere with importing the intended module.
2. **Importing from the wrong directory.** Python must be able to find the module through its import search path.
3. **Forgetting top-level code runs on import.** Statements outside functions and classes can run when the module is imported.
4. **Assuming every file needs `main()`.** A `main()` function is a useful convention, not a requirement for every Python file.
5. **Putting unrelated responsibilities in one module.** Modules are most useful when their contents have a clear purpose.

## Key Takeaways

- A module is a `.py` file containing reusable Python code.
- Modules can contain variables, functions, classes, and top-level statements.
- Modules improve reuse, organization, readability, and maintainability.
- A file may be imported as a module, run as a script, or do both.
- A package organizes related modules; a library is a broader collection of reusable code.
- Avoid names that shadow standard-library modules and remember that top-level code can execute during import.
