## 1. Basic `import` Statement

The `import` statement makes a module available to your code. With a regular module import, use dot notation to access its members.

````python
import math

print(math.sqrt(25))
print(math.pi)
print(math.factorial(5))
````

Output:

```text
5.0
3.141592653589793
120
```

The prefix `math.` makes it clear where each name comes from and helps reduce ambiguity when different modules define names with the same spelling.

## 2. Import Specific Members

Use `from ... import ...` to bind selected names directly in the current namespace.

````python
from math import sqrt, pi

print(sqrt(36))
print(pi)
````

Output:

```text
6.0
3.141592653589793
```

Because `sqrt` and `pi` were imported directly, use `sqrt()` and `pi` rather than `math.sqrt()` and `math.pi`. This syntax can import functions, variables, and classes exposed by a module.

````python
from datetime import date

today = date.today()
print(today)
print(today.year)
````

This example imports the `date` class from the standard-library `datetime` module, then creates and uses a date object.

## 3. Import Aliases with `as`

An alias gives an imported module or member a different local name.

````python
import math as m

print(m.sqrt(49))
````

Output:

```text
7.0
```

Aliases are common in Data Science. For example, NumPy and pandas are conventionally imported as `np` and `pd`. Those packages must be installed in your Python environment before these imports will work.

````python
# Common conventions; run these if NumPy and pandas are installed.
import numpy as np
import pandas as pd

print(np.array([1, 2, 3]))
print(pd.Series([10, 20, 30]))
````

A specific member can also be aliased:

````python
from math import factorial as fact

print(fact(5))
````

## 4. Import Multiple Members

Import multiple names from one module by separating them with commas. For long lists, parentheses make the statement easier to read.

````python
from math import sqrt, factorial, pi, ceil, floor

print(sqrt(81))
print(ceil(4.2))
print(floor(4.8))
````

Output:

```text
9.0
5
4
```

````python
from math import (
    sqrt,
    factorial,
    pi,
    ceil,
    floor,
)

print(sqrt(16))
print(factorial(4))
print(pi)
````

You can also import names from different modules in separate statements. Keep imports explicit and grouped clearly so readers can see where each name originates.

````python
from math import sqrt
from statistics import mean

print(sqrt(64))
print(mean([10, 20, 30]))
````

## 5. Wildcard Imports

A wildcard import uses `*` to import names from a module:

````python
from math import *

print(sqrt(64))
print(factorial(5))
````

Although this works in many cases, wildcard imports are generally discouraged in application code:

- It is harder to tell which module provided a name.
- Names can overwrite other names in the current namespace.
- Code becomes less readable and can be harder to maintain.
- The names imported by `*` can depend on a module's `__all__` definition; without it, Python generally excludes names beginning with an underscore.

Prefer explicit imports:

````python
from math import sqrt, factorial

print(sqrt(64))
print(factorial(5))
````

## 7. Import Behavior and Common Mistakes

### Import initialization and caching

When Python imports a module for the first time in a process, it executes the module's top-level code to initialize it. Python stores the loaded module in `sys.modules`, so later imports in the same process generally reuse the loaded module instead of executing it from scratch again. Reloading is a separate, explicit operation and has its own caveats.

````python
import math
import sys

print(math.__name__)
print("math" in sys.modules)
````

### Common import errors

| Error | Typical cause |
|---|---|
| `ModuleNotFoundError` | Python cannot find the requested module or one of its dependencies. |
| `ImportError` | A module was found, but the requested name or import could not be resolved. |
| `AttributeError` | The imported module does not have the attribute you tried to access. |

Example: `from math import cube_root` raises `ImportError` because `math` does not provide a `cube_root` name.

````python
# Demonstrate handling an import error safely.
try:
    from math import cube_root
except ImportError as error:
    print(type(error).__name__)
    print("The requested name is not provided by math.")
````

### Name collisions and shadowing

Avoid naming your own file `math.py`, `random.py`, `statistics.py`, or `datetime.py`. A local file can shadow the intended standard-library module and lead to confusing errors. If an import behaves unexpectedly, check the current project directory and the imported module's `__file__` attribute when available.

````python
import math
print(getattr(math, "__file__", "This module has no __file__ attribute."))
````

### Circular imports: introductory awareness

A circular import occurs when module A imports module B while module B also imports module A, directly or through another module. Depending on which names are accessed and when, one module may be only partially initialized when the other tries to use it.

This notebook only introduces the issue. The later best-practices notebook will discuss cleaner ways to structure dependencies.

### Example A: A calculator module

Save this as `calculator.py`:

### Example B: Import a custom class

Save this as `student.py`:

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)
```

Then in `main.py`:

```python
from student import Student

student = Student("Asha", [80, 90, 100])
print(student.name)
print(student.average())
```

Expected output:

```text
Asha
90.0
```

### Example C: Diagnose an import problem

If Python reports `ModuleNotFoundError` for a custom module:
1. Confirm the filename and spelling.
2. Confirm the file is in a directory Python can search.
3. Check that you are running the intended script or notebook kernel.
4. Check whether another file with the same name is shadowing the intended module.
5. If the module is third-party, confirm it is installed in the active Python environment.

If Python reports `ImportError`, verify that the module actually defines the requested name and that the import statement matches its spelling.

## Key Takeaways

- `import module` keeps names under a module prefix; `from module import name` binds selected names directly.
- `as` creates a local alias for a module or imported member.
- Parentheses can make long multi-name imports readable.
- Explicit imports are usually clearer and safer than wildcard imports.
- Custom modules can expose variables, functions, and classes.
- Python initializes a module on import and usually caches it in `sys.modules`.
- Check spelling, search paths, file names, and requested members when debugging import errors.
