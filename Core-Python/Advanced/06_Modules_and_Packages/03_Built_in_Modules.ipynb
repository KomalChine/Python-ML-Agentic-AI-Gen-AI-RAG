## 1. The `math` Module

The `math` module provides mathematical constants and functions. Import it with `import math` and access members using dot notation.

````python
import math

print(math.sqrt(64))
print(math.factorial(5))
print(math.ceil(4.2))
print(math.floor(4.8))
print(math.pi)
````

Output:

```text
8.0
120
5
4
3.141592653589793
```

- `sqrt(x)`: square root of `x`.
- `factorial(n)`: factorial of a non-negative integer.
- `ceil(x)`: smallest integer greater than or equal to `x`.
- `floor(x)`: largest integer less than or equal to `x`.

````python
import math

print(math.ceil(-4.2))   # -4
print(math.floor(-4.2))  # -5
print(math.e)
print(math.inf)
````

`ceil()` and `floor()` behave differently for negative values because they move toward positive and negative infinity, respectively. `math.inf` represents positive infinity.

### Powers, logarithms, trigonometry, and rounding

````python
import math

print(math.pow(2, 3))        # 8.0
print(math.log(math.e))      # 1.0 (natural logarithm)
print(math.log10(1000))      # 3.0
print(math.sin(math.pi / 2)) # 1.0
print(math.cos(0))           # 1.0
print(round(4.567, 2))       # 4.57
print(math.floor(4.567))     # 4
````

`math.pow()` returns a floating-point result. The built-in `round()` rounds a number to a chosen number of decimal places, while `ceil()` and `floor()` return integer bounds. Python's `round()` uses ties-to-even behavior for exact halfway cases represented by the input, so it is not always equivalent to “round half up.”

## 2. The `random` Module

Use `random` for pseudorandom numbers, selections, games, and simulations. Its output may vary between runs.

````python
import random

print(random.random())          # float in [0.0, 1.0)
print(random.randint(1, 10))    # integer from 1 through 10
print(random.randrange(0, 10, 2))  # one of 0, 2, 4, 6, 8
print(random.choice(["Python", "SQL", "Excel"]))
````

Other useful functions:
- `choices(sequence, k=n)`: sample with replacement; items can repeat.
- `sample(sequence, k=n)`: sample without replacement; selected positions are unique.
- `shuffle(list)`: shuffles a mutable sequence in place and returns `None`.

````python
import random

topics = ["Python", "SQL", "Excel", "Statistics"]
print(random.choices(topics, k=3))
print(random.sample(topics, k=2))

numbers = [1, 2, 3, 4, 5]
result = random.shuffle(numbers)
print(numbers)
print(result)  # None
````

### Reproducible results with `seed()`

Setting a seed makes a pseudorandom sequence reproducible in the same relevant environment and with the same sequence of random calls. This is useful for experiments and demonstrations.

````python
import random

random.seed(42)
print(random.randint(1, 100))
````

Do not use `random` for passwords, security tokens, or security-sensitive randomness. Python's `secrets` module is designed for those purposes.

## 3. The `datetime` Module

The `datetime` module provides types for dates, times, combined date-times, and durations.
- `date`: calendar date.
- `time`: time of day.
- `datetime`: date and time together.
- `timedelta`: duration or difference between dates/times.

````python
from datetime import date, time, datetime, timedelta

today = date.today()
now = datetime.now()
time_example = time(14, 30)

print("Today:", today)
print("Current date and time:", now)
print("Example time:", time_example)

future_date = today + timedelta(days=7)
print("After seven days:", future_date)
````

### Formatting with `strftime()`

`strftime()` converts a date or datetime object to a formatted string.

````python
from datetime import datetime

now = datetime.now()
formatted = now.strftime("%d-%m-%Y")
print(formatted)
````

For example, `"%d-%m-%Y"` displays a date like `15-10-2026`. Common format directives include `%d` (day), `%m` (month), `%Y` (four-digit year), `%H` (hour in 24-hour format), `%M` (minute), and `%S` (second).

### Parsing strings with `strptime()`

`strptime()` parses a string according to a specified format and returns a datetime object.

````python
from datetime import datetime

date_text = "15-10-2026"
parsed_date = datetime.strptime(date_text, "%d-%m-%Y")
print(parsed_date)
print(parsed_date.date())
````

### Date differences and adding days

````python
from datetime import date, timedelta

start = date(2026, 10, 1)
end = date(2026, 10, 15)

difference = end - start
print("Days between:", difference.days)

next_week = start + timedelta(days=7)
print("Seven days after start:", next_week)
````

### Basic time-zone awareness

A naive datetime has no time-zone information. An aware datetime has a time zone or UTC offset. For a simple UTC-aware timestamp, use `timezone.utc`.

````python
from datetime import datetime, timezone

utc_now = datetime.now(timezone.utc)
print(utc_now)
print(utc_now.tzinfo)
````

For applications spanning locations or daylight-saving changes, use a suitable time zone such as one from `zoneinfo` rather than assuming every location uses the same UTC offset.

## 4. The `os` Module

The `os` module provides utilities for interacting with the operating system. Directory paths and environment variables depend on the machine and current process, so outputs can vary.

````python
import os

print("Current directory:", os.getcwd())
print("PATH environment variable exists:", os.environ.get("PATH") is not None)
print("Directory contents:", os.listdir("."))
````

`os.environ` provides access to environment variables. Prefer checking whether a variable exists rather than printing potentially sensitive values such as tokens or credentials.

### Create directories and check paths

````python
import os

os.makedirs("practice_data", exist_ok=True)
print(os.path.exists("practice_data"))
print(os.path.isdir("practice_data"))
````

`exist_ok=True` prevents an error if the target directory already exists. `os.path` also provides functions such as `isfile()`, `join()`, and `abspath()`.

````python
import os

print("Absolute current directory:", os.path.abspath("."))
print("Operating system name:", os.name)
````

## 5. The `sys` Module

The `sys` module provides information and utilities for the current Python interpreter.

````python
import sys

print("Python version:", sys.version)
print("Platform:", sys.platform)
print("Number of search-path entries:", len(sys.path))
````

- `sys.version`: Python version information.
- `sys.platform`: platform identifier.
- `sys.path`: directories Python searches for imports.
- `sys.argv`: command-line arguments passed to a script.
- `sys.stdin`, `sys.stdout`, and `sys.stderr`: standard input, output, and error streams.
- `sys.exit()`: requests program termination by raising `SystemExit`.

### Understanding `sys.argv`

For a script executed like `python greet.py Komal`, `sys.argv` would usually contain the script name followed by `"Komal"`. The notebook's own arguments are environment-specific, so use this illustrative script:

```python
# greet.py
import sys

if len(sys.argv) > 1:
    print("Hello,", sys.argv[1])
else:
    print("Hello, learner!")
```

Use `sys.exit()` carefully; it is generally more useful in scripts than in interactive notebook cells, where exiting may interrupt the notebook session.

## 6. The `pathlib` Module

`pathlib.Path` represents filesystem paths as objects. It is often easier to read than manually combining path strings.

````python
from pathlib import Path

folder = Path("practice_data")
folder.mkdir(exist_ok=True)

file_path = folder / "notes.txt"
file_path.write_text("Learning Python modules!", encoding="utf-8")

print(file_path.exists())
print(file_path.is_file())
print(file_path.read_text(encoding="utf-8"))
````

Output:

```text
True
True
Learning Python modules!
```

This example creates a directory and writes a file in the current working directory. Run it in a practice folder and be aware of where the notebook is running.

### Paths, joining, and checks

````python
from pathlib import Path

relative_path = Path("data") / "students.csv"
print("Path:", relative_path)
print("Absolute path:", relative_path.resolve())
print("Exists:", relative_path.exists())
print("Is file:", relative_path.is_file())
print("Is directory:", relative_path.is_dir())
````

A relative path is interpreted from the process's current working directory. An absolute path identifies a location from the filesystem root. `Path.resolve()` produces an absolute resolved path; whether the target exists depends on the filesystem.

````python
from pathlib import Path

folder = Path("practice_data") / "reports"
folder.mkdir(parents=True, exist_ok=True)

report_file = folder / "summary.txt"
report_file.write_text("Student statistics report", encoding="utf-8")
print(report_file.read_text(encoding="utf-8"))
````

Use `encoding="utf-8"` explicitly for predictable text-file encoding. `pathlib` is usually a good default for new Python code that manipulates paths.

## 7. Other Useful Standard-Library Modules

### `statistics`: descriptive statistics

````python
import statistics

marks = [72, 85, 85, 91, 67]

print("Mean:", statistics.mean(marks))
print("Median:", statistics.median(marks))
print("Mode:", statistics.mode(marks))
````

`statistics.mode()` returns a mode according to Python's documented behavior; for multiple tied modes, use `statistics.multimode()` if you want all values tied for most frequent.

````python
import statistics

marks = [70, 80, 80, 90, 90]
print(statistics.multimode(marks))
````

### `collections`: specialized containers

````python
from collections import Counter, defaultdict, deque

fruits = ["apple", "banana", "apple", "orange", "banana", "apple"]
print(Counter(fruits))

scores = defaultdict(list)
scores["Asha"].append(90)
scores["Asha"].append(95)
print(dict(scores))

queue = deque(["first", "second"])
queue.append("third")
print(queue.popleft())
print(queue)
````

- `Counter` counts occurrences.
- `defaultdict` creates a default value for a missing key.
- `deque` supports efficient appends and pops at both ends.

### `itertools`: iteration tools

````python
from itertools import combinations, count, islice

print(list(combinations(["A", "B", "C"], 2)))
print(list(islice(count(10, 2), 5)))
````

`combinations()` creates combinations of a chosen length. `count()` is an infinite iterator, so the example uses `islice()` to take only five values.

### `functools`: function utilities

````python
from functools import reduce, partial

numbers = [1, 2, 3, 4]
total = reduce(lambda a, b: a + b, numbers)
print(total)

def power(base, exponent):
    return base ** exponent

square = partial(power, exponent=2)
print(square(5))
````

`reduce()` combines an iterable into a single result; for simple sums, the built-in `sum()` is usually clearer. `partial()` creates a callable with some arguments already supplied.

### Finding reliable documentation

When you need a module's full behavior, examples, supported arguments, or edge cases, consult the official Python standard-library documentation. It is especially useful when platform behavior or version differences matter.

### B. Age calculation from a birth date

This example calculates age in years using the current date. Replace the sample birth date with a real date if needed.

````python
from datetime import date

birth_date = date(2003, 12, 9)
today = date.today()

age = today.year - birth_date.year
if (today.month, today.day) < (birth_date.month, birth_date.day):
    age -= 1

print("Age in years:", age)
````

### D. Command-line calculator concept

Save the following as `calculator_cli.py` and run it from a terminal. `sys.argv` holds command-line arguments.

```python
# Usage: python calculator_cli.py add 10 5
import sys

if len(sys.argv) != 4:
    print("Usage: python calculator_cli.py <add|subtract|multiply> <a> <b>")
    raise SystemExit(1)

operation = sys.argv[1]
a = float(sys.argv[2])
b = float(sys.argv[3])

if operation == "add":
    result = a + b
elif operation == "subtract":
    result = a - b
elif operation == "multiply":
    result = a * b
else:
    print("Unknown operation:", operation)
    raise SystemExit(1)

print(result)
```

### E. Student marks statistics

````python
import statistics
from collections import Counter

students = {
    "Asha": [80, 90, 85],
    "Ravi": [70, 75, 80],
    "Mina": [90, 95, 95],
}

all_marks = [mark for marks in students.values() for mark in marks]
print("Overall mean:", statistics.mean(all_marks))
print("Overall median:", statistics.median(all_marks))
print("Most common mark(s):", Counter(all_marks).most_common(3))
````
