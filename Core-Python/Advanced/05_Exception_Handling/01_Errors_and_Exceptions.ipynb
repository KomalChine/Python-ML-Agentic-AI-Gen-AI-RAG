# 01 — Errors and Exceptions

## Advanced Python — Exception Handling

This notebook introduces what can go wrong in Python programs, the major types of errors, exceptions, common built-in exceptions, tracebacks, and why exception handling is needed.

> This notebook introduces exceptions but does not teach `try`, `except`, `else`, or `finally` in detail. Those belong to the following notebooks.

## 1. Introduction

Even when Python code has correct syntax, something can still go wrong while the program is running.

Example:

```python
a = 10
b = 0

print(a / b)
```

The code is syntactically valid, but division by zero is not allowed, so Python raises an exception.

````python
a = 10
b = 0

print(a / b)
````

The basic flow is:

```text
Correct syntax
      ↓
Program starts running
      ↓
Problem occurs
      ↓
Exception
```

## 2. What Is an Error?

An error is a problem that prevents a program from behaving or executing as intended.

Three broad categories are useful here:

```text
Errors
├── Syntax Errors
├── Runtime Errors / Exceptions
└── Logical Errors
```

They are different kinds of problems.

## 3. Syntax Errors

A syntax error occurs when Python cannot understand the program because the code violates Python's syntax rules.

Example:

```python
if 10 > 5
    print("Hello")
```

Python reports a `SyntaxError`.

Another example:

```python
print("Hello"
```

Syntax errors are detected while Python is parsing the code. The affected code cannot execute normally until the syntax is fixed.

## 4. Runtime Errors

Some problems occur while the program is running.

Example:

```python
a = 10
b = 0

print(a / b)
```

This results in `ZeroDivisionError`.

Another example:

```python
numbers = [10, 20, 30]
print(numbers[5])
```

This results in `IndexError`.

Many runtime errors in Python are represented as exceptions.

## 5. Logical Errors

A logical error occurs when the program runs successfully but produces an incorrect result.

Example:

```python
length = 10
width = 5

area = length + width
```

The code runs, but the area formula is incorrect.

The correct calculation is:

```python
area = length * width
```

Important distinction:

```text
Syntax Error
→ Code cannot be parsed normally

Exception
→ Problem occurs during execution

Logical Error
→ Program runs but produces an incorrect result
```

## 6. Syntax Errors vs Runtime Errors vs Logical Errors

| Type | When detected | Program runs? | Example |
|---|---|---|---|
| Syntax Error | During parsing | No | Missing `:` |
| Runtime Error / Exception | During execution | Stops at the exception if unhandled | Division by zero |
| Logical Error | During/after execution | Yes | Wrong formula |

This distinction is important when identifying the cause of a problem.

## 7. What Is an Exception?

An **exception** is an event that occurs during program execution that disrupts the normal flow of the program.

Example:

```python
number = int("abc")
```

Python raises a `ValueError`.

Conceptually:

```text
Program execution
       ↓
Problematic situation
       ↓
Exception occurs
       ↓
Python raises the exception
       ↓
Program may terminate if it is not handled
```

Exception handling itself will be studied in the next notebooks.

## 8. Exception vs Error

For this course, use this practical distinction:

- Syntax errors are detected when Python parses invalid syntax.
- Exceptions are raised during execution.
- Logical errors are mistakes in program logic that Python may not detect.

Python's terminology commonly uses **exception** for runtime conditions that can be raised and handled.

## 9. Common Built-in Exceptions

### `ZeroDivisionError`

```python
print(10 / 0)
```

Occurs when an invalid division by zero is attempted.

### `ValueError`

```python
age = int("twenty")
```

Occurs when a value is inappropriate for an operation.

### `TypeError`

```python
result = 10 + "20"
```

Occurs when an operation is applied to incompatible types.

### `NameError`

```python
print(username)
```

Occurs when a name has not been defined.

### `IndexError`

```python
numbers = [10, 20, 30]
print(numbers[5])
```

Occurs when a sequence index is outside the valid range.

### `KeyError`

```python
student = {"name": "Komal"}
print(student["age"])
```

Occurs when a requested dictionary key does not exist.

### `AttributeError`

```python
name = "Python"
name.append("!")
```

Occurs when an object does not have the requested attribute or method.

### `FileNotFoundError`

```python
file = open("missing_file.txt")
```

Occurs when a requested file cannot be found.

### `ModuleNotFoundError`

```python
import something_that_does_not_exist
```

Occurs when Python cannot find the requested module.

### `ImportError`

`ImportError` is an import-related exception. The detailed mechanics of imports belong to the Modules and Packages section.

## 10. Reading an Exception Message

Consider:

```python
numbers = [10, 20, 30]
print(numbers[5])
```

A typical traceback contains information similar to:

```text
Traceback (most recent call last):
    ...
IndexError: list index out of range
```

Important parts:

```text
Traceback
   ↓
Where the problem occurred
   ↓
Exception type
   ↓
Description of the problem
```

## 11. Understanding a Traceback

A traceback helps identify:

- where the exception occurred,
- which line caused the problem,
- the type of exception,
- the message describing the problem.

Example:

```python
def divide(a, b):
    return a / b

x = divide(10, 0)
```

The traceback points toward the line where the division occurs.

Tracebacks are therefore an important debugging tool.

````python
def divide(a, b):
    return a / b

# This line produces a ZeroDivisionError.
x = divide(10, 0)
````

When reading a traceback, pay attention to:

1. The location of the problem.
2. The line involved.
3. The exception type.
4. The exception message.

## 12. `raise` — Basic Introduction

The `raise` statement allows a program to explicitly raise an exception.

Example:

```python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative")
```

The basic idea is:

> `raise` allows a program to explicitly raise an exception.

Detailed use of `raise` belongs to `05_Raising_Exceptions.ipynb`.

## 13. What Happens When an Exception Is Not Handled?

Consider:

```python
print("Start")

result = 10 / 0

print("End")
```

The output begins with `Start`, then the exception occurs.

`End` is never printed because execution is interrupted at the exception.

```text
Start
  ↓
Exception occurs
  ↓
Current execution is interrupted
  ↓
Program terminates if the exception is unhandled
```

## 14. Why Do We Need Exception Handling?

Consider:

```python
age = int(input("Enter your age: "))
```

If the user enters:

```text
abc
```

Python raises a `ValueError`.

Without exception handling, unexpected input can terminate the program.

Exception handling allows us to design programs that respond appropriately to such situations.

## 16. Exception Hierarchy — Basic Introduction

Python exceptions are organized into a class hierarchy.

A simplified view is:

```text
BaseException
│
├── Exception
│   │
│   ├── ArithmeticError
│   │   ├── ZeroDivisionError
│   │   └── ...
│   │
│   ├── LookupError
│   │   ├── IndexError
│   │   └── KeyError
│   │
│   ├── ValueError
│   ├── TypeError
│   └── ...
│
└── SystemExit
```

The goal here is only to understand that exceptions are organized into classes and inheritance relationships.

Detailed hierarchy is not necessary yet.

## 17. Exceptions Are Objects

Because you have recently studied OOP, this is a useful connection.

Exceptions are represented by Python classes and objects.

For example:

```python
ValueError
```

is an exception class.

An exception object can carry information such as its message.

Example:

```python
ValueError("Invalid age")
```

````python
exception_object = ValueError("Invalid age")

print(type(exception_object))
print(exception_object)
````

This example creates a `ValueError` object without raising it.

The important OOP connection is:

```text
Exception class
      ↓
Exception object
      ↓
Exception information/message
```

## 18. Common Mistakes

### Mistake 1 — Thinking every error is a syntax error

Runtime exceptions are different from syntax errors.

### Mistake 2 — Confusing logical errors with exceptions

A wrong formula may execute without raising an exception.

### Mistake 3 — Ignoring the traceback

The traceback provides useful debugging information.

### Mistake 4 — Assuming exceptions only happen because of bad code

Exceptions can also result from:

- invalid user input,
- missing files,
- unavailable resources.

### Mistake 5 — Trying to handle every possible exception blindly

Exception-handling best practices will be covered later.
