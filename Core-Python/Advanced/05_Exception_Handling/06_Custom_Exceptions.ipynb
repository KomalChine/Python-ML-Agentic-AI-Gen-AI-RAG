# 06 — Custom Exceptions

## Exception Handling

The previous notebook taught `raise` with built-in exceptions. This notebook introduces **custom exception classes for application-specific situations**.

Since classes, inheritance, `__init__`, and methods were already covered in OOP, this notebook connects those concepts to exceptions rather than reteaching them.

> **Boundary:** Custom exception classes are the focus here. Broader exception-handling best practices come later.

## 1. Introduction

A general exception such as `ValueError` may not clearly communicate a domain-specific problem.

For example, a banking application can use:

```python
raise InsufficientBalanceError("Insufficient balance")
```

instead of a generic `ValueError` when that distinction matters.

## 2. What Is a Custom Exception?

````python
class InsufficientBalanceError(Exception):
    pass

try:
    raise InsufficientBalanceError("Not enough balance")
except InsufficientBalanceError as e:
    print(type(e).__name__)
    print(e)
````

```text
Exception
    ↑
InsufficientBalanceError
```

## 3. Why Do We Need Custom Exceptions?

Examples include:

```text
Banking → InsufficientBalanceError
Student system → InvalidMarksError
Login system → AuthenticationError
Shopping system → OutOfStockError
Employee system → InvalidSalaryError
```

Custom exceptions make application-specific conditions clearer and easier to handle.

## 4. Creating a Basic Custom Exception

````python
class InvalidAgeError(Exception):
    pass

age = -5

try:
    if age < 0:
        raise InvalidAgeError("Age cannot be negative")
except InvalidAgeError as e:
    print("Error:", e)
````

## 5. Why Inherit from `Exception`?

The normal application pattern is:

```python
class MyError(Exception):
    pass
```

Conceptually:

```text
BaseException
    │
    └── Exception
          │
          └── MyError
```

For normal application-defined exceptions, inherit from `Exception`, not directly from `BaseException`.

## 6. Raising a Custom Exception

````python
class InvalidMarksError(Exception):
    pass

marks = 150

try:
    if not 0 <= marks <= 100:
        raise InvalidMarksError("Marks must be between 0 and 100")
except InvalidMarksError as e:
    print("Error:", e)
````

## 7. Handling a Custom Exception

Custom exceptions are handled with `except` just like built-in exceptions.

## 8. Custom Exception with a Function

A function can detect an invalid condition and raise a custom exception while the caller decides how to respond.

## 9. Custom Exception with `__init__`

Because exceptions are classes, they can store their own attributes.

## 10. Accessing Custom Exception Attributes

````python
class InvalidAgeError(Exception):
    def __init__(self, age):
        self.age = age
        super().__init__(f"Invalid age: {age}")

try:
    raise InvalidAgeError(-5)
except InvalidAgeError as e:
    print("Invalid age:", e.age)
    print("Message:", e)
````

## 11. Custom Exception with Multiple Attributes

````python
class InvalidMarksError(Exception):
    def __init__(self, marks, message="Invalid marks"):
        self.marks = marks
        super().__init__(message)

try:
    raise InvalidMarksError(150, "Marks must be between 0 and 100")
except InvalidMarksError as e:
    print("Marks:", e.marks)
    print("Error:", e)
````

## 12. Custom Exception Hierarchy

````python
class BankingError(Exception):
    pass

class InsufficientBalanceError(BankingError):
    pass

class InvalidAmountError(BankingError):
    pass

print(issubclass(InsufficientBalanceError, BankingError))
print(issubclass(BankingError, Exception))
````

## 13. Handling the Custom Hierarchy

Specific custom exceptions should come before their broader parent handler.

## 14. General Custom Exception Handler

A parent custom exception can handle several related child exceptions together.

## 15. Practical Example — Banking System

````python
class InsufficientBalanceError(Exception):
    pass

class InvalidAmountError(Exception):
    pass

def withdraw(balance, amount):
    if amount <= 0:
        raise InvalidAmountError("Withdrawal amount must be positive")
    if amount > balance:
        raise InsufficientBalanceError("Insufficient balance")
    return balance - amount

try:
    print(withdraw(5000, 6000))
except InvalidAmountError as e:
    print("Invalid amount:", e)
except InsufficientBalanceError as e:
    print("Transaction failed:", e)
````

## 16. Practical Example — Student System

````python
class InvalidMarksError(Exception):
    pass

def validate_marks(marks):
    if not 0 <= marks <= 100:
        raise InvalidMarksError("Marks must be between 0 and 100")
    return True

try:
    validate_marks(120)
except InvalidMarksError as e:
    print("Invalid marks:", e)
````

## 17. Practical Example — Login System

````python
class AuthenticationError(Exception):
    pass

def login(username, password):
    if username != "admin":
        raise AuthenticationError("Invalid username")
    if password != "1234":
        raise AuthenticationError("Invalid password")
    return "Login successful"

try:
    print(login("admin", "wrong"))
except AuthenticationError as e:
    print("Login failed:", e)
````

## 18. Custom Exceptions with Properties and OOP

````python
class InvalidMarksError(Exception):
    pass

class Student:
    def __init__(self, marks):
        self.marks = marks

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):
        if not 0 <= value <= 100:
            raise InvalidMarksError("Marks must be between 0 and 100")
        self._marks = value

try:
    student = Student(150)
except InvalidMarksError as e:
    print("Error:", e)
````

## 19. Custom Exceptions vs Built-in Exceptions

Use built-in exceptions when they accurately describe the problem. Use custom exceptions when the condition is meaningful to the application's domain.

Built-in examples: `ValueError`, `TypeError`, `IndexError`, `KeyError`, `FileNotFoundError`.

Custom examples: `InsufficientBalanceError`, `InvalidMarksError`, `AuthenticationError`, `OutOfStockError`.

## 20. Don't Create Custom Exceptions for Everything

Do not create a custom class merely to rename an existing built-in exception. If `ValueError` already communicates the problem correctly, use it.

## 21. Custom Exception Naming Convention

A common convention is `PascalCase + Error`, such as `InvalidMarksError` and `InsufficientBalanceError`.

## 22. Custom Exceptions in Separate Modules

In a larger project, custom exceptions can be placed in a file such as `exceptions.py`:

```text
project/
├── main.py
├── banking.py
└── exceptions.py
```

This is only an organizational example; modules and packages are covered separately.

## 23. Re-Raising Custom Exceptions

A lower-level exception can be translated into a domain-specific exception with:

```python
raise NewException(...) from e
```

This preserves the original exception as the cause.

## 24. Exception Chaining — Basic Idea

Conceptually:

```text
Original problem
ValueError
     ↓
   caused
     ↓
DataProcessingError
```

This preserves useful debugging information without going deeply into advanced exception chaining.

## 25. Common Mistakes

- Forgetting to inherit from `Exception`.
- Using `BaseException` directly for normal application exceptions.
- Creating too many custom exceptions when built-ins are sufficient.
- Using vague names such as `Error1`.
- Losing the original cause when translating exceptions.
