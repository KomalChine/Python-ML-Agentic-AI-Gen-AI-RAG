## 1. Introduction

A Python program can encounter exceptions while it is running. If an exception is not handled, normal execution stops and Python reports the error.

`try-except` gives us a way to handle an expected exception and allow the program to continue.

The basic idea is:

1. Put code that may cause an exception inside `try`.
2. Provide an `except` block to handle the exception.
3. If an exception occurs, Python moves to the appropriate `except` block.

````python
print("Before")

try:
    result = 10 / 0
except:
    print("Something went wrong")

print("After")
````

### What happens without handling?

If `10 / 0` is executed without an appropriate handler, Python raises `ZeroDivisionError` and the program stops at that point.

With `try-except`, we can respond to that exception instead of allowing it to terminate the current flow of the program.

## 2. Basic `try-except` Syntax

The fundamental structure is:

```python
try:
    # code that may cause an exception
except:
    # code that runs if an exception occurs
```

The `try` block contains the statements Python should attempt to execute.

If an exception occurs in that block, Python looks for a matching `except` handler.

````python
try:
    result = 10 / 0
except:
    print("Something went wrong")
````

### Basic execution flow

```text
try block
   ↓
Exception?
 ┌─┴─┐
No  Yes
↓    ↓
continue → except
```

If there is no exception, the `except` block is skipped.

If an exception occurs, execution inside the `try` block stops at that point and Python moves to the appropriate handler.

## 3. How `try` Works

Python executes statements inside a `try` block normally.

If every statement completes successfully, Python skips the `except` block.

````python
try:
    print("A")
    print("B")
except:
    print("Error")
````

Output:

```text
A
B
```

There is no exception, so `except` is not executed.

The `try` block does **not** mean that Python expects an error every time. It means that Python should monitor the enclosed statements for an exception.

## 4. How `except` Works

When an exception occurs inside `try`:

1. Python stops executing the remaining statements in that `try` block.
2. Python looks for a matching `except`.
3. The matching `except` block executes.
4. Python continues with the code after the `try-except` statement.

````python
try:
    print(10 / 0)
except:
    print("Division is not possible")
````

The `except` block runs because `10 / 0` raises an exception.

The statements after the point where the exception occurred inside the `try` block are not executed.

## 5. What Happens After an Exception?

Consider the complete flow:

````python
print("Start")

try:
    print(10 / 0)
    print("This will not execute")
except:
    print("Exception handled")

print("End")
````

Output:

```text
Start
Exception handled
End
```

Notice that `"This will not execute"` is never printed.

Once the exception occurs, Python leaves the `try` block, executes the handler, and then continues with the statement after the entire `try-except` structure.

## 6. Handling a Specific Exception

Instead of using a completely generic `except`, we can specify the exception type.

```python
except ZeroDivisionError:
```

Example:

````python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
````

A specific exception handler makes the code more precise.

It tells the reader exactly which problem the program expects to handle.

When you know the expected exception type, prefer a specific handler rather than a completely generic `except`.

## 7. `try-except` with `ValueError`

`ValueError` commonly occurs when a value has the right general type for an operation but has an inappropriate value.

For example, converting `"abc"` to an integer raises `ValueError`.

````python
try:
    age = int(input("Enter your age: "))
    print("Age:", age)
except ValueError:
    print("Please enter a valid number")
````

### If the user enters `25`

The conversion succeeds, so the `except` block is skipped.

### If the user enters `abc`

`int("abc")` raises `ValueError`, so Python executes:

```python
print("Please enter a valid number")
```

These are two different execution paths through the same program.

## 8. `try-except` with `IndexError`

`IndexError` occurs when a sequence index is outside the valid range.

````python
numbers = [10, 20, 30]

try:
    print(numbers[5])
except IndexError:
    print("Index does not exist")
````

The valid indexes are `0`, `1`, and `2`.

Index `5` does not exist, so Python raises `IndexError`, which is handled by the `except IndexError` block.

## 9. `try-except` with `KeyError`

`KeyError` occurs when a dictionary lookup requests a key that does not exist.

````python
student = {
    "name": "Komal",
    "age": 22
}

try:
    print(student["marks"])
except KeyError:
    print("Key does not exist")
````

The dictionary contains `"name"` and `"age"`, but not `"marks"`.

Therefore, the lookup raises `KeyError`.

## 10. `try-except` with `TypeError`

`TypeError` can occur when an operation is performed on incompatible types.

````python
try:
    result = 10 + "20"
except TypeError:
    print("Incompatible data types")
````

An integer and a string cannot be added directly in this way, so Python raises `TypeError`.

## 11. `try-except` with `FileNotFoundError`

This is a practical example connected to file handling.

If Python tries to open a file that does not exist at the specified location, it can raise `FileNotFoundError`.

````python
try:
    file = open("data.txt")
except FileNotFoundError:
    print("File not found")
````

Here, the exception is handled instead of allowing the missing-file operation to stop the program.

The exact result depends on whether `data.txt` exists in the current working directory.

## 12. Getting the Exception Object

An exception can be stored in a variable using `as`.

```python
except ZeroDivisionError as e:
```

Here, `e` refers to the exception object created when the exception occurs.

````python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print(type(e))
    print(e)
````

The exception object can be inspected.

For example:

```python
print(type(e))
print(e)
```

The variable name `e` is only a conventional choice. Another valid variable name could be used.

## 13. `except Exception as e`

Python provides the general `Exception` class for many ordinary runtime exceptions.

Example:

````python
try:
    result = 10 / 0
except Exception as e:
    print("Error:", e)
````

`Exception` is broad and can catch many ordinary exceptions.

However, do not use it everywhere simply because it is convenient.

When you know the specific exception that should be handled, prefer the specific type:

```python
except ZeroDivisionError:
```

Use a broad `Exception` handler only when broad handling is actually appropriate.

## 14. Bare `except` vs Specific `except`

### Bare `except`

```python
try:
    result = 10 / 0
except:
    print("Error")
```

### Specific `except`

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
```

The second version communicates exactly what the program expects and handles.

A bare `except` can catch exceptions more broadly than intended and can hide unexpected problems. For normal application code, specific exception types are generally clearer.

## 15. Multiple Statements Inside `try`

A `try` block can contain several related operations.

For example:

````python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print(result)
except ValueError:
    print("Enter a valid integer")
except ZeroDivisionError:
    print("Number cannot be zero")
````

This example contains more than one `except` handler.

The detailed topic of handling multiple exception types and ordering multiple handlers is covered in the next notebook.

For now, notice the basic idea: different operations in one `try` block can potentially produce different exceptions.

## 16. Important Rule — What Goes Inside `try`?

A `try` block should contain operations that may reasonably raise the exception you are trying to handle.

Avoid putting an entire program inside one huge `try` block.

### Too broad

```python
try:
    # 100 lines of code
except:
    print("Error")
```

This makes it difficult to determine which operation caused the problem and may hide unexpected problems.

### More focused

```python
number = int(input("Enter number: "))

try:
    result = 100 / number
except ZeroDivisionError:
    print("Cannot divide by zero")
```

This is a basic introduction to focused exception handling. More exception-handling best practices are covered later.

## 17. Nested `try-except`

A `try` block can exist inside another `try` block.

Example:

````python
try:
    try:
        result = 10 / 0
    except ZeroDivisionError:
        print("Inner exception handled")
except:
    print("Outer exception handled")
````

The inner handler gets the opportunity to handle the exception first.

If the inner handler handles the exception, execution continues after the inner `try-except`.

Nested exception handling is useful in some situations, but it should not be added unnecessarily. More advanced exception-handling patterns come later.

## 18. `try-except` with Functions

Exception handling can be placed inside a user-defined function.

This connects exception handling with the Functions section.

````python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"


print(divide(10, 2))
print(divide(10, 0))
````

The function handles `ZeroDivisionError` internally.

For `divide(10, 2)`, the calculation succeeds.

For `divide(10, 0)`, the exception is caught and the function returns the message instead.

## 19. Exception Handling with User Input

A common use of `try-except` is validating user input.

### Basic input validation

````python
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)
except ValueError:
    print("Invalid input")
````

The conversion succeeds when the user enters a valid integer.

If the user enters text that cannot be converted to an integer, `ValueError` is raised and handled.

### Combining input conversion and division

The following example can encounter two different exception types:

````python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print(result)
except ValueError:
    print("Please enter an integer")
except ZeroDivisionError:
    print("Zero is not allowed")
````

The detailed treatment of multiple `except` blocks is the subject of the next notebook. Here, focus on the overall idea: user input can produce exceptions, and `try-except` can handle them.

## 20. Flow of `try-except`

The complete basic flow can be represented as:

```text
              Start
                │
                ▼
          Execute try
                │
          ┌─────┴─────┐
          │           │
     No exception   Exception
          │           │
          ▼           ▼
      Skip except   Find matching
          │          except
          │           │
          └─────┬─────┘
                ▼
        Continue program
```

The key idea is that the `try` block is attempted first.

- **No exception:** the `except` handler is skipped.
- **Exception:** Python stops the current `try` block, finds a matching handler, executes it, and then continues after the `try-except`.

## 21. Common Mistakes

### Mistake 1 — Putting `except` without `try`

```python
except ValueError:
    print("Error")
```

An `except` block must belong to a corresponding `try` statement. Used by itself, this is invalid syntax.

### Mistake 2 — Catching the wrong exception

```python
try:
    print(10 / 0)
except ValueError:
    print("Error")
```

`10 / 0` raises `ZeroDivisionError`, not `ValueError`, so this handler does not match that exception.

### Mistake 3 — Putting unrelated code inside `try`

Keep the `try` block focused on operations for which exception handling is intended.

### Mistake 4 — Using bare `except` everywhere

A bare `except` is very broad and can hide unexpected problems.

### Mistake 5 — Forgetting that execution stops inside `try` after an exception

```python
try:
    print("A")
    10 / 0
    print("B")
except ZeroDivisionError:
    print("C")
```

Output:

```text
A
C
```

`B` is not executed because the exception occurs before it.

````python
try:
    print("A")
    10 / 0
    print("B")
except ZeroDivisionError:
    print("C")
````
