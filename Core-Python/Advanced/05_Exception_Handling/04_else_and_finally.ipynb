# 04 — `else` and `finally`

## Exception Handling

This notebook completes the main `try-except` control-flow structure by introducing **`else`** and **`finally`**.

### Central idea

```text
try     → Run risky code
except  → Run when a matching exception occurs
else    → Run when no exception occurs
finally → Run regardless of whether an exception occurs
```

### Learning goals

By the end of this notebook, you should understand:

- what the `else` block does
- when `else` executes
- why `else` is useful
- what the `finally` block does
- when `finally` executes
- why `finally` is useful for cleanup
- the complete `try-except-else-finally` flow
- valid combinations of `try`, `except`, `else`, and `finally`
- `try-finally` without `except`
- the difference between `else` and `finally`
- `finally` with functions and `return`
- why returning from `finally` is generally avoided

> **Boundary:** `raise` is not covered in detail here. It becomes the main topic in the next notebook.

## 1. Introduction

From the previous notebooks, we know how to handle an exception:

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero")
```

Sometimes we also need to:

- execute code only when **no exception occurs**
- execute code **regardless of whether an exception occurs**

Python provides:

```python
else
finally
```

## 2. Complete `try-except-else-finally` Structure

The full structure is:

```python
try:
    # risky code

except SomeException:
    # runs if matching exception occurs

else:
    # runs if no exception occurs

finally:
    # runs regardless
```

### Basic flow

```text
                  try
                   │
             ┌─────┴─────┐
             │           │
       No exception    Exception
             │           │
             ▼           ▼
           else        except
             │           │
             └─────┬─────┘
                   ▼
                finally
                   │
                   ▼
                Continue
```

## 3. What Is the `else` Block?

The `else` block executes **only when the `try` block completes without raising an exception**.

Example:

````python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Result:", result)
````

Output:

```text
Result: 5.0
```

Since no exception occurred, `else` executes.

## 4. `else` When an Exception Occurs

Now consider:

````python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Result:", result)
````

Output:

```text
Cannot divide by zero
```

The `else` block is skipped because an exception occurred.

### Mental model

```text
Exception?
│
├── Yes → except
│          ↓
│       skip else
│
└── No  → skip except
           ↓
          else
```

## 5. Why Use `else`?

A beginner might write:

```python
try:
    result = 10 / 2
    print("Result:", result)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

This works.

However, separating successful follow-up code can keep the risky portion of `try` more focused:

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Result:", result)
```

A useful mental model is:

- `try` → operation that may fail
- `except` → failure handling
- `else` → successful follow-up

## 6. `else` with User Input

Example:

````python
try:
    age = int(input("Enter your age: "))
except ValueError:
    print("Invalid age")
else:
    print("Your age is:", age)
````

If the user enters:

```text
22
```

the conversion succeeds and `else` runs.

If the user enters:

```text
abc
```

`ValueError` occurs and the `except` block runs instead.

## 7. `else` with Multiple Exceptions

The `else` block can be used after multiple `except` handlers.

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    result = a / b

except ValueError:
    print("Enter valid integers")

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Result:", result)
```

`else` executes only if the **entire `try` block** completes successfully without raising an exception.

````python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    result = a / b

except ValueError:
    print("Enter valid integers")

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Result:", result)
````

## 8. What Is the `finally` Block?

The `finally` block is executed when control leaves the `try` statement, whether an exception occurred or not.

Basic example:

````python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Execution completed")
````

`finally` is intended for actions that should happen regardless of whether the protected operation succeeds or fails.

## 9. `finally` When No Exception Occurs

Example:

````python
try:
    print(10 / 2)
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Finally executed")
````

Output:

```text
5.0
Finally executed
```

The `finally` block executes even though there was no exception.

## 10. `finally` When an Exception Occurs

Example:

````python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Finally executed")
````

Output:

```text
Cannot divide by zero
Finally executed
```

Therefore:

```text
No exception       → finally executes
Handled exception  → finally executes
```

## 11. Why Do We Need `finally`?

`finally` is useful for cleanup actions that should happen whether an operation succeeds or fails.

Examples include:

- closing files
- releasing resources
- closing database connections
- cleanup operations
- resetting temporary state

Conceptually:

```python
try:
    print("Using resource")
except Exception:
    print("Something went wrong")
finally:
    print("Clean up resource")
```

The important idea is:

> `finally` is for actions that should be performed regardless of the success or failure of the operation.

## 12. `finally` with File Handling

Since file handling was learned earlier, it is a useful example here.

````python
file = None

try:
    file = open("data.txt", "r")
    content = file.read()
    print(content)

except FileNotFoundError:
    print("File does not exist")

finally:
    if file is not None:
        file.close()
````

The file is closed in `finally` if it was successfully opened.

This illustrates why cleanup belongs in `finally`: the cleanup should be attempted whether reading succeeds or an expected exception occurs.

Later, **context managers (`with`)** provide a cleaner solution for many resource-management situations. Context managers are a separate Advanced Python topic and are not covered deeply here.

## 13. `try-except-else-finally` Together

This is the central example of the notebook:

````python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Please enter a valid integer")

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Result:", result)

finally:
    print("Program execution finished")
````

This single structure demonstrates all four parts:

- `try` → performs the risky operations
- `except` → handles expected exceptions
- `else` → runs only after successful execution
- `finally` → runs regardless of whether an exception occurred



## 17. Different Valid Combinations

All four blocks are not always required.

### `try-except`

```python
try:
    ...
except ValueError:
    ...
```

### `try-except-else`

```python
try:
    ...
except ValueError:
    ...
else:
    ...
```

### `try-except-finally`

```python
try:
    ...
except ValueError:
    ...
finally:
    ...
```

### `try-except-else-finally`

```python
try:
    ...
except ValueError:
    ...
else:
    ...
finally:
    ...
```

Python also allows:

```python
try:
    ...
finally:
    ...
```

without an `except` block.

## 18. `try-finally` Without `except`

Example:

````python
try:
    print("Running operation")
finally:
    print("Cleanup")
````

`finally` is not itself an exception handler.

Consider an unhandled exception:

````python
try:
    print(10 / 0)
finally:
    print("Cleanup")
````

`finally` executes, but because there is no `except` handler, the `ZeroDivisionError` still propagates afterward.

This distinction is important:

> `finally` guarantees the finalization code is attempted; it does not automatically handle an exception.

## 19. `else` vs `finally`

| Feature | `else` | `finally` |
|---|---|---|
| Runs when no exception occurs | Yes | Yes |
| Runs after a handled exception | No | Yes |
| Used for successful follow-up code | Yes | Usually no |
| Used for cleanup | Usually no | Yes |

### Mental model

```text
else
→ success path

finally
→ cleanup/finalization path
```

## 20. Execution Order

Consider:

````python
try:
    print("Try")
except:
    print("Except")
else:
    print("Else")
finally:
    print("Finally")
````

When no exception occurs, the output is:

```text
Try
Else
Finally
```

When an exception occurs and is handled:

```text
Try
Except
Finally
```

The key idea is:

- `else` belongs to the successful path.
- `finally` runs after the applicable `try`/`except`/`else` path.

## 21. `finally` with Functions

Exception control flow can also be used inside functions.

````python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print("Result:", result)
    finally:
        print("divide() finished")


divide(10, 2)
divide(10, 0)
````

For `divide(10, 2)`:

```text
Result: 5.0
divide() finished
```

For `divide(10, 0)`:

```text
Cannot divide by zero
divide() finished
```

The `finally` block runs in both cases.

## 22. `finally` and `return`

This behavior is important because it can surprise beginners.

Consider:

````python
def test():
    try:
        return "From try"
    finally:
        print("Finally executed")


print(test())
````

Expected output:

```text
Finally executed
From try
```

The `finally` block executes before control actually leaves the function.

The `return` from the `try` is still returned after the `finally` code finishes.

## 23. Avoid `return` Inside `finally`

Consider this pattern:

````python
def test():
    try:
        return 10
    finally:
        return 20


print(test())
````

The `return` inside `finally` overrides the earlier `return 10`, so the function returns `20`.

Returning from `finally` should generally be avoided because it can make control flow confusing and can suppress other outcomes.

A `finally` block is usually better used for cleanup/finalization rather than changing the function's result.

## 24. Practical Example — Safe Division

A complete practical example:

````python
def divide():
    try:
        a = float(input("Enter first number: "))
        b = float(input("Enter second number: "))
        result = a / b

    except ValueError:
        print("Invalid numeric input")

    except ZeroDivisionError:
        print("Cannot divide by zero")

    else:
        print("Result:", result)

    finally:
        print("Division operation completed")


divide()
````

This combines:

- user input
- `ValueError`
- `ZeroDivisionError`
- successful follow-up with `else`
- final cleanup/finalization with `finally`

## 25. Common Mistakes

### Mistake 1 — Thinking `else` always executes

`else` executes only when the `try` block completes without an exception.

### Mistake 2 — Thinking `finally` handles exceptions

It does not.

```python
finally:
```

provides finalization behavior, not exception matching.

### Mistake 3 — Using `else` for cleanup

Cleanup normally belongs in `finally`.

### Mistake 4 — Putting too much code inside `try`

Keep the risky portion focused where practical.

### Mistake 5 — Assuming `finally` only runs after `except`

`finally` also runs when no exception occurs.

### Mistake 6 — Returning from `finally`

A `return` inside `finally` can override an earlier return and complicate exception behavior.
